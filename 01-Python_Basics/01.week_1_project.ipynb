```python
import random
```

### Step 2: Create the Questions

Store the quiz questions in a list of dictionaries.

Each dictionary should contain:

- Four answer choices: A, B, C, and D.

```python
questions = [
    {
        "question": "Which keyword is used to define a function in Python?",
        "A": "function",
        "B": "def",
        "C": "define",
        "D": "fun",
        "answer": "B"
    },
    {
        "question": "Which data type stores multiple items in an ordered collection?",
        "A": "Integer",
        "B": "String",
        "C": "List",
        "D": "Boolean",
        "answer": "C"
    },
    {
        "question": "Which loop is commonly used to iterate over a sequence?",
        "A": "if",
        "B": "for",
        "C": "def",
        "D": "try",
        "answer": "B"
    }
]
```

### Step 3: Shuffle the Questions

Use `random.shuffle()` to mix the order of questions before displaying them.

```python
random.shuffle(questions)
```

This ensures that the questions do not always appear in the same order each time the quiz is played.

### Step 4: Initialise the Score

Create a variable to keep track of the user's score.

```python
score = 0
```

The score should increase every time the user selects the correct answer.

### Step 5: Display the Quiz Questions

Use a `for` loop to go through the questions and display them one at a time.

For each question, display:

- The question text.
- All four answer choices.

Example:

```python
for question in questions:

    print("\n" + question["question"])

    print("A.", question["A"])
    print("B.", question["B"])
    print("C.", question["C"])
    print("D.", question["D"])
```

### Step 6: Accept and Validate User Input

Ask the user to select an answer.

The application should only accept A, B, C, or D.

Use a `while` loop and `try/except` to handle invalid input. If the user enters an invalid choice, display an error message and ask them to try again.

Example:

```python
while True:
    try:
        answer = input("Enter your answer (A/B/C/D): ").upper().strip()

        if answer not in ["A", "B", "C", "D"]:
            raise ValueError("Please enter A, B, C, or D.")

        break

    except ValueError as error:
        print(error)
```

**What is happening here?**

- `input()` collects the user's answer.
- `upper()` converts the answer to uppercase.
- `strip()` removes unnecessary spaces.
- The `if` statement checks whether the answer is valid.
- `raise ValueError` raises an exception when the input is invalid.
- `except` catches the exception and displays the error message.
- `break` exits the loop once a valid answer is entered.

### Step 7: Check the Answer and Update the Score

Compare the user's answer with the correct answer stored in the dictionary.

If the answer is correct:
- Display a message confirming the correct answer.
- Increase the score by 1.

If the answer is incorrect:
- Display an incorrect-answer message.
- Show the correct answer.

Example:

```python
if answer == question["answer"]:
    print("Correct!")
    score += 1
else:
    print("Incorrect!")
    print("Correct answer:", question["answer"])
```

### Step 8: Calculate the Final Score

After all questions have been displayed, calculate the user's final score and percentage.

```python
total_questions = len(questions)
percentage = (score / total_questions) * 100
```

Display the results:

```python
print("\nQUIZ RESULTS")
print("Your score:", score, "/", total_questions)
print("Percentage:", round(percentage, 2), "%")
```

### Step 9: Provide Performance Feedback

Use conditional statements to give the user feedback based on their final percentage.

Apply the following grading criteria:

| Percentage | Feedback |
|---|---|
| 80% and above | Excellent |
| 50%–79% | Good |
| Below 50% | Try Again |

Example:

```python
if percentage >= 80:
    print("Excellent! You did very well.")
elif percentage >= 50:
    print("Good! Keep practising.")
else:
    print("Try Again! Review the concepts and attempt the quiz again.")
```

### Step 10: Combine Everything

Bring all the steps together to create a complete quiz application.

Your application should:

1. Display a welcome message.
2. Shuffle the questions.
3. Display questions one at a time.
4. Accept and validate user answers.
5. Check answers and update the score.
6. Display the final score and percentage.
7. Provide performance feedback.

**Reference implementation:**

```python
import random

questions = [
    {
        "question": "Which keyword is used to define a function in Python?",
        "A": "function",
        "B": "def",
        "C": "define",
        "D": "fun",
        "answer": "B"
    },
    {
        "question": "Which data type stores multiple items in an ordered collection?",
        "A": "Integer",
        "B": "String",
        "C": "List",
        "D": "Boolean",
        "answer": "C"
    },
    {
        "question": "Which loop is commonly used to iterate over a sequence?",
        "A": "if",
        "B": "for",
        "C": "def",
        "D": "try",
        "answer": "B"
    }
]

# Shuffle the questions
random.shuffle(questions)

# Initialise the score
score = 0

print("WELCOME TO THE PYTHON QUIZ!")
print("---------------------------")

# Display questions
for question in questions:

    print("\n" + question["question"])

    print("A.", question["A"])
    print("B.", question["B"])
    print("C.", question["C"])
    print("D.", question["D"])

    # Validate user input
    while True:
        try:
            answer = input("Enter your answer (A/B/C/D): ").upper().strip()

            if answer not in ["A", "B", "C", "D"]:
                raise ValueError("Please enter A, B, C, or D.")

            break

        except ValueError as error:
            print(error)

    # Check the answer
    if answer == question["answer"]:
        print("Correct!")
        score += 1
    else:
        print("Incorrect!")
        print("Correct answer:", question["answer"])

# Calculate final results
total_questions = len(questions)
percentage = (score / total_questions) * 100

# Display results
print("\nQUIZ RESULTS")
print("------------")
print("Your score:", score, "/", total_questions)
print("Percentage:", round(percentage, 2), "%")

# Provide feedback
if percentage >= 80:
    print("Excellent! You did very well.")
elif percentage >= 50:
    print("Good! Keep practising.")
else:
    print("Try Again! Review the concepts and attempt the quiz again.")
```

**Expected outcome:** A working quiz application that runs in the terminal and provides a final score and performance feedback.

Create the remaining functions based on their responsibilities.

### Step 3: Move Input Validation into a Function

Move the input validation logic from Stage 1 into `get_answer()`.

Remember to retain the `try/except` handling so that invalid choices do not interrupt the application.

```python
def get_answer():

    while True:
        try:
            answer = input("Enter your answer (A/B/C/D): ").upper().strip()

            if answer not in ["A", "B", "C", "D"]:
                raise ValueError("Please enter A, B, C, or D.")

            return answer

        except ValueError as error:
            print(error)
```

```python
def display_results(score, total_questions):

    percentage = (score / total_questions) * 100

    print("\nQUIZ RESULTS")
    print("------------")
    print("Your score:", score, "/", total_questions)
    print("Percentage:", round(percentage, 2), "%")

    if percentage >= 80:
        print("Excellent! You did very well.")
    elif percentage >= 50:
        print("Good! Keep practising.")
    else:
        print("Try Again! Review the concepts and attempt the quiz again.")
```

# Stage 3: Store Questions in a CSV File

## Objective

Modify your application to load questions from an external CSV file instead of storing them directly in the Python script.

CSV files allow us to store structured data in a tabular format and make it easier to manage larger sets of questions.

### Step 1: Create the CSV File

Create a file named `questions.csv`.

Use the following column headings:

| question | A | B | C | D | answer |
|---|---|---|---|---|---|
| Which keyword defines a function in Python? | function | def | define | fun | B |
| Which data type stores multiple items in an ordered collection? | Integer | String | List | Boolean | C |
| Which loop is commonly used to iterate over a sequence? | if | for | def | try | B |

Add at least 10 questions to your CSV file.

Make sure the answer column contains the correct option letter: A, B, C, or D.

### Step 2: Import the CSV Module

At the beginning of your Python file, import the `csv` module.

```python
import csv
import random
```

The `csv` module allows Python to read from and write to CSV files.

### Step 3: Create a Function to Load Questions

Create a function called `load_questions()`.

Use `csv.DictReader()` to read the CSV file.

`DictReader()` reads each row as a dictionary, using the column headings as keys.

```python
def load_questions():

    questions = []

    try:
        with open("questions.csv", "r", newline="", encoding="utf-8") as file:

            reader = csv.DictReader(file)

            for row in reader:
                row["answer"] = row["answer"].strip().upper()
                questions.append(row)

    except FileNotFoundError:
        print("Error: The questions.csv file was not found.")

    except PermissionError:
        print("Error: You do not have permission to read the file.")

    except OSError:
        print("Error: Unable to access the file.")

    return questions
```

### Step 4: Integrate CSV Loading into Your Application

Your application should also check whether any questions were successfully loaded before starting the quiz.

For example:

```python
questions = load_questions()

if not questions:
    print("No questions available. Please check your CSV file.")
```
