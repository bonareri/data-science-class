## Applying Web Services

### Introduction to APIs?
An **API (Application Programming Interface)** is a set of rules that allows different software applications to communicate with each other.

Think of it like a **waiter** at a restaurant:
- You (the client) give the waiter (API) your order.
- The waiter takes it to the kitchen (server).
- The kitchen prepares your food and gives it back to the waiter.
- The waiter delivers it to you.

### Examples where APIs are used?
<ol>
 	<li><strong>Data Integration</strong>: APIs allow different databases or software services to share data. For instance, a weather application might use an API to gather weather data from a central database.</li>
 	<li><strong>E-commerce</strong>: Online stores use APIs to connect with payment processors, ensuring smooth transactions.</li>
 	<li><strong>Social Media Integration</strong>: Many websites and apps use social media APIs to allow users to sign in with their social media accounts, share content, or import their profile data.</li>
 	<li><strong>IoT Devices</strong>: In the Internet of Things (IoT), smart thermostats and fitness trackers use APIs to communicate data to servers or other devices.</li>
 	<li><strong>Cloud Computing</strong>: APIs are crucial in cloud services, allowing applications to access cloud resources and perform tasks like storing data or running computations.</li>
</ol>

---

In Python, working with APIs usually involves sending HTTP requests to web services and processing the responses. Python provides several libraries to facilitate API interactions, such as <code>requests</code>, <code>http.client</code>, and <code>urllib</code>.

### API Terms to Know

| Term               | Description                                                                 |
|--------------------|-----------------------------------------------------------------------------|
| **Endpoint**        | URLs to which you send requests (e.g., `https://api.example.com/users`).
| **Request**         | The action sent to the API (e.g., GET, POST)                                |
| **Response**        | The data returned by the API (usually in JSON or XML format)                       |
| **Status Code**     | Tells if the request was successful (e.g., `200`, `404`)                    |
| **Authentication**  | Many APIs require an API Key or token to access                             |
| **HTTP Methods**    | Common methods include GET, POST, PUT, DELETE                               |
| **Headers**         | Metadata sent with the request (e.g., authentication tokens, content type)  |
| **Parameters**      | Data sent with the request (query params in URL or in the request body)     |

---

### Common HTTP Methods

| Method | Description          |
|--------|----------------------|
| GET    | Retrieve data        |
| POST   | Send data            |
| PUT    | Update existing data |
| DELETE | Remove data          |

<h2>Understanding Common API Status Codes</h2>
<p class="rm">Every request to a web server returns a status code indicating what happened with the request. Here are some common codes relevant to GET requests:</p>

<ul>
 	<li><code>200</code>: Everything went okay, and the result has been returned (if any).</li>
 	<li><code>301</code>: The server is redirecting you to a different endpoint. This can happen when a company switches domain names, or an endpoint name is changed.</li>
 	<li><code>400</code>: The server thinks you made a bad request. This happens when you send incorrect data or make other client-side errors.</li>
 	<li><code>401</code>: The server thinks you're not authenticated. Many APIs require login credentials, so this happens when you don't send the right credentials to access an API.</li>
 	<li><code>403</code>: The resource you're trying to access is forbidden: you don't have the right permissions to see it.</li>
 	<li><code>404</code>: The resource you tried to access wasn't found on the server.</li>
 	<li><code>503</code>: The server is not ready to handle the request.</li>
</ul>
<p class="rm">Notice that all the codes starting with 4 indicate some sort of client-side error, while 5xx codes point to server-side issues. The first digit of the status code indicates its category. Knowing this makes it easy to quickly identify if a request was successful (2xx) or if there was an error (4xx or 5xx).</p>

<h4>Working with APIs in Python</h4>
The <code>requests</code> library is the most popular way to interact with APIs in Python. Here's a basic example:

```python
import requests

# Send a GET request to an API endpoint

# Check if the request was successful
```

## 1. Introduction to Object-Oriented Programming

So far, we have been writing Python programs using variables, conditions, loops, and functions. As programs become larger, managing data and functionality separately can become difficult.

**Object-Oriented Programming (OOP)** is a programming approach that organizes code around objects. Objects combine data (attributes) and behavior (methods) into a single unit.

For example, consider a student management system. A student has information such as:

- Name
- Student ID
- Course

A student can also perform actions such as:

- Display their details
- Enroll in a course
- Submit an assignment

OOP allows us to represent real-world entities in our programs.

### Why use OOP?

- Organizes code into reusable structures.
- Makes it easier to manage large programs.
- Reduces repetition when creating similar entities.
- Makes code easier to maintain and extend.

---

## 2. Classes and Objects

Classes and objects are the fundamental concepts of Object-Oriented Programming.

### What is a class?

A **class** is a blueprint or template used to create objects. It defines the attributes and behaviors that objects created from it can have.

### What is an object?

An **object** is an instance of a class. It is an actual entity created using that blueprint.

Think of a class as a blueprint for a house, and objects as the actual houses built from that blueprint. You can build many houses from one blueprint, but each house can have different properties.

### Creating a class

In Python, we use the `class` keyword to define a class.

```python
class Student:
    pass
```

- `class` is the keyword used to define a class.
- `Student` is the class name.
- `pass` is a placeholder used when a class has no implementation yet.

Python convention is to use **PascalCase** for class names, for example, `Student`, `BankAccount`, and `Employee`.

### Creating objects

We can create objects from the `Student` class.

```python
class Student:
    pass

student1 = Student()
student2 = Student()
```

Here:

- `Student` is the class.
- `student1` and `student2` are objects.
- `Student()` creates a new instance of the class.

We can create as many objects as we need from the same class.

```python
print(student1)
print(student2)
```

Each object is a separate instance of the class.

---

## 3. Attributes

**Attributes** are variables that store information associated with an object.

For example, each student has a name, age, and course.

We can assign attributes to objects after creating them.

```python
class Student:
    pass

student1 = Student()

student1.name = "Mary"
student1.age = 22
student1.course = "Data Science"

print(student1.name)
print(student1.age)
print(student1.course)
```

However, assigning attributes individually after creating objects can become tedious. We need a better way to initialize objects, which brings us to the `__init__()` method.

---

## 4. The `__init__()` Method

The `__init__()` method is a special method that Python automatically calls when a new object is created.

It is commonly used to initialize an object's attributes.

```python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

student1 = Student("Mary", 22, "Data Science")
student2 = Student("John", 25, "Software Engineering")

print(student1.name)
print(student2.course)
```

### Understanding the code

- `__init__()` initializes an object when it is created.
- `name`, `age`, and `course` are parameters receiving the values passed in.
- `self.name`, `self.age`, and `self.course` are attributes belonging to the particular object.
- `Student("Mary", 22, "Data Science")` creates and initializes a new student object.

### What is `self`?

`self` refers to the current object being created or used.

It allows each object to store and access its own data.

```python
self.name = name
```

- `self.name` is the object's attribute.
- `name` is the parameter passed to the constructor.

When we create different objects, Python stores different values in each object.

```python
student1 = Student("Mary", 22, "Data Science")
student2 = Student("John", 25, "Software Engineering")

print(student1.name)
print(student2.name)
```

Output:

```text
Mary
John
```

**Note:** `self` is not a Python keyword, but it is the standard naming convention. Always use it as the first parameter in instance methods.

## 5. Methods

**Methods** are functions defined inside a class. They describe what an object can do.

For example, a student object can display its details or calculate their average marks.

```python
class Student:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    def introduce(self):
        print(f"My name is {self.name}, and I study {self.course}.")

student1 = Student("Mary", "Data Science")

student1.introduce()
```

Notice that we call the method using the object followed by the method name.

### Adding more methods

```python
class Student:
    def __init__(self, name, course, marks):
        self.name = name
        self.course = course
        self.marks = marks

    def introduce(self):
        print(f"I am {self.name}, studying {self.course}.")

    def calculate_average(self):
        return sum(self.marks) / len(self.marks)

student1 = Student("Mary", "Data Science", [78, 85, 90])

student1.introduce()

average = student1.calculate_average()
print("Average:", average)
```

### Attributes vs Methods

| Attributes | Methods |
|---|---|
| Store data about an object | Define what an object does |
| Variables associated with an object | Functions defined inside a class |
| `student.name` | `student.introduce()` |
| `student.marks` | `student.calculate_average()` |

Each bank account has:

- Account holder
- Balance

It can:

- Deposit money
- Withdraw money

```python
class BankAccount:
```

## 7. Instance Attributes vs Class Attributes

So far, our attributes have belonged to individual objects. These are called **instance attributes**.

Python also allows us to define **class attributes**, which are shared at the class level.

```python
class Student:
    school = "Zindua School"  # Class attribute

    def __init__(self, name, course):
        self.name = name       # Instance attribute
        self.course = course   # Instance attribute

student1 = Student("Mary", "Data Science")
student2 = Student("John", "Software Engineering")

print(student1.school)
print(student2.school)
```

Output:

```text
Zindua School
Zindua School
```

| Class Attributes | Instance Attributes |
|---|---|
| Defined directly inside the class | Usually defined inside `__init__()` using `self` |
| Shared at the class level | Belong to individual objects |
| `school` | `name`, `course` |

## 8. The Four Pillars of OOP

The four major principles of Object-Oriented Programming are encapsulation, inheritance, polymorphism, and abstraction.

### 8.1 Inheritance

**Inheritance** allows a child class to acquire attributes and methods from a parent class. It promotes code reuse and allows us to extend existing functionality.

For example, different types of employees may share common attributes such as name and salary, but have different roles.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_details(self):
        print(self.name, self.salary)


class DataAnalyst(Employee):
    def analyze_data(self):
        print(f"{self.name} is analyzing data.")


class SoftwareDeveloper(Employee):
    def write_code(self):
        print(f"{self.name} is writing code.")


analyst = DataAnalyst("Mary", 60000)
developer = SoftwareDeveloper("John", 80000)

analyst.display_details()
analyst.analyze_data()

developer.display_details()
developer.write_code()
```

Notice that `DataAnalyst` and `SoftwareDeveloper` don't have their own `__init__()` methods, but they can still use the one inherited from `Employee`.

- `Employee` is the parent class (superclass).
- `DataAnalyst` and `SoftwareDeveloper` are child classes (subclasses).

### 8.2 Polymorphism

**Polymorphism** allows different classes to implement the same method name, each with its own behavior.

```python
class Dog:
    def speak(self):
        print("Woof!")

class Cat:
    def speak(self):
        print("Meow!")

class Cow:
    def speak(self):
        print("Moo!")

animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.speak()
```

Output:

```text
Woof!
Meow!
Moo!
```

All three objects have the same method `speak()`, but each produces a different result.

### 8.3 Encapsulation

**Encapsulation** combines data and the methods that operate on that data within a class. It also controls how attributes are accessed or modified.

Python uses naming conventions to indicate the visibility of attributes:

| Naming | Access Level | Meaning |
|---|---|---|
| `name` | Public | Can be accessed and modified from anywhere. |
| `_name` | Protected | Intended for use within the class and its subclasses. |
| `__name` | Private | Intended for use within the class. Python uses name mangling to make direct access more difficult. |

**Note:** Python does not strictly enforce public, protected, and private access. These naming conventions guide developers on how attributes should be used.

#### Example

```python
class BankAccount:
    def __init__(self, account_holder, balance, pin):
        self.account_holder = account_holder  # Public
        self._balance = balance               # Protected
        self.__pin = pin                      # Private

    def display_account(self):
        print(f"Account Holder: {self.account_holder}")
        print(f"Balance: {self._balance}")

    def verify_pin(self, pin):
        return self.__pin == pin


account = BankAccount("Melody", 5000, 1234)

# Accessing public attribute
print(account.account_holder)

# Accessing protected attribute
print(account._balance)

# Accessing private attribute through a class method
print(account.verify_pin(1234))
```

**Output:**

```text
Melody
5000
True
```

#### Understanding the Three Naming Conventions

**1. Public Attributes**

Public attributes can be accessed or modified from anywhere, inside or outside the class.

```python
print(account.account_holder)

account.account_holder = "Jane"
print(account.account_holder)
```

**2. Protected Attributes**

Protected attributes use a single underscore (`_`). They can still be accessed from outside the class, but the underscore indicates that they are intended for internal use or use by subclasses.

```python
print(account._balance)
```

Although this works, developers are discouraged from accessing protected attributes directly unless necessary.

**3. Private Attributes**

Private attributes use double underscores (`__`). Python changes their names internally through a process called name mangling, making accidental access or modification more difficult.

```python
print(account.verify_pin(1234))
```

The PIN is accessed through a method defined inside the class rather than being accessed directly.

Attempting to access it directly:

```python
print(account.__pin)
```

Raises an `AttributeError` because Python changes the attribute name internally.

The double underscore discourages direct access to the attribute. Python does not enforce strict private access in the same way as some other languages.

## 8.4 Abstraction

### 1. Introduction to Abstraction

**Abstraction** is the concept of hiding unnecessary implementation details and exposing only the essential features of an object.

It allows users to interact with an object without needing to understand how everything works internally.

For example, when withdrawing money from an ATM, you only need to enter your PIN, select the withdrawal option, and enter the amount. You don't need to understand how the ATM communicates with the bank or processes the transaction internally.

In Python, abstraction helps us simplify complex systems by exposing only the functionality that a user needs.

### 2. Why Do We Use Abstraction?

Abstraction helps us:

- Reduce complexity by hiding unnecessary implementation details.
- Make code easier to use and understand.
- Separate what an object does from how it does it.
- Improve code organization and maintainability.
- Allow different classes to follow a common structure while implementing their functionality differently.

### 3. Practical Example 1: ATM Machine

Consider an ATM machine. A user only needs to call the `withdraw()` method without understanding how the transaction is processed internally.

```python
class ATM:
    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):
        if amount <= self.balance:
            self._verify_transaction()
            self.balance -= amount
            print(f"Withdrawn: Ksh {amount}")
            print(f"Remaining balance: Ksh {self.balance}")
        else:
            print("Insufficient funds.")

    def _verify_transaction(self):
        print("Verifying transaction...")


atm = ATM(10000)

atm.withdraw(3000)
```

**Output:**

```text
Verifying transaction...
Withdrawn: Ksh 3000
Remaining balance: Ksh 7000
```

**Understanding the Example:**

- `withdraw()` is the main method exposed to the user.
- `_verify_transaction()` represents an internal operation.
- The user only needs to call `withdraw()` without handling each internal step.

The user interacts with a simple interface while the class handles the internal operations.

### 4. Practical Example 2: Payment System Using Abstract Classes

Imagine an online shopping application that supports different payment methods, such as M-Pesa and credit cards.

Although the payment methods work differently, they all need to perform a payment.

We can use an abstract class to define a common structure for all payment methods.

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class MpesaPayment(Payment):
    def pay(self, amount):
        print(f"Processing M-Pesa payment of Ksh {amount}")
        print("Payment successful!")


class CardPayment(Payment):
    def pay(self, amount):
        print(f"Processing card payment of Ksh {amount}")
        print("Payment successful!")


payment1 = MpesaPayment()
payment2 = CardPayment()

payment1.pay(1500)
payment2.pay(2000)
```

**Output:**

```text
Processing M-Pesa payment of Ksh 1500
Payment successful!
Processing card payment of Ksh 2000
Payment successful!
```

**Understanding the Example:**

- `Payment` is an abstract class.
- `pay()` is an abstract method.
- `MpesaPayment` and `CardPayment` inherit from `Payment`.
- Each subclass provides its own implementation of `pay()`.

The abstract class defines what every payment method must do, while the subclasses determine how the payment is processed.

### 5. Practical Example 3: Vehicle Management System

Different vehicles can start their engines differently. For example, a car uses an ignition system, while an electric vehicle uses an electric motor.

However, all vehicles should have a `start()` method.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):
    def start(self):
        print("Car engine started.")


class ElectricCar(Vehicle):
    def start(self):
        print("Electric motor started.")


class Motorcycle(Vehicle):
    def start(self):
        print("Motorcycle engine started.")


vehicles = [Car(), ElectricCar(), Motorcycle()]

for vehicle in vehicles:
    vehicle.start()
```

**Output:**

```text
Car engine started.
Electric motor started.
Motorcycle engine started.
```

**Understanding the Example:**

- `Vehicle` defines a common structure for all vehicles.
- Every vehicle must implement the `start()` method.
- Each vehicle provides its own implementation.
- The user can call `start()` without knowing the internal process of starting each vehicle.

This makes it easier to add new vehicle types without changing the overall structure.

### 6. Practical Example 4: Employee Management System

Consider a company that has different types of employees, such as full-time and contract employees.

Both types of employees receive salaries, but their salary calculations may differ.

We can use abstraction to define a common salary calculation method.

```python
from abc import ABC, abstractmethod

class Employee(ABC):

    def __init__(self, name):
        self.name = name

    @abstractmethod
    def calculate_salary(self):
        pass


class FullTimeEmployee(Employee):
    def __init__(self, name, monthly_salary):
        super().__init__(name)
        self.monthly_salary = monthly_salary

    def calculate_salary(self):
        return self.monthly_salary


class ContractEmployee(Employee):
    def __init__(self, name, hourly_rate, hours_worked):
        super().__init__(name)
        self.hourly_rate = hourly_rate
        self.hours_worked = hours_worked

    def calculate_salary(self):
        return self.hourly_rate * self.hours_worked


employee1 = FullTimeEmployee("Jane", 60000)
employee2 = ContractEmployee("John", 500, 80)

print(f"{employee1.name}: Ksh {employee1.calculate_salary()}")
print(f"{employee2.name}: Ksh {employee2.calculate_salary()}")
```

**Output:**

```text
Jane: Ksh 60000
John: Ksh 40000
```

**Understanding the Example:**

- `Employee` is an abstract class.
- `calculate_salary()` is an abstract method.
- `FullTimeEmployee` calculates salary using a fixed monthly amount.
- `ContractEmployee` calculates salary using the hourly rate and hours worked.
- Both employee types follow the same structure but implement salary calculation differently.

### 7. Important Concepts in Abstraction

| Concept | Meaning |
|---|---|
| Abstract Class | A class that provides a common structure for other classes. |
| Abstract Method | A method declared in an abstract class that must be implemented by subclasses. |
| `ABC` | A base class used to create abstract classes in Python. |
| `@abstractmethod` | A decorator used to declare an abstract method. |

**Important:** An abstract class cannot be instantiated directly if it contains abstract methods that have not been implemented.

For example:

```python
payment = Payment()
```

This raises a `TypeError` because `Payment` has an abstract method that has not been implemented.

# Special Methods in Python

## 1. Introduction to Special Methods

**Special methods** are built-in methods in Python that allow objects to interact with Python's built-in operations such as printing, addition, comparison, and length checking.

They are also called **magic methods** because they have double underscores at the beginning and end of their names (e.g., `__init__`, `__str__`).

Special methods allow us to define how objects behave when we use certain Python operations.

For example:

- `print(object)` calls `__str__()`.
- `object1 + object2` calls `__add__()`.
- `len(object)` calls `__len__()`.
- `object1 == object2` calls `__eq__()`.

We don't usually call these methods directly. Python calls them automatically when we use the corresponding operation.

---

## 2. The `__init__()` Method

The `__init__()` method is a special method that initializes an object when it is created.

It is used to assign initial values to the object's attributes.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Melody", 25)

print(student1.name)
print(student1.age)
```

**Output:**

```text
Melody
25
```

When we create `student1`, Python automatically calls `__init__()` and assigns the values to the object's attributes.

---

## 3. The `__str__()` Method

The `__str__()` method defines how an object should be represented as a readable string when we print it.

Without `__str__()`, printing an object usually displays a generic representation of the object instead of meaningful information.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Student: {self.name}, Age: {self.age}"

student1 = Student("Melody", 25)

print(student1)
```

**Output:**

```text
Student: Melody, Age: 25
```

When `print(student1)` is executed, Python automatically calls `student1.__str__()`.

> The `__str__()` method should return a string.

---

## 4. The `__len__()` Method

The `__len__()` method defines what should be returned when we use the `len()` function on an object.

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __len__(self):
        return self.pages

book1 = Book("Python Basics", 250)

print(len(book1))
```

**Output:**

```text
250
```

Python automatically calls the `__len__()` method when we use `len(book1)`.

---

## 5. The `__add__()` Method

The `__add__()` method defines how two objects behave when we use the `+` operator.

For example, we can define how two bank account objects combine their balances.

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def __add__(self, other):
        return self.balance + other.balance

account1 = BankAccount(5000)
account2 = BankAccount(3000)

print(account1 + account2)
```

**Output:**

```text
8000
```

Python automatically calls `account1.__add__(account2)` when we use `account1 + account2`.

---

## 6. The `__eq__()` Method

The `__eq__()` method defines how two objects are compared using the `==` operator.

By default, comparing two objects using `==` may check whether they are considered the same object. We can customize this behavior to compare their attributes instead.

```python
class Student:
    def __init__(self, name, admission_number):
        self.name = name
        self.admission_number = admission_number

    def __eq__(self, other):
        return self.admission_number == other.admission_number

student1 = Student("Jane", 101)
student2 = Student("Jane", 101)
student3 = Student("John", 102)

print(student1 == student2)
print(student1 == student3)
```

**Output:**

```text
True
False
```

Python calls the `__eq__()` method when we compare the objects using `==`.

---

## 7. Common Special Methods

| Special Method | Purpose | Example |
|---|---|---|
| `__init__()` | Initializes an object | `Student("Jane", 20)` |
| `__str__()` | Returns a readable string representation | `print(student)` |
| `__len__()` | Defines the length of an object | `len(book)` |
| `__add__()` | Defines behavior of `+` | `account1 + account2` |
| `__eq__()` | Defines behavior of `==` | `student1 == student2` |
| `__gt__()` | Defines behavior of `>` | `student1 > student2` |
| `__lt__()` | Defines behavior of `<` | `student1 < student2` |
| `__repr__()` | Returns a developer-oriented representation of an object | `repr(student)` |

---

## 8. Why Do We Use Special Methods?

Special methods allow us to customize how our objects behave when interacting with Python operations.

Instead of writing separate methods like:

```python
account1.add_account(account2)
student1.compare(student2)
book.get_length()
```

We can make our objects behave more naturally:

```python
account1 + account2
student1 == student2
len(book)
print(student1)
```

This makes our classes easier to use, more readable, and more compatible with Python's built-in functions and operators.

---

## 9. Practical Example: Book Class

Let's create a `Book` class that uses multiple special methods.

```python
class Book:
    def __init__(self, title, price, pages):
        self.title = title
        self.price = price
        self.pages = pages

    def __str__(self):
        return f"{self.title} - Ksh {self.price}"

    def __len__(self):
        return self.pages

    def __eq__(self, other):
        return self.title == other.title

book1 = Book("Python Basics", 1500, 250)
book2 = Book("Python Basics", 1800, 300)
book3 = Book("Data Science", 2000, 400)

print(book1)
print(len(book1))
print(book1 == book2)
print(book1 == book3)
```

**Output:**

```text
Python Basics - Ksh 1500
250
True
False
```

Notice that `book1` and `book2` are considered equal because they have the same title, even though their prices and page counts are different. This is because we defined our own comparison logic in `__eq__()`.

## 10. Key Takeaways

- Special methods have double underscores before and after their names.
- They allow us to customize how objects behave with Python operators and built-in functions.
- Python automatically calls them when we perform certain operations.
- `__init__()` initializes an object.
- `__str__()` controls how an object is displayed when printed.
- `__len__()` allows us to use `len()` on our objects.
- `__add__()` allows us to customize addition.
- `__eq__()` allows us to customize object equality.

**Remember:** Special methods help us make our own objects behave like built-in Python objects.

```python
class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def display_info(self):
        print(f"{self.name}: KSh {self.price}, Stock: {self.quantity}")

    def restock(self, amount):
        self.quantity += amount
        print(f"{self.name} restocked. New stock: {self.quantity}")

    def sell(self, amount):
        if amount <= self.quantity:
            self.quantity -= amount
            print(f"Sold {amount} {self.name}(s).")
        else:
            print("Insufficient stock.")

    def total_stock_value(self):
        return self.price * self.quantity


product1 = Product("Laptop", 80000, 5)
product2 = Product("Mouse", 1500, 10)

product1.display_info()
product1.sell(2)
product1.restock(3)

product2.display_info()
product2.sell(12)

print("Laptop stock value:", product1.total_stock_value())
print("Mouse stock value:", product2.total_stock_value())
```
