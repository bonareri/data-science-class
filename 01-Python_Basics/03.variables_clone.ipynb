# PART 2: Values, Variables and Data Types (20 minutes)

## 2.1 Understanding Values

A **value** is a piece of information that a program can work with.

Every value in Python has a data type. A data type determines the kind of information represented and the operations that can be performed on it.

### Basic Python Data Types

| Data type | Description | Example |
|---|---|---|
| Integer (`int`) | Whole numbers | `25` |
| Float (`float`) | Decimal numbers | `89.5` |
| String (`str`) | Text | `"Brian"` |
| Boolean (`bool`) | True or False values | `True` |

# PART 3: Variable Naming and Type Casting (15 minutes)

## 3.1 Variable Naming Rules

Variable names help us identify and understand the information stored in a program.

Python has specific naming rules:

- Names must start with a letter or underscore.
- Names cannot start with a number.
- Spaces are not allowed.
- Names can contain letters, numbers and underscores.
- Variable names are case-sensitive.
- Python reserved keywords cannot be used as variable names.

Python conventionally uses `snake_case` for variable names.

Example:

```python
student_name = "Brian"
student_age = 24
assessment_score = 90
```

## Activity 1: Debugging Challenge

Present this code:

```python
student name = "Brian"
2nd_score = 90
student_age = 24
class = "Data Science"
```

Ask students to identify the invalid variable names and explain why they are invalid.

**Reference solution:**

```python
student_name = "Brian"
second_score = 90
student_age = 24
class_name = "Data Science"
```

Discuss why `class` cannot be used as a variable name: it is a reserved Python keyword.

## 3.2 Type Casting

**Type casting** is the process of converting a value from one data type to another.

Sometimes information is stored in one data type but needs to be used in another.

For example:

```python
age = "25"

print(age + 5)
```

Ask students to predict what will happen.

When executed, Python raises a `TypeError` because it cannot directly add a string and an integer.

We can convert the string into an integer:

```python
age = "25"

age = int(age)

print(age + 5)
```

Output:

```text
30
```

### Common Type Conversion Functions

| Function | Purpose | Example |
|---|---|---|
| `int()` | Converts to an integer | `int("25")` |
| `float()` | Converts to a float | `float("3.5")` |
| `str()` | Converts to a string | `str(100)` |
| `bool()` | Converts to a Boolean | `bool(1)` |

### Activity 2: Type Conversion

Ask students to convert:

1. `"45"` into an integer.
2. `"78.5"` into a float.
3. `100` into a string.
4. `1` into a Boolean.

**Reference solution:**

```python
a = int("45")
b = float("78.5")
c = str(100)
d = bool(1)

print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
```

Expected values: `45`, `78.5`, `"100"` and `True`.

**Important discussion:** What happens if we attempt `int("hello")`?

Explain that not every value can be converted successfully into every data type.

# PART 4: Python Operators (35 minutes)

Operators allow us to perform calculations, compare values and make decisions.

**Scenario:**

A school wants to calculate student performance using three assessments. We need to calculate total marks, average marks and determine whether a student has met the required score.

## 4.1 Arithmetic Operators

Arithmetic operators perform mathematical calculations.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 5` |
| `-` | Subtraction | `10 - 5` |
| `*` | Multiplication | `10 * 5` |
| `/` | Division | `10 / 5` |
| `//` | Floor division | `10 // 3` |
| `%` | Modulus | `10 % 3` |
| `**` | Exponentiation | `2 ** 3` |

Example:

```python
assignment = 75
practical = 80
final_exam = 65

total = assignment + practical + final_exam

print(total)
```

### Activity 1: Explore Arithmetic Operators

Given:

```python
x = 10
y = 3
```

Ask students to predict the output of:

```python
print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x // y)
print(x % y)
print(x ** y)
```

**Reference answers:**

| Expression | Output |
|---|---:|
| `x + y` | 13 |
| `x - y` | 7 |
| `x * y` | 30 |
| `x / y` | 3.333... |
| `x // y` | 3 |
| `x % y` | 1 |
| `x ** y` | 1000 |

**Explain the important differences:**

- `/` performs division and returns a floating-point result.
- `//` performs floor division, returning the floor of the quotient.
- `%` returns the remainder.
- `**` raises a number to a power.

### Activity 2: Apply Arithmetic to a Real Problem

A student has three assessment marks:

```python
assignment = 75
practical = 80
final_exam = 65
```

Ask students to calculate:
1. Total marks.
2. Average marks.
3. Difference between the highest and lowest marks (without using built-in functions).

**Reference solution:**

```python
total = assignment + practical + final_exam
average = total / 3

highest = max(assignment, practical, final_exam)
lowest = min(assignment, practical, final_exam)

difference = highest - lowest

print("Total:", total)
print("Average:", average)
print("Difference:", difference)
```

## 4.2 Comparison Operators

Comparison operators compare two values and return a Boolean result (`True` or `False`).

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

Example:

```python
average = 73

print(average >= 70)
```

Output:

```text
True
```

### Activity 3: Predict the Results

```python
score = 65

print(score == 65)
print(score != 65)
print(score > 50)
print(score < 50)
print(score >= 70)
print(score <= 65)
```

**Reference answers:**

```text
True
False
True
False
False
True
```

Ask students to explain why each expression returns its result.

**Important distinction:** `=` assigns a value, while `==` compares two values.

## 4.3 Logical Operators

Logical operators combine multiple conditions.

| Operator | Meaning |
|---|---|
| `and` | Both conditions must be True |
| `or` | At least one condition must be True |
| `not` | Reverses a Boolean result |

Example:

A student must score at least 50 in both the assignment and final examination.

```python
assignment_score = 75
final_exam_score = 65

assignment_passed = assignment_score >= 50
exam_passed = final_exam_score >= 50

overall_pass = assignment_passed and exam_passed

print(overall_pass)
```

### Activity 4: Experiment with Logical Operators

Ask students to predict what happens when:
- Both marks are above 50.
- Only the assignment mark is above 50.
- Only the examination mark is above 50.
- Neither mark is above 50.

**Reference solution:**

```python
assignment_score = 75
final_exam_score = 40

if assignment_score >= 50 and final_exam_score >= 50:
    print("Passed")
else:
    print("Failed")
```

Output: `Failed`.

**Extension:** Ask students to write a condition that allows a student to pass if either assessment score is at least 50, using `or`.

## 4.4 Assignment Operators

Assignment operators update the value of an existing variable.

| Operator | Equivalent expression |
|---|---|
| `+=` | `x = x + value` |
| `-=` | `x = x - value` |
| `*=` | `x = x * value` |
| `/=` | `x = x / value` |
| `//=` | `x = x // value` |
| `%=` | `x = x % value` |
| `**=` | `x = x ** value` |

Example:

```python
total = 100

total += 50
print(total)

total -= 20
print(total)

total *= 2
print(total)
```

**Reference output:**

```text
150
130
260
```

### Activity 5: Trace the Variable

```python
balance = 1000

balance -= 200
balance += 500
balance *= 2

print(balance)
```

**Reference answer:** `2600`.

Ask students to explain how the balance changes after each operation.

# PART 5: Expressions and Operator Precedence (15 minutes)

## 5.1 Understanding Expressions

An **expression** is a combination of values, variables, operators or function calls that Python evaluates to produce a result.

Examples:

```python
2 + 3
```

```python
total = 10 + 5 * 2
```

```python
name = "Hello " + "Brian"
```

```python
age = 25
is_adult = age >= 18
```

Ask students to identify what each expression produces.

## 5.2 Operator Precedence

Operator precedence determines the order in which Python evaluates operations in an expression.

For example:

```python
result = 2 + 3 * 4

print(result)
```

Ask: Is the answer 20 or 14?

The correct answer is 14 because multiplication is evaluated before addition.

Parentheses can change the order:

```python
result = (2 + 3) * 4

print(result)
```

Output:

```text
20
```

### Activity: Predict Before Running

Ask students to predict:

```python
print(10 + 5 * 2)
print((10 + 5) * 2)
print(20 / 5 + 2)
print(2 ** 3 + 1)
```

**Reference answers:**

```text
20
30
6.0
9
```

**Key concept:** Parentheses are useful for controlling evaluation order and making expressions easier to understand.

# PART 7: Main Practical Project (25 minutes)

## Mini Project: Personal Expense Calculator

**Scenario:**

You have been asked to develop a simple Python application that helps someone track their daily expenses.

The application should collect spending information and determine whether the user has stayed within a daily budget.

### Project Requirements

Your program should:

1. Ask for the user's name.
2. Ask how much they spend on transport.
3. Ask how much they spend on food.
4. Ask how much they spend on other expenses.
5. Calculate total expenditure.
6. Calculate the remaining amount from a KSh 1,000 budget.
7. Determine whether the user stayed within budget.
8. Display an expense summary.

**Additional challenge:** Calculate how much the user would spend over five working days if their daily expenditure remained the same.

### Starter Code

```python
# 1. Collect the user's name

# 2. Collect daily expenses

# 3. Calculate total expenditure

# 4. Calculate the remaining budget

# 5. Check whether the user stayed within budget

# 6. Calculate five-day expenditure

# 7. Display the expense summary
```

### Before Coding: Ask Students

- What variables will you need?
- Which data types should they contain?
- How will you collect the information?
- Which arithmetic operators will you use?
- How will you determine whether someone exceeded the budget?
- How will you display the results?

Allow students to work in pairs and develop their own solutions.

### Instructor Reference Solution

```python
name = input("Enter your name: ")

transport = float(input("Enter transport expenses: "))
food = float(input("Enter food expenses: "))
other = float(input("Enter other expenses: "))

total_expenses = transport + food + other

budget = 1000

remaining = budget - total_expenses

within_budget = total_expenses <= budget

five_day_expenses = total_expenses * 5

print("\nEXPENSE SUMMARY")
print("----------------------")
print(f"Name: {name}")
print(f"Transport: KSh {transport}")
print(f"Food: KSh {food}")
print(f"Other expenses: KSh {other}")
print(f"Total expenses: KSh {total_expenses}")
print(f"Remaining budget: KSh {remaining}")
print(f"Within budget: {within_budget}")
print(f"Five-day expenditure: KSh {five_day_expenses}")
```

### Extension Challenge

Ask students to improve their programs by:

1. Displaying a message saying "Within budget" or "Budget exceeded" instead of `True` or `False`.
2. Calculating the percentage of the budget spent.
3. Identifying the largest expense category.
4. Displaying a warning when expenditure exceeds KSh 1,000.

**Reference solution for the first extension:**

```python
if within_budget:
    print("You stayed within your budget.")
else:
    print("You exceeded your budget.")
```

**Discussion:** What happens when the user exceeds their budget? Why does the remaining amount become negative?

# Instructor Teaching Guidelines

Throughout the lesson:

1. **Predict before executing:** Never run an example immediately. Ask students what they expect first.
2. **Use incomplete code:** Let students fill in missing variables, operators and expressions.
3. **Introduce debugging:** Give students incorrect code and allow them to identify the errors.
4. **Ask why:** Encourage students to explain why they selected a particular data type or operator.
5. **Encourage experimentation:** Let students change values and observe how the output changes.
6. **Use pair programming:** Have students discuss the problem before writing code.
7. **Reserve time for independent coding:** Every student should complete at least one task without following your demonstration.

**The central teaching principle:** Students should understand how information is represented, stored, converted and manipulated in Python. The goal is not to memorise syntax but to develop the ability to translate real-world problems into working programs.
