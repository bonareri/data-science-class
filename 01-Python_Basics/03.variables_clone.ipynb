# Interactive Python Lesson Plan: Values, Variables, Data Types and Operators

**Duration:** 2 hours 30 minutes  
**Approach:** Problem-based learning, pair programming, prediction exercises, debugging and independent practice.

## Learning outcomes

By the end of the lesson, students should be able to:

1. Identify basic Python data types.
2. Create and use variables.
3. Apply Python variable naming rules.
4. Inspect and convert data types.
5. Use arithmetic, comparison, logical and assignment operators.
6. Understand expressions and operator precedence.
7. Collect information using `input()`.
8. Build a simple interactive Python program.

# PART 2: Values and data types

**Time: 20 minutes**

### 1. Introduce values

Explain:

"A value is a piece of information that a program can work with. Every value in Python has a data type. The data type determines what kind of information it represents and what operations we can perform on it."

Introduce the four basic data types.

| Data type | Description | Example |
|---|---|---|
| Integer (`int`) | Whole numbers | `25` |
| Float (`float`) | Decimal numbers | `89.5` |
| String (`str`) | Text | `"Brian"` |
| Boolean (`bool`) | True or False | `True` |

### 2. Student activity: Identify the data type

Display these values:

```python
"Kenya"
25
89.5
True
"100"
3.14
False
```

Ask students to classify each value.

**Expected answers:**

- `"Kenya"` – String
- `25` – Integer
- `89.5` – Float
- `True` – Boolean
- `"100"` – String
- `3.14` – Float
- `False` – Boolean

### Discussion question

"Why is `"100"` a string while `100` is an integer?"

Expected explanation: The quotation marks indicate that the value is text, even though it contains numeric characters.

### 3. Introduce variables

Explain:

"Now that we understand values and data types, how can we store these values so that we can use them later?"

Introduce variables as names that refer to values.

```python
student_name = "Brian"
student_age = 24
assessment_score = 85.5
has_passed = True
```

Ask students:

- What does each variable represent?
- What data type does each variable contain?
- Why would we use descriptive variable names instead of just `x`, `y` and `z`?

### 4. Predict the output

Display:

```python
age = 20
age = 25

print(age)
```

Ask students to predict the output before executing the code.

**Expected answer:** `25`

Explain that assigning a new value to an existing variable updates the value it refers to.

### 5. Let students investigate

Ask students to execute:

```python
student_name = "Brian"
student_age = 24
assessment_score = 85.5
has_passed = True

print(type(student_name))
print(type(student_age))
print(type(assessment_score))
print(type(has_passed))
```

Ask them to explain the results.

# PART 3: Variable naming and type casting

**Time: 15 minutes**

### 1. Debugging challenge

Display the following code:

```python
student name = "Brian"
2nd_score = 90
student_age = 24
class = "Data Science"
```

Ask students:

"Which lines are valid Python? Which lines will cause errors? How would you fix them?"

**Expected corrections:**

```python
student_name = "Brian"
second_score = 90
student_age = 24
class_name = "Data Science"
```

Explain the naming rules:

- Variable names must start with a letter or underscore.
- They cannot start with a number.
- Spaces are not allowed.
- They can contain letters, numbers and underscores.
- Variable names are case-sensitive.
- Python keywords cannot be used as variable names.

### 2. Type casting

Explain:

"Sometimes, the value we have is not in the data type we need. Python allows us to convert values from one type to another. This is called type casting."

Display:

```python
age = "25"

print(age + 5)
```

Ask students to predict what will happen.

Allow them to execute it and observe the error.

Then introduce:

```python
age = int(age)

print(age + 5)
```

Ask:

- Why did the first example fail?
- What changed in the second example?
- What is the difference between `int()`, `float()` and `str()`?

Demonstrate:

```python
x = int("25")
y = float("3.5")
z = str(100)

print(x, type(x))
print(y, type(y))
print(z, type(z))
```

### Mini challenge

Ask students to convert:

- `"45"` into an integer.
- `"78.5"` into a float.
- `100` into a string.

# PART 4: Python operators

**Time: 35 minutes**

Introduce the problem:

"A school wants to calculate the performance of students based on three assessments. We need Python to calculate the total marks, average marks and determine whether a student has passed."

### 1. Arithmetic operators

Start with:

```python
assignment = 75
practical = 80
final_exam = 65
```

Ask:

"How would we calculate the total marks?"

Allow students to suggest the expression.

Then write:

```python
total = assignment + practical + final_exam

print(total)
```

Introduce the arithmetic operators:

| Operator | Meaning |
|---|---|
| `+` | Addition |
| `-` | Subtraction |
| `*` | Multiplication |
| `/` | Division |
| `//` | Floor division |
| `%` | Modulus |
| `**` | Exponentiation |

Demonstrate using:

```python
x = 10
y = 3

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x // y)
print(x % y)
print(x ** y)
```

### Student prediction exercise

Ask students to predict:

```python
print(10 // 3)
print(10 % 3)
print(2 ** 4)
```

Expected answers: `3`, `1`, `16`.

Explain the difference between floor division and modulus.

### 2. Comparison operators

Ask:

"How can we determine whether a student has achieved an average of at least 70?"

Introduce:

```python
average = 73

print(average >= 70)
```

Explain that comparison operators return Boolean values.

Introduce:

`==`, `!=`, `>`, `<`, `>=`, `<=`.

### Student activity

Ask students to determine whether:

```python
score = 65

print(score == 65)
print(score != 65)
print(score > 50)
print(score < 50)
print(score >= 70)
print(score <= 65)
```

Ask them to explain each result.

### 3. Logical operators

Present this requirement:

"A student must score at least 50 in both the assignment and final examination to meet the assessment requirements."

Ask:

"How do we check two conditions at the same time?"

Introduce:

```python
assignment_score = 75
final_exam_score = 65

assignment_passed = assignment_score >= 50
exam_passed = final_exam_score >= 50

overall_pass = assignment_passed and exam_passed

print(overall_pass)
```

Introduce `and`, `or` and `not`.

Ask students what would happen if one of the two marks fell below 50.

### 4. Assignment operators

Explain that assignment operators allow us to update variable values.

```python
total = 100

total += 50
print(total)

total -= 20
print(total)

total *= 2
print(total)
```

Ask students to predict the result after each operation.

Explain that:

- `total += 50` is equivalent to `total = total + 50`.
- `total -= 20` is equivalent to `total = total - 20`.
- `total *= 2` is equivalent to `total = total * 2`.

# PART 6: User input and interactive programs

**Time: 20 minutes**

### Introduce user input

Explain:

"So far, we have been assigning values directly to variables. But real applications need to collect information from users. Python provides the `input()` function for this."

Demonstrate:

```python
name = input("Enter your name: ")

print(f"Hello, {name}!")
```

Ask students to run the program and enter their names.

### Explain an important concept

"The `input()` function always returns a string. If we need to perform calculations using the input, we must convert it into the appropriate data type."

Demonstrate:

```python
age = input("Enter your age: ")

print(type(age))
```

Then:

```python
age = int(input("Enter your age: "))

print(f"In 5 years, you will be {age + 5}")
```

Explain the difference between `int()` and `float()` when collecting numerical input.

### Practical exercise: Shopping calculator

Ask students to build a program that:

1. Asks for the price of a product.
2. Asks for the quantity.
3. Calculates the total cost.
4. Displays the result.

Example:

```python
product_price = float(input("Enter product price: "))
quantity = int(input("Enter quantity: "))

total_cost = product_price * quantity

print(f"Your total cost is {total_cost}")
```

Let students implement this independently.

# PART 7: Main practical challenge

**Time: 25 minutes**

## Mini project: Personal Expense Calculator

### Scenario

You have been asked to build a simple Python application that helps someone track their daily expenses.

The application should collect the user's spending information and determine whether they have stayed within their daily budget.

### Requirements

Your program should:

1. Ask the user for their name.

# Instructor's answer key

## Expense calculator

One possible solution:

```python
name = input("Enter your name: ")

transport = float(input("Enter transport expenses: "))
food = float(input("Enter food expenses: "))
other = float(input("Enter other expenses: "))

total_expenses = transport + food + other

budget = 1000

remaining = budget - total_expenses

within_budget = total_expenses <= budget

weekly_expenses = total_expenses * 5

print("\nEXPENSE SUMMARY")
print(f"Name: {name}")
print(f"Transport: KSh {transport}")
print(f"Food: KSh {food}")
print(f"Other expenses: KSh {other}")
print(f"Total expenses: KSh {total_expenses}")
print(f"Remaining budget: KSh {remaining}")
print(f"Within budget: {within_budget}")
print(f"Five-day expenditure: KSh {weekly_expenses}")
```

**Discussion:** What happens when the user exceeds the budget? The remaining amount becomes negative. This is an opportunity to discuss how the program represents overspending.

## Exit ticket solution

```python
name = input("Enter student name: ")

mark1 = float(input("Enter first mark: "))
mark2 = float(input("Enter second mark: "))
mark3 = float(input("Enter third mark: "))

average = (mark1 + mark2 + mark3) / 3

passed = average >= 70

print("\nSTUDENT RESULTS")
print(f"Name: {name}")
print(f"Average: {average}")
print(f"Passed: {passed}")
```

# Teaching reminders

1. Never write a long block of code without involving students.
2. Ask students to predict outputs before executing code.
3. Give students time to experiment with different values.
4. Encourage students to explain errors rather than immediately fixing them for them.
5. Ask students to explain what a line of code does and why it is necessary.
6. Allow students to suggest the next line of code during demonstrations.
7. Use pair programming to encourage discussion.
8. Make sure every student gets an opportunity to write code independently.

**The most important principle:** Students should be doing more thinking and coding than simply watching you code. Your role is to facilitate their learning, ask the right questions and help them develop problem-solving skills.
