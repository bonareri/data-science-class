# 1. Introduction to Control Flow

Normally, Python executes instructions sequentially, from top to bottom. However, real-world programs need to make decisions and repeat certain operations.

**Control flow** refers to the order in which instructions are executed in a program.

Python provides two important mechanisms:

- **Conditional statements:** Allow programs to make decisions.
- **Loops:** Allow programs to repeat instructions.

**Think about it:** How would you build a program that determines whether someone is eligible to vote? Would you need to execute every possible instruction, or should the program make a decision based on the person's age?

# 2. Conditional Statements

Conditional statements allow a program to execute different instructions depending on whether a condition is `True` or `False`.

## 2.1 The `if` Statement

An `if` statement executes a block of code only when its condition evaluates to `True`.

Syntax:

```python
if condition:
    # Code to execute when condition is True
```

Example:

```python
age = 20

if age >= 18:
    print("You are eligible to vote")
```

**Important concepts:**
- A condition is an expression that evaluates to `True` or `False`.
- The colon (`:`) marks the beginning of the conditional block.
- Indentation identifies the instructions belonging to the condition.

### Activity: Predict the Output

## 2.2 The `if-else` Statement

An `if-else` statement provides two possible execution paths.

```python
number = 7

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

**Concepts:**
- `%` is the modulo operator. It returns the remainder after division.
- `==` checks whether two values are equal.
- `else` executes when the `if` condition is false.

## 2.3 Chained Conditionals: `if-elif-else`

When a program needs to evaluate more than two possibilities, we use `elif`.

Python evaluates conditions from top to bottom. The first condition that evaluates to `True` executes, and the remaining branches are skipped.

Example: Grading System

| Score | Grade |
|---|---|
| 90–100 | A |
| 80–89 | B |
| 70–79 | C |
| Below 70 | F |

```python
score = 85

if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
else:
    print("Grade F")
```

## 2.4 Nested Conditionals

A nested conditional is an `if` statement placed inside another conditional statement.

Example: Examination Access System

```python
registered = True
fees_paid = False

if registered:
    if fees_paid:
        print("You can access the examination")
    else:
        print("Please clear your fees")
else:
    print("You are not registered")
```

The inner condition is evaluated only when the outer condition is satisfied.

**Student Exercise:** A student can access a learning platform if they have registered and paid their fees. Write the same program using logical operators instead of nested conditionals.

**Solution:**

```python
registered = True
fees_paid = False

if registered and fees_paid:
    print("Access granted")
else:
    print("Access denied")
```

## 2.5 Logical Operators

Logical operators combine multiple conditions.

| Operator | Meaning | Example |
|---|---|---|
| `and` | Both conditions must be True | `age >= 18 and age <= 35` |
| `or` | At least one condition must be True | `day == "Saturday" or day == "Sunday"` |
| `not` | Reverses a Boolean result | `not is_registered` |

Example:

```python
age = 25
has_id = True

if age >= 18 and has_id:
    print("Access granted")
```

### Student Exercise

**Solution:**

```python
is_student = False
age = 65

if is_student or age >= 60:
    print("Discount available")
else:
    print("No discount")
```

## 2.6 Short-Circuit Evaluation and the Guardian Pattern

Python evaluates logical expressions from left to right. With `and`, if the first condition is false, Python does not need to evaluate the remaining conditions.

This is called **short-circuit evaluation**.

It can prevent errors when an operation depends on a particular condition being satisfied.

Example: Preventing Division by Zero

```python
x = 10
y = 0

if y != 0 and x / y > 2:
    print("Passed")
else:
    print("Safe")
```

The first condition checks that `y` is not zero before attempting division.

This protective check is known as the **guardian pattern**.

# 4. Loops in Python

Loops allow us to execute a block of code repeatedly without writing the same instructions multiple times.

Python provides two main types of loops:

- `while` loops
- `for` loops

## 4.1 Updating Variables

Before understanding loops, students need to understand variable updates.

An assignment such as:

```python
x = x + 1
```

means: Take the current value of `x`, add 1, and store the new value in `x`.

Other examples:

```python
x = 0       # Initialization
x = x + 1   # Increment
x = x - 1   # Decrement
```

The shorthand operators are:

```python
x += 1
x -= 1
```

## 4.2 The `while` Loop

A `while` loop executes a block of code as long as its condition remains `True`.

Syntax:

```python
while condition:
    # Instructions to repeat
```

Example:

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

**How it works:**
1. Python checks whether `count <= 5`.
2. If true, it executes the loop body.
3. It updates the variable.
4. It checks the condition again.
5. When the condition becomes false, the loop stops.

Each execution of the loop body is called an **iteration**.

### Activity: Trace the Loop

Ask students to predict the output and complete the missing values.

```python
number = 5

while number > 0:
    print(number)
    number -= 1

print("Blastoff!")
```

**Questions:**

- What happens when `number` becomes zero?
- Why is the final message outside the loop?

### Student Exercise

## 4.3 Infinite Loops

An infinite loop occurs when a loop continues executing without reaching a termination condition.

Example:

```python
count = 1

while count <= 5:
    print(count)
```

**Problem:** The value of `count` never changes, so the condition remains true.

**Correction:**

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

**Student Exercise:** Identify and correct the error in this code:

```python
number = 10

while number >= 1:
    print(number)
    number += 1
```

**Solution:** Change `number += 1` to `number -= 1`.

## 4.4 The `break` Statement

The `break` statement immediately terminates a loop, even when its condition is still true.

Example:

```python
numbers = [3, 7, 10, 15, 20]

for number in numbers:
    print("Checking:", number)

    if number == 15:
        print("Number found!")
        break
```

Once 15 is found, the loop stops.

## 4.5 The `continue` Statement

The `continue` statement skips the remaining instructions in the current iteration and moves to the next iteration.

Example:

```python
numbers = [1, 2, 3, 4, 5]

for number in numbers:
    if number == 3:
        continue

    print(number)
```

Output:

```text
1
2
4
5
```

**Remember:**
- `break` stops the entire loop.
- `continue` skips only the current iteration.

### Student Exercise

**Solution:**

```python
for number in range(1, 11):
    if number % 2 != 0:
        continue

    print(number)
```

# 5. The `for` Loop

A `for` loop iterates over items in a sequence, such as a list, string or range of numbers.

Syntax:

```python
for variable in sequence:
    # Instructions
```

Example:

```python
students = ["Mary", "John", "Peter", "Faith"]

for student in students:
    print(student)
```

**Important concepts:**
- The sequence contains the items being processed.
- The iteration variable holds the current item.
- The loop executes once for every item in the sequence.

## 5.1 Iterating Through Strings

Strings are sequences of characters.

```python
word = "Python"

for letter in word:
    print(letter)
```

Python processes one character at a time.

## 5.2 Using `range()`

The `range()` function generates a sequence of integers.

Its common syntax is:

```python
range(start, stop, step)
```

- `start`: Starting value (default 0).
- `stop`: Ending boundary, which is excluded.
- `step`: Difference between consecutive values (default 1).

Example:

```python
for number in range(1, 6):
    print(number)
```

Output: 1, 2, 3, 4, 5.

### Student Exercise

Write programs to:
1. Print numbers from 1 to 10.
2. Print even numbers between 2 and 20.
3. Count backwards from 10 to 1.

**Solutions:**

```python
# 1. Numbers from 1 to 10
for number in range(1, 11):
    print(number)

# 2. Even numbers from 2 to 20
for number in range(2, 21, 2):
    print(number)

# 3. Countdown
for number in range(10, 0, -1):
    print(number)
```

## 5.3 `for` vs `while`

| `for` loop | `while` loop |
|---|---|
| Iterates through a sequence. | Repeats while a condition is true. |
| Often used when processing known collections. | Often used when the stopping point depends on a condition. |
| Automatically moves to the next item. | Requires appropriate condition updates where necessary. |

# 6. Combining Loops and Conditional Statements

Loops become more powerful when combined with conditions. This allows us to filter, count and analyse information.

Example: Finding High Scores

```python
scores = [45, 88, 72, 91, 65, 100]

for score in scores:
    if score >= 90:
        print("High score:", score)
```

The loop processes each score, while the conditional determines which scores are displayed.

## 6.1 Common Loop Patterns

These are common programming patterns used when processing data.

### A. Counting

Counts how many items satisfy a condition.

```python
numbers = [2, 4, 7, 9, 12]
count = 0

for number in numbers:
    if number > 5:
        count += 1

print("Count:", count)
```

### B. Summation

Calculates the total of values.

```python
numbers = [5, 10, 15]
total = 0

for number in numbers:
    total += number

print("Total:", total)
```

### C. Finding the Maximum

Tracks the largest value encountered.

```python
numbers = [3, 9, 1, 12, 7]
maximum = numbers[0]

for number in numbers:
    if number > maximum:
        maximum = number

print("Maximum:", maximum)
```

### D. Finding the Minimum

Tracks the smallest value encountered.

```python
numbers = [8, 2, 10, 5]
minimum = numbers[0]

for number in numbers:
    if number < minimum:
        minimum = number

print("Minimum:", minimum)
```

### E. Searching

Checks whether a particular item exists.

```python
names = ["Alice", "Bob", "Charlie"]
found = False

for name in names:
    if name == "Bob":
        found = True
        break

print("Found!" if found else "Not found.")
```

# 7. Nested Loops

A nested loop is a loop placed inside another loop.

The inner loop completes all its iterations before the outer loop moves to its next iteration.

Example:

```python
for row in range(1, 4):
    for column in range(1, 4):
        print(row, column)
```

The outer loop controls the rows, while the inner loop controls the columns.

**Concept:** If the outer loop executes 3 times and the inner loop executes 3 times for every outer iteration, the inner block executes 9 times in total.

## 7.1 Generating a Grid

```python
for row in range(3):
    for column in range(3):
        print("*", end=" ")
    print()
```

Output:

```text
* * *
* * *
* * *
```

`end=" "` keeps the output on the same line, while the final `print()` moves to a new line.

## 7.2 Multiplication Tables

```python
for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end="\t")
    print()
```

The outer loop selects the number, while the inner loop generates its multiplication results.

# 8. Final Practical Challenge: Student Performance Analyzer

**Scenario:** You are working as a junior data analyst. A school has provided you with a list of examination scores and wants to understand student performance.

```python
scores = [45, 78, 90, 56, 82, 34, 95, 67, 88, 72]
```

Build a program that:

1. Classifies each student as passed or failed (pass mark: 50).
2. Counts the number of students who passed.
3. Counts the number who failed.
4. Calculates the total marks.
5. Calculates the average score.
6. Finds the highest and lowest scores.

**Requirements:**
- Use a `for` loop.
- Use conditional statements.
- Use variables to track totals and counts.
- Do not use `sum()`, `max()` or `min()` for these calculations.

### Instructor Solution

```python
scores = [45, 78, 90, 56, 82, 34, 95, 67, 88, 72]

passed = 0
failed = 0
total = 0
highest = scores[0]
lowest = scores[0]

for score in scores:

    if score >= 50:
        print(score, "Passed")
        passed += 1
    else:
        print(score, "Failed")
        failed += 1

    total += score

    if score > highest:
        highest = score

    if score < lowest:
        lowest = score

average = total / len(scores)

print("Students passed:", passed)
print("Students failed:", failed)
print("Total marks:", total)
print("Average:", average)
print("Highest:", highest)
print("Lowest:", lowest)
```
