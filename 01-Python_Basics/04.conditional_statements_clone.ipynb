- **Conditional statements:** Allow programs to make decisions.
- **Loops:** Allow programs to repeat instructions.

## 2.3 Chained Conditionals: `if-elif-else`

What happens when a program needs to choose from more than two possibilities?

Consider a school grading system.

| Score | Grade |
|---|---|
| 90–100 | A |
| 80–89 | B |
| 70–79 | C |
| 60–69 | D |
| Below 60 | F |

Using several separate `if` statements can lead to unnecessary checks.

Python provides `elif`, which means "else if".

### Example

```python
score = 85

if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
elif score >= 60:
    print("Grade D")
else:
    print("Grade F")
```

Python evaluates the conditions from top to bottom.

The first condition that evaluates to `True` executes. The remaining branches are skipped.

### Important Concept: Order Matters

Consider this incorrect grading system:

```python
score = 95

if score >= 60:
    print("Grade D")
elif score >= 70:
    print("Grade C")
elif score >= 80:
    print("Grade B")
elif score >= 90:
    print("Grade A")
else:
    print("Grade F")
```

**Predict the output before running it.**

Why does a score of 95 not receive Grade A?

The first condition is already `True`, so Python never evaluates the remaining conditions.

This demonstrates why conditions should be arranged carefully.

## 2.4 Nested Conditionals

A nested conditional is an `if` statement placed inside another conditional statement.

Imagine a student wants to access an examination.

The student must:

1. Be registered.
2. Have paid the required fees.

Both requirements must be satisfied.

### Example: Examination Access System

```python
registered = True
fees_paid = False

if registered:
    if fees_paid:
        print("You can access the examination")
    else:
        print("Please clear your fees")
else:
    print("You are not registered")
```

The inner condition is evaluated only when the outer condition is satisfied.

## 2.5 Logical Operators

Logical operators allow us to combine multiple conditions.

Python provides three main logical operators:

| Operator | Meaning | Example |
|---|---|---|
| `and` | Both conditions must be True | `age >= 18 and has_id` |
| `or` | At least one condition must be True | `is_student or is_senior` |
| `not` | Reverses a Boolean result | `not is_registered` |

### Understanding `and`

Both conditions must be `True`.

```python
age = 25
has_id = True

if age >= 18 and has_id:
    print("Access granted")
else:
    print("Access denied")
```

### Understanding `or`

At least one condition must be `True`.

```python
is_student = False
age = 65

if is_student or age >= 60:
    print("Discount available")
else:
    print("No discount")
```

### Understanding `not`

The `not` operator reverses a Boolean result.

```python
is_registered = False

if not is_registered:
    print("Please register first")
```

## 2.6 Short-Circuit Evaluation and the Guardian Pattern

Python evaluates logical expressions from left to right.

With `and`, if the first condition is `False`, Python does not need to evaluate the remaining conditions.

This is called **short-circuit evaluation**.

It can help prevent errors when an operation depends on another condition being satisfied first.

### Example: Preventing Division by Zero

```python
x = 10
y = 0

if y != 0 and x / y > 2:
    print("Passed")
else:
    print("Safe")
```

Python first checks:

```python
y != 0
```

Since `y` is zero, this condition is `False`.

Python does not attempt to evaluate:

```python
x / y > 2
```

This prevents a division-by-zero error.

This protective arrangement is known as the **guardian pattern**.

### Think and Discuss

What would happen if we reversed the order of the conditions?

```python
if x / y > 2 and y != 0:
    print("Passed")
```

Why is this unsafe?

**Key takeaway:** When one operation depends on another condition being satisfied, place the protective condition first.

## 4.2 The `while` Loop

A `while` loop executes a block of code as long as its condition remains `True`.

### Syntax

```python
while condition:
    # Instructions to repeat
```

### Example

Suppose we want to print numbers from 1 to 5.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

### How It Works

1. Python checks whether `count <= 5`.
2. If the condition is `True`, the loop body executes.
3. The value of `count` increases by 1.
4. Python checks the condition again.
5. When the condition becomes `False`, the loop stops.

Each execution of the loop body is called an **iteration**.

## 4.5 The `continue` Statement

The `continue` statement skips the remaining instructions in the current iteration and moves to the next iteration.

### Example

```python
numbers = [1, 2, 3, 4, 5]

for number in numbers:
    if number == 3:
        continue

    print(number)
```

Output:

```text
1
2
4
5
```

When Python encounters 3, it skips the `print()` instruction and moves to the next number.

### Understanding the Difference

| `break` | `continue` |
|---|---|
| Stops the entire loop. | Skips the current iteration. |
| No further iterations occur. | The next iteration continues. |

## 6.1 Common Loop Patterns

When working with datasets, we often need to perform repeated operations.

Some common patterns include counting, summation, finding maximum and minimum values, and searching.

### A. Counting

Counting allows us to determine how many items satisfy a condition.

Example: Count how many scores are above 70.

```python
numbers = [2, 4, 7, 9, 12]

count = 0

for number in numbers:
    if number > 5:
        count += 1

print("Count:", count)
```

**Understand the logic:**

- Start with a count of zero.
- Check each number.
- If the number satisfies the condition, increase the count.
- Display the final count.

### B. Summation

Summation calculates the total of values.

```python
numbers = [5, 10, 15]

total = 0

for number in numbers:
    total += number

print("Total:", total)
```

The variable `total` accumulates the values as the loop executes.

### C. Finding the Maximum

Finding the maximum means identifying the largest value in a collection.

```python
numbers = [3, 9, 1, 12, 7]

maximum = numbers[0]

for number in numbers:
    if number > maximum:
        maximum = number

print("Maximum:", maximum)
```

We initialise `maximum` using the first value in the list.

Each new value is compared with the current maximum. If it is larger, the maximum is updated.

### D. Finding the Minimum

Finding the minimum means identifying the smallest value.

```python
numbers = [8, 2, 10, 5]

minimum = numbers[0]

for number in numbers:
    if number < minimum:
        minimum = number

print("Minimum:", minimum)
```

### E. Searching

Searching checks whether a particular item exists in a collection.

```python
names = ["Alice", "Bob", "Charlie"]

found = False

for name in names:
    if name == "Bob":
        found = True
        break

if found:
    print("Name found")
else:
    print("Name not found")
```

We start with `found = False`.

When the item is found, we change the variable to `True` and stop searching.

# 7. Nested Loops

A nested loop is a loop placed inside another loop.

The inner loop completes all its iterations before the outer loop moves to its next iteration.

### Example

```python
for row in range(1, 4):
    for column in range(1, 4):
        print(row, column)
```

The outer loop controls the rows, while the inner loop controls the columns.

### Understanding the Execution

If the outer loop executes 3 times and the inner loop executes 3 times for every outer iteration, the inner block executes:

3 × 3 = 9 times.

## 7.1 Generating a Grid

Nested loops can be used to generate patterns.

The `end=" "` argument keeps the output on the same line.

```python
scores = [45, 78, 90, 56, 82, 34, 95, 67, 88, 72]

passed = 0
failed = 0
total = 0

highest = scores[0]
lowest = scores[0]

# Write your loop below
```
