# 1. Introduction to Using Web Services

## 1.1 Where Does Data Come From?

Throughout data analysis, we work with data from different sources.

So far, you have worked with data stored in files such as:

- CSV files
- Excel spreadsheets
- JSON files
- Databases

However, not all useful data is stored in files that we can download directly.

Consider these situations:

- A business wants to analyse current exchange rates.
- A researcher wants to collect information about countries around the world.
- A company wants to monitor product prices from an online store.
- A data analyst wants to collect information about customer reviews on a website.
- A researcher wants to study news headlines published online.

In these situations, downloading a static CSV file may not be possible or practical. The information may be available directly through a website or an online service.

This is where **web services** become useful.

## 1.2 What Are Web Services?

Web services allow different applications and systems to communicate and exchange information over the internet.

For example, imagine a mobile banking application that displays the current exchange rate between the Kenyan shilling and the US dollar.

The banking application may not calculate or store the exchange rate itself. Instead, it can request the latest information from another online service and display the response to the user.

There are two common ways of collecting information from websites:

1. **Accessing APIs:** Requesting data from an online service through a defined interface.

We will learn both approaches.

## 1.3 APIs vs Web Scraping

Imagine you want to collect the names and populations of countries around the world.

There are two possible approaches.

**Using an API**

A service may provide an endpoint that allows you to request country information. The response may contain country names, population figures, currencies and other details in a structured format.

**Using web scraping**

You could visit a website displaying country information, inspect its HTML structure and write Python code to extract the country names and populations.

The main difference is how we access the information.

| APIs | Web Scraping |
|---|---|
| Accesses data through a defined interface | Extracts information from web pages |
| Usually returns structured data such as JSON | Usually starts with HTML content |
| Uses API endpoints | Uses HTML elements and their attributes |
| Often has documentation explaining how to request data | Requires inspecting the page structure |
| May require an API key or other authentication | May be restricted by website rules or technical controls |

**Important:** APIs and web scraping are not competing technologies. They are different approaches to collecting online data. When a suitable API is available, it is generally preferable to use it rather than scrape the website.

# 2. Understanding How Websites Communicate

Before writing code to retrieve information from the internet, we need to understand what happens when we access a website.

## 2.1 What Happens When You Visit a Website?

Imagine you open your browser and visit an online bookstore.

You enter the website address, and the browser displays the page containing book titles, prices and descriptions.

But where does that information come from?

A simplified explanation is:

1. Your browser sends a request to the website's server.
2. The server receives the request.
3. The server processes the request and retrieves the appropriate information.
4. The server sends a response back to your browser.
5. Your browser displays the returned content.

This communication follows a set of rules called **HTTP (Hypertext Transfer Protocol)**.

HTTP defines how clients and servers communicate over the web.

There are three important components to understand:

- **Client:** The application requesting information. For example, your browser or a Python program.
- **Server:** The computer or system that receives the request, processes it and sends back a response.
- **Database:** An organized place where information is stored. A server may communicate with a database to retrieve, add or update information when handling a request.

**The server and the database are not the same thing.** The server handles the request, while the database stores the information. They may run on separate computers, or they may run on the same computer, but they perform different roles.

For example, imagine you use a school website to check your grades:

1. Your browser (the client) sends a request to the school's server.
2. The server receives and processes the request.
3. If the grades are stored in a database, the server requests the relevant grades from that database.
4. The database returns the grades to the server.
5. The server sends a response to your browser.
6. Your browser displays the grades.

The same basic communication process happens when Python retrieves information from an online service. In that case, your Python program acts as the client.

## 2.2 Understanding Requests and Responses

There are two important terms to understand:

**Request:** A message sent by a client asking a server to perform an action or provide information.

**Response:** The message returned by the server after processing the request.

For example, suppose a Python program wants to retrieve information about Kenya from an online country information service.

The program sends a request asking for Kenya's information.

The server may respond with:

- The country's official name.
- Its capital city.
- Its population.
- Its currency.
- Its region.

The response contains the information requested.

This is the foundation of working with APIs and web scraping.

## 2.3 Understanding URLs and Endpoints

A URL (Uniform Resource Locator) is the address used to locate a resource on the internet.

For example:

`https://restcountries.com/v3.1/name/kenya`

This URL points to an online service that provides information about Kenya.

An **API endpoint** is a specific URL through which an API makes a particular resource or operation available.

For example:

- An endpoint may return information about one country.
- Another endpoint may return information about several countries.
- Another may return information about a country's currency.

The endpoint tells the service what information we want.

## 2.4 Understanding HTTP Methods

HTTP supports different methods that describe what a client wants to do.

Some common methods include:

| Method | Purpose | Example |
|---|---|---|
| `GET` | Retrieve information | Get a list of countries |
| `POST` | Send information to create a resource | Submit a new customer record |
| `PUT` | Replace an existing resource | Replace a customer profile |
| `PATCH` | Update part of an existing resource | Change a customer's email |
| `DELETE` | Delete a resource | Delete a customer record |

In this lesson, we will mainly use **GET requests** because our goal is to retrieve data for analysis.

## 2.5 Understanding Status Codes

After sending a request, the server returns a response with a status code.

The status code helps us understand what happened to our request.

| Status code | Meaning | Explanation |
|---|---|---|
| `200` | OK | The request was successful. |
| `201` | Created | A resource was successfully created. |
| `400` | Bad Request | The request was not valid. |
| `401` | Unauthorized | Authentication is required or invalid. |
| `403` | Forbidden | Access to the requested resource is not allowed. |
| `404` | Not Found | The requested resource could not be found. |
| `429` | Too Many Requests | Too many requests were sent within a given period. |
| `500` | Internal Server Error | Something went wrong on the server. |

For data collection, we should check whether a request was successful before attempting to process its response.

# 3. Accessing Data Through APIs

## 3.1 What Is an API?

API stands for **Application Programming Interface**.

An API is a defined way for one software application to communicate with another.

Think of an API as a service desk.

Imagine visiting a restaurant:

1. You look at the menu.
2. You place an order with the waiter.
3. The waiter communicates your order to the kitchen.
4. The kitchen prepares the food.
5. The waiter brings the food to you.

You do not need to enter the kitchen or understand how the food is prepared.

An API works in a similar way.

- Your Python program is the customer.
- The API is the waiter.
- The server is the kitchen.
- The returned data is the food.

Your program sends a request through the API, and the API provides the information made available by the service.

## 3.2 Why Are APIs Important in Data Analysis?

APIs allow data analysts to retrieve information directly from online sources.

For example:

- Financial analysts can retrieve exchange rates.
- Researchers can retrieve population statistics.
- Business analysts can retrieve sales information from business platforms.
- Data scientists can retrieve weather information for predictive modelling.
- Developers can retrieve information from social media or other online platforms where access is permitted.

Instead of manually downloading and copying information, we can automate the collection process using Python.

## 3.3 Understanding API Responses

APIs commonly return data in formats such as JSON or XML.

In this lesson, we will focus on JSON because it is widely used by web APIs.

JSON stands for **JavaScript Object Notation**.

It is a format used to store and exchange structured information.

Consider the following JSON example:

```json
{
    "name": "Kenya",
    "capital": "Nairobi",
    "population": 55000000,
    "region": "Africa"
}
```

This response represents information about one country.

It contains:

- Keys: `name`, `capital`, `population`, and `region`.
- Values: `"Kenya"`, `"Nairobi"`, `55000000`, and `"Africa"`.

JSON uses key-value pairs, making it easy to organize and retrieve information.

JSON data has structures that are similar to Python dictionaries and lists.

For example:

| JSON structure | Python equivalent |
|---|---|
| Object `{}` | Dictionary |
| Array `[]` | List |
| String | String |
| Number | Integer or float |
| `true` / `false` | `True` / `False` |
| `null` | `None` |

This similarity makes it easier to work with API responses in Python.

## 3.5 Converting an API Response into Python Data

A successful request does not automatically mean that we have a Python dictionary or list that we can analyse.

The response contains the data returned by the server. We need to convert it into a format that Python can work with.

When an API returns JSON, we can use the `.json()` method.

```python
data = response.json()
data
```

This converts the JSON response into Python objects.

You may notice that the response is a list containing dictionaries.

Why?

The API can return more than one matching country, so the response is structured as a list.

We can inspect the first record:

```python
country = data["data"]["objects"][0]
```

We can then retrieve individual values using dictionary keys.

Using the API response:

1. Store the first country record in a variable.
2. Extract the country's common name.
3. Extract its population.
4. Extract its region.
5. Print the results.


```python
country = data["data"]["objects"][0]

print(f"Country: {country['names']['common']}")
print(f"Capital: {country['capitals'][0]['name']}")
print(f"Population: {country['population']:,}")
print(f"Region: {country['region']}")
print(f"Currency: {country['currencies'][0]['name']}")
print(f"Flag: {country['flag']['emoji']}")
```

Notice that some information is nested.

For example:

```python
country["name"]["common"]
```

This means:

1. Access the `name` dictionary.
2. Retrieve the value associated with the `common` key.

This is called **nested data**.

## 3.6 Retrieving Multiple Records

So far, we have retrieved information about one country.

However, data analysis often involves comparing multiple records.

For example, we may want to compare the populations of Kenya, Uganda, Tanzania and Rwanda.

The REST Countries API allows us to request information about multiple countries.

```python
url = "https://restcountries.com/v3.1/region/africa"

response = requests.get(url)

data = response.json()
```

This request retrieves country information for countries in Africa.

We can inspect how many records were returned:

```python
print(len(data))
```

We can loop through the records to display the country names:

```python
for country in data:
    print(country["name"]["common"])
```

We can also retrieve multiple fields:

```python
for country in data:
    name = country["name"]["common"]
    population = country["population"]
    region = country["region"]

    print(name, population, region)
```

This demonstrates how we can retrieve multiple records from an API and extract the information we need.

## 3.7 Converting API Data into a Pandas DataFrame

Once we retrieve data from an API, we often want to analyse it using Pandas.

Remember that Pandas works with tabular data arranged into rows and columns.

However, API responses may contain nested dictionaries and lists. We may therefore need to select the fields we want before creating a DataFrame.

Let's extract country names, populations and regions.

```python
import pandas as pd

country_data = []

for country in data:
    country_data.append({
        "country": country["name"]["common"],
        "population": country["population"],
        "region": country["region"]
    })
```

Convert the extracted information into a DataFrame:

```python
df = pd.DataFrame(country_data)

df.head()
```

Now we can use familiar Pandas operations.

For example, calculate the average population:

```python
df["population"].mean()
```

Find the five most populated countries in the returned data:

```python
df.nlargest(5, "population")
```

Count the number of countries:

```python
df["country"].count()
```

This is an important step because it connects API data collection to the data analysis techniques you have already learned.

## 3.8 Using Query Parameters

Sometimes, an API allows us to customize a request using additional values called **query parameters**.

For example, a data service may allow us to request:

- A specific country.
- A particular date range.
- A specific category.
- A certain number of records.

Query parameters are additional values passed to an API to control the information returned.

They are commonly included in a URL after a question mark (`?`).

For example:

```text
https://example.com/api/products?category=phones
```

Here:

- `category` is the parameter.
- `phones` is its value.

When there are multiple parameters, they are separated using `&`.

```text
https://example.com/api/products?category=phones&limit=10
```

This requests phone products and limits the response to 10 records, assuming the API supports those parameters.

In Python, we can pass parameters using the `params` argument.

```python
url = "https://example.com/api/products"

params = {
    "category": "phones",
    "limit": 10
}

response = requests.get(url, params=params)
```

The example illustrates how query parameters are passed. The endpoint and parameter names must be supported by the actual API being used.

## 3.9 Handling API Errors

Not every API request will succeed.

Some common reasons include:

- Incorrect endpoint.
- Internet connection problems.
- Invalid authentication credentials.
- Requests exceeding the API's allowed limits.
- Server problems.

We should therefore check the response before processing it.

One useful method is `raise_for_status()`.

```python
response = requests.get(url)

response.raise_for_status()

data = response.json()
```

If the server returns an unsuccessful HTTP status code, `raise_for_status()` raises an exception.

We can handle errors using `try` and `except`.

```python
try:
    response = requests.get(url, timeout=10)

    response.raise_for_status()

    data = response.json()

    print(data)

except requests.exceptions.RequestException as error:
    print("An error occurred:", error)
```

The `timeout` argument prevents the program from waiting indefinitely for a response.

# 4. Introduction to Web Scraping

## 4.1 What Is Web Scraping?

Not every website provides an API that allows us to retrieve its information.

Sometimes, the information we need is displayed directly on a web page.

For example, imagine you want to analyse book prices from an online bookstore.

You visit the website and see:

- Book titles.
- Prices.
- Ratings.
- Availability.

You could manually copy this information into Excel, but that would become time-consuming if the website contains hundreds or thousands of books.

Instead, you can write a Python program to extract the information automatically.

This process is called **web scraping**.

Web scraping is the process of automatically collecting information from web pages by extracting data from their HTML content.

It is useful when:

- Information is publicly displayed on web pages.
- No suitable API is available.
- You need to collect information from many pages.
- You want to automate repetitive data collection tasks.

Before scraping a website, always check its terms of service, permissions and scraping rules.

## 4.2 Understanding HTML Before Scraping

To understand web scraping, we first need to understand how web pages are structured.

Web pages are commonly built using HTML.

HTML stands for **HyperText Markup Language**.

It describes the structure and content of a web page.

For example, consider the following HTML:

```html
<h1>Available Books</h1>

<div class="book">
    <h3>Python for Beginners</h3>
    <p class="price">£25.99</p>
    <p class="availability">In stock</p>
</div>
```

Let's understand what each element represents.

| HTML element | Purpose |
|---|---|
| `<h1>` | Represents a main heading |
| `<div>` | Groups related content |
| `<h3>` | Represents a heading |
| `<p>` | Represents a paragraph |
| `class` | Gives an element a class name that can be used to identify it |

In this example, the book information is grouped inside a `div` element.

The book title is inside an `h3` element.

The price and availability are stored inside separate paragraph elements.

This structure is important because web scraping works by locating and extracting information from these HTML elements.

## 4.3 Understanding HTML Attributes

HTML elements can have attributes that provide additional information about them.

For example:

```html
<p class="price">£25.99</p>
```

Here:

- `p` is the HTML element.
- `class` is an attribute.
- `price` is the attribute value.
- `£25.99` is the text displayed inside the element.

Why is this important?

Imagine a web page contains many paragraph elements:

```html
<p>Welcome to our bookstore.</p>
<p>Discover our latest books.</p>
<p class="price">£25.99</p>
```

If we want to extract only prices, we need a way to identify the correct paragraph.

The `class="price"` attribute helps us do that.

## 4.4 Understanding HTML Elements and Nested Structures

HTML elements can contain other elements.

For example:

```html
<article class="book">
    <h3>
        <a href="book.html">Python for Beginners</a>
    </h3>

    <p class="price">£25.99</p>
</article>
```

In this example:

- The `article` element represents one book.
- The `h3` element contains a heading.
- The `a` element contains the book title and a link.
- The `p` element contains the price.

The elements are nested inside one another.

When scraping a website, understanding this structure helps us locate the information we want.

Website:

https://books.toscrape.com/

Imagine we want to collect book titles and prices from this website.

This sends a GET request to the website.

```text
200
```

## 5.3 Parsing HTML with BeautifulSoup

Parsing means analysing the structure of content so that we can locate and work with specific parts of it.

BeautifulSoup converts the HTML text into a structure that Python can navigate.

```python
soup = BeautifulSoup(response.text, "html.parser")
```

Let's understand this statement:

- `BeautifulSoup()` creates a BeautifulSoup object.
- `response.text` provides the HTML content.
- `"html.parser"` tells BeautifulSoup to use Python's built-in HTML parser.

We can now use `soup` to search through the web page.

## 5.4 Finding HTML Elements

Before extracting information, we need to know which HTML elements contain it.

Suppose the page contains the following structure:

```html
<article class="product_pod">
    <h3>
        <a title="Python for Beginners">Book title</a>
    </h3>

    <p class="price_color">£25.99</p>
</article>
```

We want to extract the book title.

We know that the title is inside an `a` element.

BeautifulSoup provides methods for locating HTML elements.

The first method is `find()`.

### Using `find()`

The `find()` method returns the first matching HTML element.

For example:

```python
book = soup.find("article")
```

This finds the first `article` element on the page.

To retrieve the first heading:

```python
title = soup.find("h3")

print(title)
```

Notice that BeautifulSoup returns the HTML element itself, not just the text.

To extract the text inside an element, use `.text`.

```python
print(title.text)
```

### Using `find_all()`

What if we want to retrieve all book titles rather than just the first one?

This is where `find_all()` becomes useful.

The `find_all()` method returns all matching elements.

```python
titles = soup.find_all("h3")
```

We can loop through the results:

```python
for title in titles:
    print(title.text)
```

The difference:

| Method | What it does |
|---|---|
| `find()` | Returns the first matching element |
| `find_all()` | Returns all matching elements |

This distinction is important when scraping pages containing multiple records.

## 5.5 Finding Elements Using Attributes

A web page can contain many elements of the same type.

For example, a page may have several paragraph elements.

But we may only want those containing book prices.

Consider:

```html
<p class="price_color">£25.99</p>
```

We can use the class attribute to locate the correct elements.

```python
prices = soup.find_all("p", class_="price_color")
```

Notice that we use `class_` instead of `class`.

This is because `class` is a reserved word in Python.

We can now extract the prices:

```python
for price in prices:
    print(price.text)
```

Similarly, we can search for elements using other attributes.

For example, an HTML link may look like this:

```html
<a href="book.html">View Book</a>
```

We can retrieve the link using:

```python
link = soup.find("a")

print(link["href"])
```

Here, `href` is an HTML attribute that stores the link destination.

## 5.6 Extracting Information from Nested Elements

Sometimes, the information we want is not directly inside the element we first locate.

Consider this HTML:

```html
<article class="product_pod">
    <h3>
        <a href="book.html" title="Python for Beginners">
            Python for Beginners
        </a>
    </h3>

    <p class="price_color">£25.99</p>
</article>
```

The book title is stored inside the `title` attribute of the `a` element.

The price is stored inside a paragraph with the class `price_color`.

We can locate the article first:

```python
book = soup.find("article", class_="product_pod")
```

Then retrieve the nested information:

```python
title = book.h3.a["title"]

price = book.find("p", class_="price_color").text

print(title)
print(price)
```

Why is this approach useful?

A website may contain many records with the same structure.

By locating one parent element first, we can extract information belonging to that specific record.

For example, each book has its own title, price and rating.

This prevents us from accidentally mixing the title of one book with the price of another.

# 6. Collecting Multiple Records

Now that we understand how to retrieve HTML and locate elements, we can combine these techniques to collect complete records.

## 6.1 Understanding the Structure of a Book Record

On Books to Scrape, each book is represented by an article element.

The article contains information such as:

- Book title.
- Price.
- Rating.
- Availability.

Each book follows a similar HTML structure.

This means we can locate all the book articles and extract information from each one.

First, find all book containers:

```python
books = soup.find_all("article", class_="product_pod")
```

Now inspect how many books were found:

```python
print(len(books))
```

## 6.2 Extracting Book Titles and Prices

We can loop through the book elements and extract the required information.

```python
for book in books:
    title = book.h3.a["title"]

    price = book.find("p", class_="price_color").text

    print(title, price)
```

Let's understand what happens:

1. `for book in books` processes one book at a time.
2. `book.h3.a["title"]` extracts the title from the link attribute.
3. `book.find()` locates the price element inside the current book.
4. `.text` extracts the displayed price.

We are now collecting information from multiple records automatically.

## 6.3 Storing the Extracted Data

Printing the information is useful for checking our results, but we usually want to store the data for further analysis.

Python lists are useful for collecting records.

We can create an empty list:

# 8. Scraping Multiple Pages

So far, we have collected information from the first page of the bookstore.

But what happens when a website contains multiple pages?

This is called **pagination**.

Pagination is the process of dividing information into separate pages rather than displaying all records on one page.

For example:

- Page 1 contains books 1–20.
- Page 2 contains books 21–40.
- Page 3 contains books 41–60.

If we only scrape the first page, we will miss the remaining records.

We therefore need to visit multiple pages.

On Books to Scrape, the next page follows a URL pattern such as:

```text
https://books.toscrape.com/catalogue/page-2.html
```

We can use a loop to retrieve information from several pages.

```python
all_books = []

for page in range(1, 4):

    if page == 1:
        url = "https://books.toscrape.com/"
    else:
        url = f"https://books.toscrape.com/catalogue/page-{page}.html"

    response = requests.get(url)

    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("article", class_="product_pod")

    for book in books:

        title = book.h3.a["title"]

        price = book.find("p", class_="price_color").text

        all_books.append({
            "title": title,
            "price": price
        })
```

Convert the results into a DataFrame:

```python
df = pd.DataFrame(all_books)

df.head()
```

We have now collected records from three pages.

**Important:** This example uses a known page URL pattern. Real websites may use different pagination systems, so always inspect how the website organizes its pages.

# 9. Responsible Web Scraping

Web scraping is a powerful data collection technique, but it must be used responsibly.

Before scraping a website, consider the following:

1. **Check website permissions:** Review the website's terms of service and scraping policies.
2. **Check robots.txt:** This file communicates crawling preferences for automated agents. It is not, by itself, legal permission to scrape.
3. **Avoid excessive requests:** Sending too many requests can overload a website's server.
4. **Respect access restrictions:** Do not bypass authentication, paywalls, CAPTCHAs or other technical controls.
5. **Protect personal information:** Avoid collecting private or sensitive information without a lawful basis and appropriate authorization.
6. **Use APIs where possible:** If a suitable API is available, it is usually a better option than scraping the website.

A responsible data analyst considers not only whether data can be collected, but also whether it should be collected.

# 10. API Access vs Web Scraping: When Should You Use Each?

Now that we have explored both techniques, let's compare when they are useful.

| Situation | Recommended approach |
|---|---|
| A service provides an API for retrieving exchange rates | API |
| You want to retrieve country population data from a public API | API |
| A website displays product prices but provides no suitable API | Web scraping, if permitted |
| You want to collect information from a website with a documented API | API |
| You need to retrieve structured information from an online service | API |
| You want to automate the collection of publicly displayed information from a permitted website | Web scraping |

The main lesson is that the choice depends on where the information is available and how the source allows it to be accessed.
