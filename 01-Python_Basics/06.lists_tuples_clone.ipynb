## 2. Lists

A list is an ordered and changeable collection of elements. Lists allow duplicate values and can store different data types.

### Creating a List

Lists are created using square brackets `[]`.

```python
students = ["Mary", "Brian", "John", "Faith"]

scores = [85, 90, 75, 85]

mixed = ["Mary", 22, True, 85.5]
```

### Characteristics of Lists

- **Ordered:** Elements maintain their insertion order.
- **Mutable:** Elements can be added, removed, or changed.
- **Allow duplicates:** The same value can appear more than once.
- **Indexed:** Elements can be accessed using their positions.
- **Allow different data types:** A list can contain strings, numbers, booleans, and other objects.

**Teaching question:** If we have a list containing five students, how can we access the third student?

### Accessing List Elements

Each element has an index starting from `0`.

```python
students = ["Mary", "Brian", "John", "Faith"]

print(students[0])   # Mary
print(students[2])   # John
print(students[-1])  # Faith
```

Explain positive and negative indexing:

- Positive indexing starts at `0` from the beginning.
- Negative indexing starts at `-1` from the end.

**Quick challenge:** Given the following list, ask students to predict the output.

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1])
print(numbers[-2])
```

**Instructor solution:**

```text
20
40
```

### List Slicing

Slicing allows us to access a portion of a list.

Syntax:

```python
list_name[start:stop:step]
```

The stop position is excluded.

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
print(numbers[:3])
print(numbers[-2:])
```

Expected output:

```text
[20, 30, 40]
[10, 20, 30]
[40, 50]
```

**Teaching question:** Why does `numbers[1:4]` return three elements instead of four?

### Modifying Lists

Because lists are mutable, their elements can be changed.

```python
students = ["Mary", "Brian", "John"]

students[1] = "Peter"

print(students)
```

Expected output:

```text
['Mary', 'Peter', 'John']
```

### Common List Methods

| Method | Purpose |
|---|---|
| `append()` | Adds one element to the end |
| `insert()` | Adds an element at a specific position |
| `extend()` | Adds multiple elements |
| `remove()` | Removes the first matching value |
| `pop()` | Removes and returns an element |
| `clear()` | Removes all elements |
| `sort()` | Sorts the list |
| `reverse()` | Reverses the list |
| `count()` | Counts occurrences of a value |
| `index()` | Returns the position of a value |

Demonstration:

```python
scores = [75, 90, 65, 80]

scores.append(95)
scores.remove(65)
scores.sort()

print(scores)
```

Expected output:

```text
[75, 80, 90, 95]
```

### Iterating Through a List

A `for` loop allows us to access each element in a list.

```python
students = ["Mary", "Brian", "John"]

for student in students:
    print(student)
```

Explain that Python accesses each student one at a time without requiring us to specify their positions.

**Ask students:** What would happen if we added another student to the list? Would we need to change the loop?

Explain that the loop automatically handles the additional element.

### Iterating Through the First Three Students Using `range()`

Sometimes we want to access only specific positions in a list.

We can use `range()` to generate indexes.

```python
students = ["Mary", "Brian", "John", "Faith", "Peter"]

for i in range(3):
    print(students[i])
```

Expected output:

```text
Mary
Brian
John
```

Break down the code:

- `range(3)` generates `0, 1, 2`.
- `i` represents the current index.
- `students[i]` accesses the student at that index.

**Challenge:** Ask students to modify the loop so that it displays only the first three students, even if the list contains fewer than three students.

**Instructor solution:**

```python
students = ["Mary", "Brian"]

for i in range(min(3, len(students))):
    print(students[i])
```

Explain:

- `len(students)` returns the number of students.
- `min(3, len(students))` chooses the smaller number.
- This prevents an `IndexError` when the list has fewer than three elements.

**Follow-up question:** What is the difference between iterating directly through a list and iterating through its indexes?

Expected understanding:
- `for student in students` gives us the actual elements.
- `for i in range(len(students))` gives us the indexes, which we can use to access or modify elements by position.

### List Comprehension

List comprehension provides a shorter way to create a new list from an existing iterable.

```python
numbers = [1, 2, 3, 4, 5]

squared = [number ** 2 for number in numbers]

print(squared)
```

Expected output:

```text
[1, 4, 9, 16, 25]
```

Explain the structure:

```python
new_list = [expression for item in iterable]
```

**Quick challenge:** Create a new list containing only the even numbers.

```python
numbers = [1, 2, 3, 4, 5, 6]
```

**Instructor solution:**

```python
even_numbers = [number for number in numbers if number % 2 == 0]

print(even_numbers)
```

Expected output:

```text
[2, 4, 6]
```

### When Should We Use Lists?

Use lists when:

- You need to store a collection that can change.
- You need to maintain the order of elements.
- You need to access elements using indexes.
- You expect to add or remove items.

**Practical scenarios:**

- Storing a list of students in a class.
- Maintaining a shopping list.
- Recording examination scores.
- Storing daily sales figures.

## 3. Tuples

A tuple is an ordered collection of elements that cannot be changed after creation.

Tuples are created using parentheses `()`.

```python
student = ("Mary", 22, "Data Science", 85)
```

### Characteristics of Tuples

- **Ordered:** Elements maintain their insertion order.
- **Immutable:** Elements cannot be changed, added, or removed after creation.
- **Allow duplicates:** The same value can appear more than once.
- **Indexed:** Elements can be accessed using their positions.
- **Allow different data types:** A tuple can store different types of values.

### Accessing Tuple Elements

Tuples support indexing and slicing just like lists.

```python
student = ("Mary", 22, "Data Science", 85)

print(student[0])
print(student[2])
print(student[-1])
```

**Quick challenge:** Ask students to access the second and last elements of this tuple.

```python
coordinates = (10, 20, 30, 40)
```

**Instructor solution:**

```python
print(coordinates[1])
print(coordinates[-1])
```

Expected output:

```text
20
40
```

### Tuple Immutability

Consider the following example:

```python
scores = (70, 80, 90)

scores[0] = 100
```

**Ask students:** What do you think will happen when this code runs?

Explain that the operation raises a `TypeError` because tuples cannot be modified after creation.

### Tuple Methods

Tuples have fewer built-in methods because they are immutable.

| Method | Purpose |
|---|---|
| `count()` | Counts occurrences of a value |
| `index()` | Returns the position of a value |

```python
scores = (70, 80, 90, 80)

print(scores.count(80))
print(scores.index(90))
```

Expected output:

```text
2
2
```

### Tuple Packing and Unpacking

**Packing** means placing multiple values into a tuple.

```python
student = "Mary", 22, "Data Science"
```

**Unpacking** means assigning tuple elements to separate variables.

```python
name, age, course = student

print(name)
print(age)
print(course)
```

A tuple with one element must include a comma.

```python
single_value = (10,)
```

Without the comma, Python treats `(10)` as an ordinary integer expression.

**Quick challenge:** Create a tuple containing a student's name, age, and course. Unpack the values into three variables and print them separately.

**Instructor solution:**

```python
student = ("Brian", 24, "Data Analytics")

name, age, course = student

print(name)
print(age)
print(course)
```

### When Should We Use Tuples?

Use tuples when:

- You need an ordered collection that should not be modified.
- You want to group related values together.
- You are representing fixed information.

**Practical scenarios:**

- Storing GPS coordinates `(latitude, longitude)`.
- Representing an RGB colour `(255, 0, 0)`.
- Storing fixed configuration values.
- Returning multiple values from a function.

**Teaching question:** If a student's examination score needs to be updated regularly, would a tuple be the best choice? Why?

Expected understanding: A list or dictionary would be more appropriate because the data needs to change.

## 4. Sets

A set is a collection of unique elements. Sets are useful when working with data where duplicate values are not needed.

Sets are created using curly brackets `{}` or the `set()` constructor.

```python
numbers = {10, 20, 30, 40}

unique_scores = {80, 90, 80, 70}

print(unique_scores)
```

Explain that duplicate values are automatically removed. The order in which elements appear when printed is not guaranteed.

### Characteristics of Sets

- **Unordered:** Sets do not guarantee element order.
- **Mutable:** Elements can be added or removed.
- **No duplicates:** Every element must be unique.
- **Not indexed:** Elements cannot be accessed using positions.
- **Elements must be hashable:** Mutable collections such as lists and dictionaries cannot be set elements.

### Creating an Empty Set

An empty set is created using `set()`.

```python
empty_set = set()
```

Compare:

```python
empty_dictionary = {}
empty_set = set()
```

Explain that an empty pair of curly brackets creates a dictionary, not a set.

### Adding and Removing Elements

```python
courses = {"Python", "SQL", "Power BI"}

courses.add("Machine Learning")

courses.remove("SQL")

print(courses)
```

Other useful set methods:

| Method | Purpose |
|---|---|
| `add()` | Adds one element |
| `update()` | Adds multiple elements |
| `remove()` | Removes an element; raises an error if absent |
| `discard()` | Removes an element without raising an error if absent |
| `clear()` | Removes all elements |

### Set Operations

Sets support mathematical operations that are useful when comparing collections of data.

Start with two sets:

```python
python_students = {"Mary", "John", "Brian"}
sql_students = {"John", "Faith", "Brian"}
```

**Union:** Returns all unique elements from both sets.

```python
print(python_students.union(sql_students))
```

**Intersection:** Returns elements found in both sets.

```python
print(python_students.intersection(sql_students))
```

**Difference:** Returns elements found in the first set but not the second.

```python
print(python_students.difference(sql_students))
```

**Symmetric Difference:** Returns elements found in either set, but not both.

```python
print(python_students.symmetric_difference(sql_students))
```

**Ask students:** Which operation would you use to identify students enrolled in both Python and SQL?

Expected answer: Intersection.

Use sets when:

- You want to remove duplicates.

## 5. Dictionaries

A dictionary stores data in key-value pairs. Each key is used to access its corresponding value.

Dictionaries are created using curly brackets `{}`.

```python
student = {
    "name": "Mary",
    "age": 22,
    "course": "Data Science",
    "score": 85
}
```

Explain that instead of accessing information using numerical indexes, we use meaningful keys.

### Characteristics of Dictionaries

- **Key-value structure:** Each item contains a key and its associated value.
- **Ordered:** Dictionaries preserve insertion order.
- **Mutable:** Elements can be added, updated, or removed.
- **Unique keys:** Duplicate keys are not allowed; assigning a value to an existing key replaces its previous value.
- **Values can be duplicated:** Different keys can have the same value.
- **Keys must be hashable:** Strings, numbers, and tuples containing hashable elements can be used as keys.

### Accessing Dictionary Values

Values are accessed using their keys rather than numerical indexes.

```python
student = {
    "name": "Mary",
    "age": 22,
    "course": "Data Science"
}

print(student["name"])
print(student["course"])
```

You can also use `get()`:

```python
print(student.get("age"))
print(student.get("score", "Not available"))
```

Explain the difference:

- `student["score"]` raises a `KeyError` if the key does not exist.
- `student.get("score")` returns `None` if the key does not exist.
- `student.get("score", "Not available")` returns the specified default value.

### Adding and Updating Dictionary Items

```python
student = {
    "name": "Mary",
    "age": 22
}

student["course"] = "Data Science"

student["age"] = 23

print(student)
```

Explain that assigning a value to a new key adds an item, while assigning a value to an existing key updates it.

### Removing Dictionary Items

```python
student = {
    "name": "Mary",
    "age": 22,
    "course": "Data Science"
}

student.pop("age")

del student["course"]

print(student)
```

### Common Dictionary Methods

| Method | Purpose |
|---|---|
| `keys()` | Returns all keys |
| `values()` | Returns all values |
| `items()` | Returns key-value pairs |
| `get()` | Retrieves a value using a key |
| `update()` | Adds or updates multiple items |
| `pop()` | Removes an item using its key |
| `clear()` | Removes all items |

### Iterating Through a Dictionary

```python
student = {
    "name": "Mary",
    "age": 22,
    "course": "Data Science"
}

for key, value in student.items():
    print(key, value)
```

Explain that `items()` provides both the key and value for each dictionary entry.

**Teaching question:** Why might a dictionary be more useful than a list when storing a student's personal details?

Expected understanding: The keys provide meaningful labels, making it easier to retrieve specific information.

## 6. Comparing the Four Data Structures

Use this comparison to reinforce the differences.

| Feature | List | Tuple | Set | Dictionary |
|---|---|---|---|---|
| Syntax | `[]` | `()` | `{}` | `{key: value}` |
| Ordered | Yes | Yes | No | Yes |
| Mutable | Yes | No | Yes | Yes |
| Allows duplicates | Yes | Yes | No | Keys: No |
| Indexed | Yes | Yes | No | Accessed by keys |
| Stores | Elements | Elements | Unique elements | Key-value pairs |

### Choosing the Right Data Structure

Discuss these scenarios with students and ask them to justify their choices.

| Scenario | Suitable Structure | Reason |
|---|---|---|
| Storing a list of student names that may change | List | Allows adding, removing, and updating names |
| Storing fixed GPS coordinates | Tuple | Keeps related values together without modification |
| Identifying unique email addresses | Set | Automatically removes duplicates |
| Finding students enrolled in two courses | Set | Supports intersection operations |
| Storing a student's name, age, and score | Dictionary | Organises information using descriptive keys |
| Storing product names and their prices | Dictionary | Associates each product with its price |
| Keeping examination scores in order | List | Supports indexing, sorting, and updating |

## 7. Integrated Practical Challenge: Student Records

Use this final activity to assess whether students can combine different data structures.

Introduce the following scenario:

You are building a simple student records system. Each student has a name, course, and score.

```python
students = [
    {"name": "Mary", "course": "Data Science", "score": 85},
    {"name": "Brian", "course": "Software Engineering", "score": 75},
    {"name": "John", "course": "Data Analytics", "score": 90},
    {"name": "Faith", "course": "Data Science", "score": 65}
]
```

Ask students to:

1. Display the names of all students.
2. Display the score of each student.
3. Calculate the average student score.
4. Identify the student with the highest score.
5. Create a set containing all unique course names.
6. Add a new student to the records.
7. Update Faith's score to 95.
8. Display all students who scored at least 80.

### Instructor Solutions

**1. Display all student names**

```python
for student in students:
    print(student["name"])
```

**2. Display all student scores**

```python
for student in students:
    print(student["score"])
```

**3. Calculate the average score**

```python
total = 0

for student in students:
    total += student["score"]

average = total / len(students)

print(average)
```

**4. Identify the student with the highest score**

```python
highest_student = students[0]

for student in students:
    if student["score"] > highest_student["score"]:
        highest_student = student

print(highest_student["name"])
print(highest_student["score"])
```

**5. Find unique courses**

```python
courses = set()

for student in students:
    courses.add(student["course"])

print(courses)
```

**6. Add a new student**

```python
new_student = {
    "name": "Peter",
    "course": "Data Science",
    "score": 88
}

students.append(new_student)
```

**7. Update Faith's score**

```python
for student in students:
    if student["name"] == "Faith":
        student["score"] = 95
```

**8. Display students who scored at least 80**

```python
for student in students:
    if student["score"] >= 80:
        print(student["name"], student["score"])
```

**Discussion questions:**

- Why is the outer structure a list?
- Why is each student's information stored in a dictionary?
- Why is a set useful for identifying unique courses?
- Could a tuple be used to store a student's record? What would be the limitation if we needed to update the score?

## 8. Key Takeaways

- **Lists:** Ordered, mutable collections that allow duplicates.
- **Tuples:** Ordered, immutable collections that allow duplicates.
- **Sets:** Mutable collections of unique elements.
- **Dictionaries:** Mutable collections of key-value pairs with unique keys.
- The appropriate data structure depends on how data needs to be organised, accessed, and modified.

The goal is not just to memorise the syntax, but to understand why and when to use each data structure.
