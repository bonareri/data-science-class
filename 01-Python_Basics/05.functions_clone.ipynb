# 1. Introduction to Functions

A function is a reusable block of code designed to perform a specific task.

Instead of writing the same code repeatedly, we define a function once and call it whenever we need it.

Think of a function as a machine:

- You provide an input.
- The function performs an operation.
- It may produce an output.

### Example

Suppose you are analysing student marks and want to calculate the average score.

Instead of writing the calculation repeatedly, you can create a function.

```python
def calculate_average():
    marks = [70, 80, 90]
    average = sum(marks) / len(marks)
    print(average)

calculate_average()
```

### Understanding the Code

- `def` – keyword used to define a function.
- `calculate_average` – name of the function.
- `()` – holds parameters when needed.
- `:` – indicates the beginning of the function body.
- Indentation – identifies the statements belonging to the function.
- `calculate_average()` – calls or executes the function.

### Why Are Functions Important?

1. **Code reusability:** Write once and use multiple times.
2. **Organisation:** Break complex programs into smaller tasks.
3. **Readability:** Make code easier to understand.
4. **Maintainability:** Update a function in one place instead of changing repeated code.
5. **Testing:** Individual functions can be tested independently.

# 4. User-Defined Functions

User-defined functions are functions that programmers create to perform specific tasks.

Unlike built-in functions, which Python already provides, user-defined functions are created using the `def` keyword.

### Syntax

```python
def function_name(parameters):
    # Statements to execute
    return result
```

### Example: Creating a Function to Calculate BMI

BMI (Body Mass Index) is calculated using:

\[
BMI = \frac{weight(kg)}{height(m)^2}
\]

```python
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi

result = calculate_bmi(70, 1.75)

print("BMI:", round(result, 2))
```

### What Happens When We Call the Function?

1. Python receives the values `70` and `1.75`.
2. The function uses them to calculate BMI.
3. The `return` statement sends the calculated result back.
4. The result is stored in the variable `result`.
5. We display the result using `print()`.

The same function can be reused with different values:

```python
print(calculate_bmi(65, 1.60))
print(calculate_bmi(80, 1.80))
print(calculate_bmi(90, 1.75))
```

# 5. Parameters and Arguments

Parameters and arguments are related, but they are not the same.

**Parameters** are the variable names specified inside the parentheses when defining a function.

**Arguments** are the actual values supplied when calling the function.

### Example

```python
def greet_student(name):
    print("Hello", name)

greet_student("Mary")
```

- `name` is the parameter.
- `"Mary"` is the argument.

## Types of Arguments

### 1. Positional Arguments

Values are assigned based on their position.

```python
def student_details(name, age):
    print(name, age)

student_details("Brian", 22)
```

Brian is assigned to `name`, while 22 is assigned to `age`.

### 2. Keyword Arguments

Values are assigned using parameter names.

```python
def student_details(name, age):
    print(name, age)

student_details(age=22, name="Brian")
```

The order does not matter when using keyword arguments.

### 3. Default Arguments

A parameter can have a default value that is used when no argument is supplied.

```python
def greet_student(name, course="Data Science"):
    print(name, "is studying", course)

greet_student("Mary")
greet_student("Brian", "Software Engineering")
```

### 4. Variable-Length Arguments

Sometimes we do not know how many arguments a function will receive.

Python provides `*args` and `**kwargs`.

**`*args`**

Allows a function to receive multiple positional arguments.

```python
def calculate_total(*numbers):
    return sum(numbers)

print(calculate_total(10, 20, 30))
print(calculate_total(5, 10, 15, 20, 25))
```

**`**kwargs`**

Allows a function to receive multiple keyword arguments.

```python
def student_details(**details):
    print(details)

student_details(name="Mary", age=22, course="Data Science")
```

The information is received as a dictionary.

# 6. Fruitful and Void Functions

Functions can also be classified based on whether they return a value.

## 6.1 Fruitful Functions

A fruitful function is a function that returns a value using the `return` statement.

The returned value can be stored in a variable, used in calculations, or passed to another function.

### Example

```python
def calculate_total(price, quantity):
    total = price * quantity
    return total

amount = calculate_total(500, 3)

print("Total amount:", amount)
```

Output:

```text
Total amount: 1500
```

The function calculates the total and returns it to the calling code.

## 6.2 Void Functions

A void function performs a task but does not explicitly return a value.

It may display information, print a message, or perform an operation.

### Example

```python
def display_message():
    print("Welcome to Python Programming!")

display_message()
```

A function without a `return` statement implicitly returns `None`.

### Difference Between Fruitful and Void Functions

| Fruitful Function | Void Function |
|---|---|
| Returns a value | Does not explicitly return a value |
| Uses `return` to provide a result | Often performs an action |
| Result can be stored | Usually called to perform a task |
| Useful for calculations | Useful for displaying or performing actions |

### Important Distinction: `print()` vs `return`

`print()` and `return` are not the same.

- `print()` displays information on the screen.
- `return` sends a value back to the part of the program that called the function.

### Example

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(10, 20)

print(result)
```

# 7. Lambda Functions

A lambda function is a small anonymous function defined using the `lambda` keyword.

Unlike regular functions, lambda functions are written in a single expression and do not require a function name.

### Syntax

```python
lambda arguments: expression
```

### Example

Regular function:

```python
def square(number):
    return number ** 2

print(square(5))
```

Equivalent lambda function:

```python
square = lambda number: number ** 2

print(square(5))
```

Both produce the same result.

## Lambda with Multiple Arguments

```python
add = lambda a, b: a + b

print(add(10, 20))
```

## Lambda with Sorting

Lambda functions are particularly useful when working with sorting and data processing.

Suppose we have a list of students and their marks:

```python
students = [
    ("Brian", 75),
    ("Mary", 90),
    ("John", 65),
    ("Faith", 85)
]

students.sort(key=lambda student: student[1])

print(students)
```

The lambda function tells Python to sort the students using their marks rather than their names.

## Lambda with `map()`

The `map()` function applies a function to every item in an iterable.

```python
numbers = [1, 2, 3, 4, 5]

squared = list(map(lambda x: x ** 2, numbers))

print(squared)
```

Output:

```text
[1, 4, 9, 16, 25]
```

## Lambda with `filter()`

The `filter()` function selects items that satisfy a condition.

```python
numbers = [10, 15, 20, 25, 30, 35]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

Output:

```text
[10, 20, 30]
```

### When Should We Use Lambda Functions?

- For short operations.
- When passing a simple function to another function.
- When sorting data using a particular column.
- When using functions such as `map()` and `filter()`.

Avoid using lambda functions for complex operations. A regular function is usually easier to read and maintain.

# 8. Recursive Functions

A recursive function is a function that calls itself to solve a problem.

Instead of solving the entire problem at once, recursion breaks it into smaller versions of the same problem.

Every recursive function needs two important components:

1. **Base case:** The condition that stops the recursion.
2. **Recursive case:** The part where the function calls itself.

## Example: Calculating Factorial

A factorial is the product of all positive integers up to a given number.

For example:

\[
5! = 5 \times 4 \times 3 \times 2 \times 1 = 120
\]

We can calculate factorial using recursion:

```python
def factorial(n):

    if n == 0 or n == 1:
        return 1

    else:
        return n * factorial(n - 1)


print(factorial(5))
```

Output:

```text
120
```

### How Does the Recursion Work?

```text
factorial(5)
    5 * factorial(4)
        4 * factorial(3)
            3 * factorial(2)
                2 * factorial(1)
                    1
```

Python works backwards to calculate the final result.

## Example: Countdown

```python
def countdown(number):

    if number == 0:
        print("Done!")

    else:
        print(number)
        countdown(number - 1)


countdown(5)
```

Output:

```text
5
4
3
2
1
Done!
```

**Important:** Without a proper base case, recursion may continue indefinitely until Python raises a `RecursionError`.

### When Are Recursive Functions Useful?

- Tree traversal.
- Searching hierarchical data.
- Divide-and-conquer algorithms.
- Processing nested structures.
- Mathematical problems such as factorial and Fibonacci sequences.

For many everyday Python tasks, loops are simpler and more memory-efficient than recursion.

# 9. Try and Except (Exception Handling)

When writing programs, errors can occur during execution.

For example:

- A user enters text instead of a number.
- A program tries to divide by zero.
- A file does not exist.
- A program attempts to access an invalid list index.

These errors are called **exceptions**.

Exception handling allows us to handle errors without necessarily stopping the entire program unexpectedly.

Python uses `try` and `except` to handle exceptions.

## 9.1 The Try and Except Structure

### Syntax

```python
try:
    # Code that may produce an error

except:
    # Code that handles the error
```

### Example: Division

```python
try:
    number1 = 10
    number2 = 0

    result = number1 / number2

    print(result)

except:
    print("Something went wrong!")
```

Without exception handling, Python would raise a `ZeroDivisionError`.

With exception handling, the program displays a message instead.

## 9.2 Handling Specific Exceptions

It is better to catch specific exceptions rather than using a general `except` statement.

### Example 1: ZeroDivisionError

```python
try:
    number1 = 10
    number2 = 0

    result = number1 / number2

    print(result)

except ZeroDivisionError:
    print("You cannot divide by zero!")
```

### Example 2: ValueError

A `ValueError` occurs when a value has the correct data type requirement but an invalid value for the operation.

```python
try:
    age = int(input("Enter your age: "))

    print("Your age is:", age)

except ValueError:
    print("Please enter a valid number.")
```

If someone enters `"hello"`, Python cannot convert it into an integer.

## 9.3 Multiple Except Blocks

A program can encounter different types of exceptions. We can handle each separately.

```python
try:
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))

    result = number1 / number2

    print("Result:", result)

except ValueError:
    print("Please enter numbers only.")

except ZeroDivisionError:
    print("You cannot divide by zero.")
```

## 9.4 The Else Block

The `else` block executes only when no exception occurs in the `try` block.

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

else:
    print("You entered:", number)
```

## 9.5 The Finally Block

The `finally` block executes whether an exception occurs or not.

It is commonly used for cleanup operations, such as closing files or releasing resources.

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

finally:
    print("Program execution completed.")
```

## 9.6 Raising Exceptions

Python also allows programmers to deliberately raise exceptions using the `raise` keyword.

```python
def check_age(age):

    if age < 0:
        raise ValueError("Age cannot be negative.")

    return age

print(check_age(-5))
```

This helps us enforce rules and validate data.

### Summary of Exception Handling Keywords

| Keyword | Purpose |
|---|---|
| `try` | Contains code that might raise an exception |
| `except` | Handles an exception |
| `else` | Runs when no exception occurs |
| `finally` | Runs whether an exception occurs or not |
| `raise` | Deliberately raises an exception |

# 10. Combining Functions with Data Analysis

Functions become particularly useful when working with datasets because they allow us to automate repetitive calculations and data-processing tasks.

### Example: Creating a Reusable Data Analysis Function

Suppose we have examination marks for different students. We want to calculate the average, highest and lowest marks.

```python
def analyse_marks(marks):

    average = sum(marks) / len(marks)
    highest = max(marks)
    lowest = min(marks)

    print("Average:", round(average, 2))
    print("Highest:", highest)
    print("Lowest:", lowest)


class_a = [70, 85, 90, 65, 80]
class_b = [60, 75, 88, 92, 70]

analyse_marks(class_a)

print("----------------")

analyse_marks(class_b)
```

Notice that we only write the logic once but can reuse it with different datasets.

This is one of the reasons functions are important in Data Science.
