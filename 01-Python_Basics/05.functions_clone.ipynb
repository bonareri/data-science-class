# 1. Introduction to Functions

## Scenario: Calculating Student Marks

Imagine you are working with student examination results. You need to calculate the average marks for different students.

You could write the calculation every time you need it:

```python
marks = [70, 80, 90]

average = sum(marks) / len(marks)

print(average)
```

But what happens when you need to calculate the average for another class, another group of students, or another dataset?

Would you keep writing the same code?

A better approach is to create a **function**.

## What Is a Function?

A function is a reusable block of code designed to perform a specific task.

Instead of writing the same instructions repeatedly, we define them once and call the function whenever we need them.

Functions help us organise programs into smaller, manageable and reusable pieces.

## 1.1 Function Syntax

Python uses the `def` keyword to define a function.

```python
def function_name():
    # Instructions to execute
```

Consider this example:

```python
def calculate_average():
    marks = [70, 80, 90]
    average = sum(marks) / len(marks)
    print(average)
```

Notice that defining a function does not automatically execute its instructions.

To execute it, we must **call the function**:

```python
calculate_average()
```

### Understanding the Syntax

| Component | Purpose |
|---|---|
| `def` | Keyword used to define a function |
| `calculate_average` | Name of the function |
| `()` | Holds parameters when needed |
| `:` | Marks the beginning of the function body |
| Indentation | Identifies the instructions belonging to the function |
| `calculate_average()` | Calls or executes the function |

## 1.2 Functions Without Parameters

A function without parameters does not receive values from the code calling it.

The function uses information already available inside its body or elsewhere in the program.

```python
def greet():
    print("Welcome to Python Programming!")

greet()
```

**Output:**

```text
Welcome to Python Programming!
```

The function does not require any additional information to perform its task.

## 1.3 Functions With Parameters

Sometimes a function needs information to perform a task.

We can provide that information through **parameters**.

```python
def greet_student(name):
    print("Welcome", name)

greet_student("Mary")
greet_student("Brian")
```

**Output:**

```text
Welcome Mary
Welcome Brian
```

Here, `name` is a parameter. It allows the function to work with different values.

We can also use multiple parameters:

```python
def calculate_area(length, width):
    area = length * width
    print("Area:", area)

calculate_area(5, 4)
calculate_area(10, 6)
```

**Output:**

```text
Area: 20
Area: 60
```

### Think About It

What is the main difference between these two functions?

```python
def greet():
    print("Hello")
```

```python
def greet(name):
    print("Hello", name)
```

The first function does not require an input. The second function expects a value to be provided when it is called.

## 1.4 Parameters vs Arguments

These two terms are related, but they are not the same.

- **Parameters** are variable names specified when defining a function.
- **Arguments** are actual values supplied when calling the function.

Example:

```python
def calculate_square(number):
    return number ** 2

calculate_square(5)
```

- `number` is the parameter.
- `5` is the argument.

## 1.5 Types of Arguments

### Positional Arguments

Values are assigned according to their position.

```python
def student_details(name, age):
    print("Name:", name)
    print("Age:", age)

student_details("Brian", 22)
```

Brian is assigned to `name`, while 22 is assigned to `age`.

### Keyword Arguments

Values are assigned using parameter names.

```python
def student_details(name, age):
    print("Name:", name)
    print("Age:", age)

student_details(age=22, name="Brian")
```

The order does not matter when using keyword arguments.

### Default Arguments

A parameter can have a default value that is used when no argument is supplied.

```python
def greet_student(name, course="Data Science"):
    print(name, "is studying", course)

greet_student("Mary")
greet_student("Brian", "Software Engineering")
```

### Variable-Length Arguments

Sometimes we do not know how many arguments a function will receive.

Python provides `*args` and `**kwargs`.

**`*args`**

Allows a function to receive multiple positional arguments.

```python
def calculate_total(*numbers):
    return sum(numbers)

print(calculate_total(10, 20, 30))
print(calculate_total(5, 10, 15, 20))
```

**`**kwargs`**

Allows a function to receive multiple keyword arguments.

```python
def student_details(**details):
    print(details)

student_details(name="Mary", age=22, course="Data Science")
```

The information is received as a dictionary.

# 2. Returning Values: `print()` vs `return`

A function can perform a task and display information, or it can calculate something and send the result back to the part of the program that called it.

Understanding this distinction is important.

## 2.1 Using `print()`

`print()` displays information on the screen.

```python
def calculate_total():
    total = 200 + 300
    print(total)

calculate_total()
```

The result is displayed, but the function does not return the calculated value for further use.

## 2.2 Using `return`

The `return` statement sends a value back to the calling code.

```python
def calculate_total():
    total = 200 + 300
    return total

result = calculate_total()

print(result)
```

We can now store the result and use it in another calculation:

```python
final_amount = calculate_total() + 100

print(final_amount)
```

## 2.3 Fruitful and Void Functions

Functions can be classified according to whether they return a value.

| Fruitful Functions | Void Functions |
|---|---|
| Return a value using `return` | Do not explicitly return a value |
| Results can be stored in variables | Usually perform an action |
| Can be used in other calculations | Often display information or perform a task |

**Fruitful function:**

```python
def calculate_discount(price, discount):
    discount_amount = price * (discount / 100)
    final_price = price - discount_amount

    return final_price

result = calculate_discount(5000, 15)

print("Final price:", result)
```

**Void function:**

```python
def display_message():
    print("Welcome to Python Programming!")

display_message()
```

A function without an explicit `return` statement automatically returns `None`.

**Remember:**

- `print()` displays information.
- `return` sends a value back.
- A function can use both, but they perform different roles.

# 5. Lambda Functions

A lambda function is a small anonymous function defined using the `lambda` keyword.

Unlike regular functions, lambda functions are written in a single expression and do not require a function name in their definition.

## 5.1 Syntax

```python
lambda arguments: expression
```

### Example: Squaring a Number

Using a regular function:

```python
def square(number):
    return number ** 2

print(square(5))
```

Using a lambda function:

```python
square = lambda number: number ** 2

print(square(5))
```

Both produce the same result.

## 5.2 Lambda With Multiple Arguments

```python
add = lambda a, b: a + b

print(add(10, 20))
```

## 5.3 Lambda With Sorting

Suppose we have a list of students and their marks:

```python
students = [
    ("Brian", 75),
    ("Mary", 90),
    ("John", 65),
    ("Faith", 85)
]

students.sort(key=lambda student: student[1])

print(students)
```

The lambda function tells Python to sort the students using their marks rather than their names.

## 5.4 Lambda With `map()`

The `map()` function applies a function to every item in an iterable.

```python
numbers = [1, 2, 3, 4, 5]

squared = list(map(lambda x: x ** 2, numbers))

print(squared)
```

**Output:**

```text
[1, 4, 9, 16, 25]
```

## 5.5 Lambda With `filter()`

The `filter()` function selects items that satisfy a condition.

```python
numbers = [10, 15, 20, 25, 30, 35]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

**Output:**

```text
[10, 20, 30]
```

### When Should We Use Lambda Functions?

- For short operations.
- When passing a simple function to another function.
- When sorting data using a particular column.
- When using functions such as `map()` and `filter()`.

Avoid using lambda functions for complex operations. A regular function is usually easier to read and maintain.

# 6. Recursive Functions

A recursive function is a function that calls itself to solve a problem.

Instead of solving the entire problem at once, recursion breaks it into smaller versions of the same problem.

Every recursive function needs two important components:

1. **Base case:** The condition that stops the recursion.
2. **Recursive case:** The part where the function calls itself.

## 6.1 Example: Calculating Factorial

A factorial is the product of all positive integers up to a given number.

For example:

\[
5! = 5 \times 4 \times 3 \times 2 \times 1 = 120
\]

We can calculate factorial using recursion:

```python
def factorial(n):

    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)


print(factorial(5))
```

**Output:**

```text
120
```

## 6.2 How Does the Recursion Work?

When we call `factorial(5)`, Python makes smaller function calls:

```text
factorial(5)
    5 * factorial(4)
        4 * factorial(3)
            3 * factorial(2)
                2 * factorial(1)
                    1
```

Once the base case is reached, Python works backwards to calculate the final result.

## 6.3 Example: Countdown

```python
def countdown(number):

    if number == 0:
        print("Done!")

    else:
        print(number)
        countdown(number - 1)


countdown(5)
```

**Output:**

```text
5
4
3
2
1
Done!
```

**Important:** Without a proper base case, recursion may continue until Python raises a `RecursionError`.

### When Are Recursive Functions Useful?

- Tree traversal.
- Searching hierarchical data.
- Divide-and-conquer algorithms.
- Processing nested structures.
- Mathematical problems such as factorial and Fibonacci sequences.

For many everyday Python tasks, loops are simpler and more memory-efficient than recursion.

# 7. Exception Handling

When writing programs, errors can occur during execution.

For example:

- A user enters text instead of a number.
- A program tries to divide by zero.
- A file does not exist.
- A program attempts to access an invalid list index.

These errors are called **exceptions**.

Exception handling allows us to handle errors without necessarily stopping the entire program unexpectedly.

## 7.1 The `try` and `except` Structure

Python uses `try` and `except` to handle exceptions.

### Syntax

```python
try:
    # Code that may produce an error

except:
    # Code that handles the error
```

### Example: Division

```python
try:
    number1 = 10
    number2 = 0

    result = number1 / number2

    print(result)

except:
    print("Something went wrong!")
```

Without exception handling, Python would raise a `ZeroDivisionError`.

With exception handling, the program displays a message instead.

## 7.2 Handling Specific Exceptions

It is better to catch specific exceptions rather than using a general `except` statement.

### Example 1: ZeroDivisionError

```python
try:
    number1 = 10
    number2 = 0

    result = number1 / number2

    print(result)

except ZeroDivisionError:
    print("You cannot divide by zero!")
```

### Example 2: ValueError

A `ValueError` occurs when a value is unsuitable for an operation.

```python
try:
    age = int(input("Enter your age: "))

    print("Your age is:", age)

except ValueError:
    print("Please enter a valid number.")
```

If someone enters `"hello"`, Python cannot convert it into an integer.

## 7.3 Multiple Except Blocks

A program can encounter different types of exceptions. We can handle each separately.

```python
try:
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))

    result = number1 / number2

    print("Result:", result)

except ValueError:
    print("Please enter numbers only.")

except ZeroDivisionError:
    print("You cannot divide by zero.")
```

## 7.4 The `else` Block

The `else` block executes only when no exception occurs in the `try` block.

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

else:
    print("You entered:", number)
```

## 7.5 The `finally` Block

The `finally` block executes whether an exception occurs or not.

It is commonly used for cleanup operations, such as closing files or releasing resources.

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

finally:
    print("Program execution completed.")
```

## 7.6 Raising Exceptions

Python also allows programmers to deliberately raise exceptions using the `raise` keyword.

```python
def check_age(age):

    if age < 0:
        raise ValueError("Age cannot be negative.")

    return age


print(check_age(-5))
```

This helps us enforce rules and validate data.

### Summary of Exception Handling Keywords

| Keyword | Purpose |
|---|---|
| `try` | Contains code that might raise an exception |
| `except` | Handles an exception |
| `else` | Runs when no exception occurs |
| `finally` | Runs whether an exception occurs or not |
| `raise` | Deliberately raises an exception |

# 8. Combining Functions With Data Analysis

Functions become particularly useful when working with datasets because they allow us to automate repetitive calculations and data-processing tasks.

### Example: Creating a Reusable Data Analysis Function

Suppose we have examination marks for different students. We want to calculate the average, highest and lowest marks.

```python
def analyse_marks(marks):

    average = sum(marks) / len(marks)
    highest = max(marks)
    lowest = min(marks)

    print("Average:", round(average, 2))
    print("Highest:", highest)
    print("Lowest:", lowest)


class_a = [70, 85, 90, 65, 80]
class_b = [60, 75, 88, 92, 70]

analyse_marks(class_a)

print("----------------")

analyse_marks(class_b)
```

Notice that we only write the logic once but can reuse it with different datasets.

This is one of the reasons functions are important in Data Science.
