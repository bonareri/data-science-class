# 1. Introduction to Functions

## The Problem: Repeating the Same Instructions

Imagine you are building a Student Performance Management System.

You have 100 students, and you need to calculate each student's average mark.

For one student, you could write:

```python
marks = [70, 80, 90]

average = sum(marks) / len(marks)

print("Average:", average)
```

But what happens when you have 100 students?

Would you write the same calculation 100 times?

That would make your program unnecessarily long, repetitive, and difficult to maintain.

Python provides a better approach: **functions**.

## What Is a Function?

A function is a reusable block of code designed to perform a specific task.

Instead of writing the same instructions repeatedly, we define them once and call the function whenever we need them.

Think of a function as a machine:

1. It can receive input.
2. It performs an operation.
3. It can produce an output.

### Example: Creating Your First Function

```python
def calculate_average():
    marks = [70, 80, 90]
    average = sum(marks) / len(marks)
    print("Average:", average)


calculate_average()
```

### Understanding the Code

- `def` is the keyword used to define a function.
- `calculate_average` is the function name.
- `()` holds parameters when the function needs input.
- `:` marks the beginning of the function body.
- Indentation identifies the statements that belong to the function.
- `calculate_average()` calls and executes the function.

**Important:** Defining a function does not automatically execute it. You must call it.

## Example: Creating a Function to Calculate BMI

BMI (Body Mass Index) is calculated using:

\[
BMI = \frac{weight(kg)}{height(m)^2}
\]

Instead of writing this calculation repeatedly, we can create a reusable function.

5. We display the result using `print()`.

The same function can be reused with different values:

```python
print(calculate_bmi(65, 1.60))
print(calculate_bmi(80, 1.80))
print(calculate_bmi(90, 1.75))
```

# 5. Parameters and Arguments

When creating reusable functions, we often need to provide them with different values.

This is where parameters and arguments become important.

**Parameters** are the variable names specified inside the parentheses when defining a function.

**Arguments** are the actual values supplied when calling the function.

### Example

```python
def greet_student(name):
    print("Hello", name)


greet_student("Mary")
```

In this example:

- `name` is the parameter.
- `"Mary"` is the argument.

The function can receive a different argument each time it is called.

## Types of Arguments

### 1. Positional Arguments

Values are assigned based on their position.

```python
def student_details(name, age):
    print(name, age)


student_details("Brian", 22)
```

Brian is assigned to `name`, while 22 is assigned to `age`.

### 2. Keyword Arguments

Values are assigned using parameter names.

```python
def student_details(name, age):
    print(name, age)


student_details(age=22, name="Brian")
```

The order does not matter when using keyword arguments.

### 3. Default Arguments

A parameter can have a default value that is used when no argument is supplied for it.

```python
def greet_student(name, course="Data Science"):
    print(name, "is studying", course)


greet_student("Mary")
greet_student("Brian", "Software Engineering")
```

If a course is not provided, Python uses `"Data Science"`.

### 4. Variable-Length Arguments

Sometimes we do not know how many arguments a function will receive.

Python provides `*args` and `**kwargs`.

**`*args`**

Allows a function to receive multiple positional arguments. They are collected into a tuple.

```python
def calculate_total(*numbers):
    return sum(numbers)


print(calculate_total(10, 20, 30))
print(calculate_total(5, 10, 15, 20, 25))
```

**`**kwargs`**

Allows a function to receive multiple keyword arguments. They are collected into a dictionary.

```python
def student_details(**details):
    print(details)


student_details(name="Mary", age=22, course="Data Science")
```

# 6. Fruitful and Void Functions

Functions can also be classified based on whether they return a value.

## 6.1 Fruitful Functions

A fruitful function is a function that returns a value using the `return` statement.

The returned value can be stored in a variable, used in calculations, or passed to another function.

### Example

```python
def calculate_total(price, quantity):
    total = price * quantity
    return total


amount = calculate_total(500, 3)

print("Total amount:", amount)
```

Output:

```text
Total amount: 1500
```

The function calculates the total and returns the result to the calling code.

## 6.2 Void Functions

A void function performs a task but does not explicitly return a value.

It may display information, print a message, or perform another operation.

### Example

```python
def display_message():
    print("Welcome to Python Programming!")


display_message()
```

A function without a `return` statement implicitly returns `None`.

## Important Distinction: `print()` vs `return`

These two are not the same.

- `print()` displays information on the screen.
- `return` sends a value back to the part of the program that called the function.

### Compare the Two Examples

**Function using print():**

```python
def add_numbers(a, b):
    print(a + b)


result = add_numbers(10, 20)

print("Stored result:", result)
```

**Function using return:**

```python
def add_numbers(a, b):
    return a + b


result = add_numbers(10, 20)

print("Stored result:", result)
```

# 7. Lambda Functions

A lambda function is a small anonymous function defined using the `lambda` keyword.

Unlike regular functions, lambda functions are written in a single expression and do not require a function name when created.

## Syntax

```python
lambda arguments: expression
```

### Example: Regular Function vs Lambda Function

Regular function:

```python
def square(number):
    return number ** 2


print(square(5))
```

Equivalent lambda function:

```python
square = lambda number: number ** 2

print(square(5))
```

Both produce the same result.

## Lambda with Multiple Arguments

```python
add = lambda a, b: a + b

print(add(10, 20))
```

## Lambda with Sorting

Lambda functions are particularly useful when sorting and processing data.

Suppose we have a list of students and their marks:

```python
students = [
    ("Brian", 75),
    ("Mary", 90),
    ("John", 65),
    ("Faith", 85)
]

students.sort(key=lambda student: student[1])

print(students)
```

The lambda function tells Python to sort the students using their marks rather than their names.

`student[1]` accesses the second item in each tuple, which represents the student's mark.

## Lambda with `map()`

The `map()` function applies a function to every item in an iterable.

Example:

```python
numbers = [1, 2, 3, 4, 5]

squared = list(map(lambda x: x ** 2, numbers))

print(squared)
```

Output:

```text
[1, 4, 9, 16, 25]
```

## Lambda with `filter()`

The `filter()` function selects items that satisfy a condition.

Example:

```python
numbers = [10, 15, 20, 25, 30, 35]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

Output:

```text
[10, 20, 30]
```

# 8. Recursive Functions

A recursive function is a function that calls itself to solve a problem.

Instead of solving the entire problem at once, recursion breaks it into smaller versions of the same problem.

Every recursive function needs two important components:

1. **Base case:** The condition that stops the recursion.
2. **Recursive case:** The part where the function calls itself.

## Example: Calculating Factorial

A factorial is the product of all positive integers up to a given number.

For example:

\[
5! = 5 \times 4 \times 3 \times 2 \times 1 = 120
\]

We can calculate factorial using recursion:

```python
def factorial(n):

    if n == 0 or n == 1:
        return 1

    else:
        return n * factorial(n - 1)


print(factorial(5))
```

Output:

```text
120
```

### How Does the Recursion Work?

Consider what happens when we call `factorial(5)`:

```text
factorial(5)
    5 * factorial(4)
        4 * factorial(3)
            3 * factorial(2)
                2 * factorial(1)
                    1
```

Once the base case is reached, Python works backwards to calculate the final result.

## Example: Countdown

```python
def countdown(number):

    if number == 0:
        print("Done!")

    else:
        print(number)
        countdown(number - 1)


countdown(5)
```

Output:

```text
5
4
3
2
1
Done!
```

### Try It Yourself: Calculate a Sum Using Recursion

Create a recursive function that calculates the sum of all positive integers from 1 up to a given number.

For example:

\[
1 + 2 + 3 + 4 + 5 = 15
\]

Your function should return 15 when given 5.

**Reference Solution**

```python
def calculate_sum(n):

    if n == 1:
        return 1

    return n + calculate_sum(n - 1)


print(calculate_sum(5))
```

### When Are Recursive Functions Useful?

- Tree traversal.
- Searching hierarchical data.
- Divide-and-conquer algorithms.
- Processing nested structures.
- Mathematical problems such as factorial and Fibonacci sequences.

For many everyday Python tasks, loops are simpler and more memory-efficient than recursion.

**Important:** Without a proper base case, recursion may continue until Python raises a `RecursionError`.

# 9. Try and Except: Exception Handling

## The Problem: What Happens When a User Makes a Mistake?

Imagine you are building a calculator.

You ask a user to enter two numbers:

```python
number1 = int(input("Enter the first number: "))
number2 = int(input("Enter the second number: "))

result = number1 / number2

print("Result:", result)
```

What happens if the user enters `"hello"` instead of a number?

What happens if the user enters `0` as the second number?

The program will encounter an error and stop executing the affected operation.

In real applications, users do not always provide the expected input. Programs therefore need a way to handle unexpected situations.

## What Are Exceptions?

Exceptions are errors that occur during program execution.

Examples include:

- A user enters text instead of a number.
- A program tries to divide by zero.
- A file does not exist.
- A program attempts to access an invalid list index.

**Exception handling** allows us to handle these errors without necessarily stopping the entire program unexpectedly.

Python uses `try`, `except`, `else`, and `finally` to handle exceptions.

## 9.1 The Try and Except Structure

### Syntax

```python
try:
    # Code that may produce an error

except:
    # Code that handles the error
```

### Example: Division

```python
try:
    number1 = 10
    number2 = 0

    result = number1 / number2

    print(result)

except:
    print("Something went wrong!")
```

Without exception handling, Python would raise a `ZeroDivisionError`.

With exception handling, the program displays a message instead.

## 9.2 Handling Specific Exceptions

It is better to catch specific exceptions rather than using a general `except` statement.

This allows us to respond differently depending on the type of error.

### Example 1: ZeroDivisionError

```python
try:
    number1 = 10
    number2 = 0

    result = number1 / number2

    print(result)

except ZeroDivisionError:
    print("You cannot divide by zero!")
```

### Example 2: ValueError

A `ValueError` occurs when a value has the correct data type requirement but an invalid value for the operation.

```python
try:
    age = int(input("Enter your age: "))

    print("Your age is:", age)

except ValueError:
    print("Please enter a valid number.")
```

If someone enters `"hello"`, Python cannot convert it into an integer.

## 9.4 The Else Block

The `else` block executes only when no exception occurs in the `try` block.

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

else:
    print("You entered:", number)
```

Notice that the `else` block runs only when the `try` block completes successfully.

## 9.5 The Finally Block

The `finally` block executes whether an exception occurs or not.

It is commonly used for cleanup operations, such as closing files or releasing resources.

```python
try:
    number = int(input("Enter a number: "))

except ValueError:
    print("Invalid input.")

finally:
    print("Program execution completed.")
```

## 9.6 Raising Exceptions

Python also allows programmers to deliberately raise exceptions using the `raise` keyword.

This helps us enforce rules and validate data.

```python
def check_age(age):

    if age < 0:
        raise ValueError("Age cannot be negative.")

    return age


print(check_age(-5))
```

The function deliberately raises a `ValueError` because a negative age is invalid.

### Summary of Exception Handling Keywords

| Keyword | Purpose |
|---|---|
| `try` | Contains code that might raise an exception |
| `except` | Handles an exception |
| `else` | Runs when no exception occurs |
| `finally` | Runs whether an exception occurs or not |
| `raise` | Deliberately raises an exception |

# 10. Combining Functions with Data Analysis

Functions become particularly useful when working with datasets because they allow us to automate repetitive calculations and data-processing tasks.

Imagine you have examination marks for two different classes.

You want to calculate the average, highest, and lowest marks for each class.

Without a function, you would have to repeat the same calculations for every class.

Instead, we can define one function and reuse it.

## Example: Creating a Reusable Data Analysis Function

```python
def analyse_marks(marks):

    average = sum(marks) / len(marks)
    highest = max(marks)
    lowest = min(marks)

    print("Average:", round(average, 2))
    print("Highest:", highest)
    print("Lowest:", lowest)


class_a = [70, 85, 90, 65, 80]
class_b = [60, 75, 88, 92, 70]

analyse_marks(class_a)

print("----------------")

analyse_marks(class_b)
```

Notice that we only write the logic once but can reuse it with different datasets.

This is one of the reasons functions are important in Data Science.

**Reference Solution**

```python
numbers = [12, 5, 8, 21, 30, 17, 4, 10]

squared = list(map(lambda x: x ** 2, numbers))

greater_than_10 = list(filter(lambda x: x > 10, numbers))

students = [
    ("Brian", 75),
    ("Mary", 90),
    ("John", 65),
    ("Faith", 85)
]

sorted_students = sorted(
    students,
    key=lambda student: student[1]
)

print("Squared:", squared)
print("Greater than 10:", greater_than_10)
print("Sorted students:", sorted_students)
```

**Reference Solution**

```python
def calculate_sum(n):

    if n == 1:
        return 1

    return n + calculate_sum(n - 1)


print(calculate_sum(5))
```
