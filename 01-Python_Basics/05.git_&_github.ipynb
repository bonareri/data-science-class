## Slide 4: What Is Version Control?

Version control is a system that helps you track changes made to files over time.

It allows you to:
- Keep a history of your changes.
- Return to an earlier version of your work.
- Identify what changed and when.
- Work on different versions of a project.
- Collaborate without constantly sharing files manually.

**Key idea:** Version control helps you manage the history of your work.

## Slide 5: What Is Git?

Git is a distributed version control system that tracks changes in your files.

With Git, you can:
- Track changes to your code.
- Save important versions of your project.
- Compare different versions.
- Create separate branches to experiment safely.
- Combine changes from different contributors.

Git works locally on your computer, even without an internet connection.

## Slide 6: What Is GitHub?

GitHub is an online platform that hosts Git repositories.

It allows you to:
- Store your projects online.
- Share your work with others.
- Collaborate with other developers.
- Review and discuss code.
- Showcase your projects in a portfolio.

**Think of Git as the tool that tracks changes and GitHub as a platform where you can store and collaborate on those projects.**

## Slide 7: Git vs GitHub

| Git | GitHub |
|---|---|
| Version control system | Online hosting platform |
| Runs locally on your computer | Accessible through the internet |
| Tracks changes | Stores and shares repositories |
| Supports branches and commits | Supports pull requests and collaboration |
| Can work offline | Requires internet for online features |

**Quick check:** Can you use Git without GitHub?

Yes! Git works independently on your computer.

## Slide 8: Important Git Terminology

Before using Git, understand these concepts:

- **Repository:** A project folder tracked by Git.
- **Working directory:** The files you are currently working on.
- **Staging area:** A space where you select changes to include in your next commit.
- **Commit:** A saved snapshot of selected changes.
- **Branch:** A separate line of development.
- **Remote repository:** A version of your repository hosted elsewhere, such as GitHub.

## Slide 9: Understanding the Git Workflow

Git follows a process when you save changes.

1. **Working directory:** Make changes to your files.
2. **Staging area:** Select the changes you want to save.
3. **Local repository:** Commit the selected changes.
4. **Remote repository:** Push your commits to GitHub.

The workflow:

`Working Directory → Staging Area → Local Repository → GitHub`

**Remember:** Saving a file is not the same as committing it.

## Slide 10: Installing and Configuring Git

To get started, you need Git installed on your computer.

Download Git from:

https://git-scm.com/downloads

After installation, open your terminal and configure your identity:

```bash
git config --global user.name "Your Name"
git config --global user.email "your-email@example.com"
```

Check your configuration:

```bash
git config --list
```

## Slide 11: Creating Your First Git Repository

A repository is a project folder that Git tracks.

Let's create one.

```bash
mkdir student-performance-analysis
cd student-performance-analysis
git init
```

What do these commands do?

- `mkdir` creates a new folder.
- `cd` moves into the folder.
- `git init` initializes Git in the folder.

Check your repository:

```bash
git status
```

## Slide 12: Tracking Files with Git

Git does not automatically include every new or modified file in a commit.

You first need to stage the files you want to track.

Create a file called `README.md` and add a short project description.

Check the status:

```bash
git status
```

Stage the file:

```bash
git add README.md
```

Stage all changes in the current directory:

```bash
git add .
```

**Think:** What is the difference between creating a file and staging a file?

## Slide 13: Saving Changes with Commits

A commit records a snapshot of your staged changes.

To create a commit:

```bash
git commit -m "Add project README"
```

A good commit message should explain what changed.

Examples:

- `Add data cleaning script`
- `Update project documentation`
- `Fix missing values in dataset`

Avoid vague messages such as:

- `Update`
- `Changes`
- `Final work`

**Activity:** Add another line to your README, stage the change, and create a second commit.

## Slide 14: Viewing Your Project History

Git allows you to review the history of your project.

View all commits:

```bash
git log
```

View a simplified history:

```bash
git log --oneline
```

Compare changes that have not been staged:

```bash
git diff
```

Compare staged changes:

```bash
git diff --staged
```

**Think:** Why might a data analyst need to review previous versions of a script?

## Slide 15: Understanding Branches

A branch allows you to work on a separate version of your project without directly changing the main branch.

For example, you might have:

- `main`: The stable version of your project.
- `data-cleaning`: Work on cleaning the dataset.
- `visualizations`: Work on creating charts.

Create a new branch:

```bash
git switch -c data-cleaning
```

List available branches:

```bash
git branch
```

Switch back to the main branch:

```bash
git switch main
```

**Activity:** Create a branch called `exploratory-analysis` and switch to it.

## Slide 16: Merging Branches

After completing work on a branch, you may want to combine those changes with the main branch.

First, switch to the main branch:

```bash
git switch main
```

Then merge your branch:

```bash
git merge data-cleaning
```

Merging combines the changes from one branch into another.

**Think:** Why is it useful to develop a new feature on a separate branch before merging it into the main project?

## Slide 18: Connecting Git to GitHub

You have created a local repository. Now you want to publish it on GitHub.

First, create an empty repository on GitHub.

Then connect your local repository:

```bash
git remote add origin https://github.com/USERNAME/REPOSITORY.git
```

Set the main branch:

```bash
git branch -M main
```

Push your changes:

```bash
git push -u origin main
```

Replace `USERNAME` and `REPOSITORY` with your actual GitHub username and repository name.

**Important:** The first push establishes the connection between your local branch and the remote branch.

## Slide 19: Essential Git Commands

| Command | Purpose |
|---|---|
| `git init` | Initialize a repository |
| `git status` | Check the current state of your files |
| `git add .` | Stage changes |
| `git commit -m "message"` | Save a commit |
| `git log` | View commit history |
| `git diff` | Compare unstaged changes |
| `git branch` | List branches |
| `git switch -c name` | Create and switch to a branch |
| `git switch main` | Switch to the main branch |
| `git merge name` | Merge a branch |
| `git push` | Upload commits to a remote repository |
| `git pull` | Fetch and integrate remote changes |
| `git clone URL` | Copy a remote repository locally |

## Slide 20: Collaborating Through GitHub

GitHub provides tools that make teamwork easier.

**Pull Requests**

A pull request allows you to propose changes and request that they be reviewed before merging them into another branch.

**Code Reviews**

Team members can examine proposed changes, leave comments, and suggest improvements.

**Issues**

Issues help teams document bugs, tasks, and project improvements.

**Example:** If you and another student are building a machine learning project, you can work on different features and use pull requests to review each other's contributions.

## Slide 21: Why Git and GitHub Matter in Data Science

As a data professional, you will work with:

- Python scripts and notebooks.
- Data cleaning and preprocessing pipelines.
- Machine learning models.
- Project documentation.
- Team-based projects.

Git and GitHub help you keep your work organized, reproducible, and accessible.

A typical data science repository might look like this:

```text
student-performance-analysis/
│
├── data/
│   └── student_performance.csv
│
├── notebooks/
│   └── exploratory_analysis.ipynb
│
├── src/
│   └── data_cleaning.py
│
├── README.md
└── requirements.txt
```

**Important:** Do not upload sensitive information, passwords, API keys, or private datasets to a public repository.
