# Introduction to Terminal Commands

## 1: Working with the Terminal

Before writing Python programs or managing projects with Git, you need to understand how to navigate and interact with your computer using the terminal.

The terminal allows you to execute commands to:
- Navigate between folders.
- Create files and directories.
- View existing files.
- Run Python programs.
- Install packages.
- Manage projects.

**Learning objectives**

By the end of this lesson, you should be able to:
- Open and navigate the terminal.
- Create and manage folders and files.
- Understand common terminal commands.
- Run Python scripts from the terminal.
- Use terminal commands when setting up Python projects.

## 2: What Is a Terminal?

A terminal is a text-based interface that allows you to communicate with your computer by typing commands.

Instead of clicking through folders and menus, you can perform tasks by entering instructions.

For example, instead of manually creating a folder, you can type:

```bash
mkdir my-project
```

Your computer interprets the command and creates the folder.

**Important terms**

- **Terminal:** The application where you enter commands.
- **Command:** An instruction given to the computer.
- **Directory:** Another name for a folder.
- **Path:** The location of a file or directory.

## 3: Understanding Your Current Location

When you open a terminal, you are usually located in a particular directory.

This is called your **current working directory**.

To find out where you are:

```bash
pwd
```

`pwd` stands for *Print Working Directory*.

Example output:

```text
/home/student
```

This tells you that your terminal is currently inside the `student` directory.

**Try it:** Open your terminal and run `pwd`. Where are you currently located?

## 4: Listing Files and Folders – `ls`

The `ls` command displays the files and folders inside your current directory.

```bash
ls
```

Example output:

```text
Documents  Downloads  Pictures  projects
```

Useful variations:

```bash
ls -l
```

Displays files with additional details, such as permissions, ownership, size, and modification date.

```bash
ls -a
```

Displays hidden files and folders as well.

```bash
ls -la
```

Displays hidden files together with detailed information.

**Try it:** Run `ls` and identify three directories on your computer.

## Slide 5: Changing Directories – `cd`

The `cd` command allows you to move from one directory to another.

`cd` stands for *Change Directory*.

Move into a folder:

```bash
cd Documents
```

Move into a project folder:

```bash
cd projects
```

Move one directory backwards:

```bash
cd ..
```

Return to your home directory:

```bash
cd ~
```

Move to a specific path:

```bash
cd /home/student/projects
```

**Important:** The folder you want to enter must exist at the location you specify.

**Try it:** Navigate into your `Documents` folder, then return to your previous directory using `cd ..`.

## Slide 6: Creating Directories – `mkdir`

The `mkdir` command creates a new directory.

`mkdir` stands for *Make Directory*.

Create a folder:

```bash
mkdir my-project
```

Create another folder:

```bash
mkdir data-analysis
```

Create multiple folders:

```bash
mkdir data notebooks scripts
```

Create nested directories:

```bash
mkdir -p projects/data-analysis/datasets
```

The `-p` option creates the parent directories if they do not already exist.

**Try it:** Create a folder called `python-practice`.

## Slide 7: Creating Files – `touch`

The `touch` command creates an empty file if it does not already exist.

For example:

```bash
touch main.py
```

This creates a Python file called `main.py`.

Create a Markdown file:

```bash
touch README.md
```

Create multiple files:

```bash
touch analysis.py cleaning.py visualization.py
```

You can also use `touch` to update a file's modification timestamp.

**Important:** `touch` creates an empty file; it does not automatically open the file or add Python code to it.

**Try it:** Create the following files:

- `main.py`
- `README.md`
- `requirements.txt`

## Slide 8: Viewing File Contents – `cat`

The `cat` command displays the contents of a file in your terminal.

For example:

```bash
cat README.md
```

If your README contains:

```text
# Student Performance Analysis

This project explores student performance data.
```

The terminal displays that content.

You can also create a file and add text using:

```bash
echo "My first Python project" > README.md
```

**Be careful:** The `>` operator replaces the existing contents of the file.

To append text without replacing existing content:

```bash
echo "Created using Python" >> README.md
```

## Slide 9: Clearing the Terminal – `clear`

As you run commands, your terminal can become crowded.

Use:

```bash
clear
```

This clears the visible terminal screen.

It does not delete your files, folders, or command history.

You can also use the keyboard shortcut `Ctrl + L` in many terminals.

## Slide 10: Copying Files – `cp`

The `cp` command copies files or directories.

Copy a file:

```bash
cp main.py backup.py
```

This creates a copy of `main.py` called `backup.py`.

Copy a file into another directory:

```bash
cp main.py scripts/
```

Copy a directory and its contents:

```bash
cp -r data-analysis backup-data-analysis
```

The `-r` option means recursive and is used when copying directories.

## Slide 11: Moving and Renaming Files – `mv`

The `mv` command moves files or directories from one location to another.

Move a file into a directory:

```bash
mv main.py scripts/
```

Rename a file:

```bash
mv old_name.py new_name.py
```

Move a directory:

```bash
mv datasets data/
```

**Remember:** The `mv` command can both move and rename files.

## Slide 12: Removing Files and Directories – `rm`

The `rm` command removes files and directories.

Remove a file:

```bash
rm main.py
```

Remove an empty directory:

```bash
rmdir empty-folder
```

Remove a directory and its contents:

```bash
rm -r old-project
```

**Warning:** Be careful when using `rm`. Deleted files may not go to your computer's recycle bin and can be difficult to recover.

Always confirm the file or directory you are deleting.

## Slide 13: Running Python Files from the Terminal

You can execute Python programs directly from the terminal.

First, create a file called `hello.py`.

Add the following code:

```python
print("Hello, Data Science!")
```

Run the file:

```bash
python hello.py
```

On some systems, you may need:

```bash
python3 hello.py
```

Expected output:

```text
Hello, Data Science!
```

**Think:** What happens if you run the command from a directory where `hello.py` does not exist?

## Slide 14: Checking Python and pip Versions

You can check which Python version is installed on your computer.

```bash
python --version
```

Or:

```bash
python3 --version
```

Check your pip version:

```bash
pip --version
```

These commands help you confirm that Python and its package installer are available.

## Slide 15: Installing Python Packages

Python packages provide additional functionality for your programs.

Use `pip install` to install a package.

For example:

```bash
pip install pandas
```

Install multiple packages:

```bash
pip install numpy matplotlib seaborn
```

View installed packages:

```bash
pip list
```

Remember to activate your project's virtual environment before installing project-specific dependencies.

## Slide 16: Understanding File Paths

A file path describes where a file or directory is located.

There are two main types:

**1. Absolute path**

The complete location of a file, starting from the root of the file system.

Example:

```text
/home/student/projects/main.py
```

**2. Relative path**

The location of a file relative to your current working directory.

Example:

```text
projects/main.py
```

Understanding paths helps you navigate projects, load datasets, and execute scripts correctly.
