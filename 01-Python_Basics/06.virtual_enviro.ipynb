# Python Virtual Environments

## Slide 1: Introduction to Virtual Environments

**Managing Python Project Dependencies**

When working on Python projects, different projects may require different libraries and even different library versions.

A virtual environment helps you manage these dependencies independently for each project.

- Install and manage Python packages.
- Deactivate and delete a virtual environment.

## Slide 3: What Is a Virtual Environment?

A virtual environment is an isolated Python environment that allows you to install and manage packages for a specific project.

Each project can have its own:
- Python packages.
- Package versions.
- Dependencies.

This means that installing a package in one project does not automatically affect another project's environment.

**Key idea:** One project, one isolated environment.

## Slide 4: Why Are Virtual Environments Important?

Virtual environments help you:

1. **Avoid dependency conflicts:** Different projects can use different package versions.
2. **Keep projects organized:** Each project has its own dependencies.
3. **Improve reproducibility:** Other developers can install the same dependencies.
4. **Simplify collaboration:** Team members can recreate your project environment.
5. **Protect your global Python installation:** You avoid installing every package globally.

## Slide 5: Understanding Python Packages

A Python package is reusable code that provides functionality you can use in your programs.

In data analysis and data science, common packages include:

| Package | Purpose |
|---|---|
| NumPy | Numerical computing |
| Pandas | Data manipulation and analysis |
| Matplotlib | Data visualization |
| Seaborn | Statistical visualization |
| Scikit-learn | Machine learning |

You can install packages using `pip`, Python's package installer.

For example:

```bash
pip install pandas
```

But where should you install these packages?

Inside your project's virtual environment.

## Slide 7: Creating a Virtual Environment

Python provides a built-in module called `venv` for creating virtual environments.

Run:

```bash
python -m venv venv
```

If your system uses `python3`, you may need:

```bash
python3 -m venv venv
```

Let's break down the command:

- `python` runs Python.
- `-m` tells Python to run a module.
- `venv` is the module used to create virtual environments.
- The final `venv` is the name of the environment directory.

After running the command, you should see a new folder named `venv` inside your project.

**Important:** The environment folder contains the isolated Python environment and its supporting files.

## Slide 8: Activating a Virtual Environment

Creating a virtual environment does not automatically activate it.

You must activate it before installing packages.

**Windows – Command Prompt:**

```bash
venv\Scripts\activate.bat
```

**Windows – PowerShell:**

```powershell
.\venv\Scripts\Activate.ps1
```

**macOS and Linux:**

```bash
source venv/bin/activate
```

When activation is successful, you should see something like this in your terminal:

```text
(venv) C:\student-performance-analysis>
```

The `(venv)` indicates that the virtual environment is active.

## Slide 11: Checking Installed Dependencies

As projects grow, you may install many packages.

It is important to keep a record of the dependencies your project needs.

Run:

```bash
pip freeze
```

This displays installed packages and their versions.

To save the dependencies into a file:

```bash
pip freeze > requirements.txt
```

The `requirements.txt` file might contain:

```text
numpy==2.0.0
pandas==2.2.0
matplotlib==3.9.0
```

The versions shown above are illustrative.

**Why is this useful?**

Another developer can use this file to install the same package versions for your project.

## Slide 12: Installing Dependencies from requirements.txt

Imagine you receive a Python project from another developer.

The project contains a `requirements.txt` file.

Instead of installing each package separately, you can install all the listed dependencies using:

```bash
pip install -r requirements.txt
```

Let's break it down:

- `pip install` installs packages.
- `-r` tells pip to read requirements from a file.
- `requirements.txt` contains the package names and versions.

This makes setting up Python projects much easier.

## Slide 13: Deactivating a Virtual Environment

When you finish working on a project, you can deactivate its virtual environment.

Run:

```bash
deactivate
```

Your terminal will return to the normal environment.

**Remember:** Deactivating an environment does not delete it. Your installed packages and environment files remain available when you activate it again.

## Slide 14: Project Structure

A typical Python project using a virtual environment might look like this:

```text
student-performance-analysis/
│
├── venv/
│
├── main.py
│
├── requirements.txt
│
└── README.md
```

Explanation:

- `venv/` contains the virtual environment.
- `main.py` contains your Python code.
- `requirements.txt` lists project dependencies.
- `README.md` explains the project.

**Important:** You should generally not upload the `venv/` directory to GitHub because it can be large and is specific to your local setup.

## Slide 15: Virtual Environments and GitHub

When working with Git and GitHub, your virtual environment should usually be excluded from version control.

Create a file called `.gitignore` in your project folder.

Add:

```text
venv/
__pycache__/
.env
```

This tells Git to ignore these files and folders.

Why?

- `venv/` contains environment-specific files.
- `__pycache__/` contains generated Python cache files.
- `.env` may contain sensitive configuration information.

Instead of uploading the environment itself, upload `requirements.txt` so another developer can recreate it.
