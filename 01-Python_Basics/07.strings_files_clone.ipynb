# Part 1: Strings

## 1. Introduction to Strings

A string is a sequence of characters. It can contain letters, numbers, spaces, and symbols.

We use strings whenever a program needs to work with text, such as:

- Student names
- Email addresses
- Phone numbers
- Product descriptions
- Customer feedback
- Dates stored as text
- Messages and search terms

A string is created by placing text inside quotation marks.

```python
student_name = "Mary"

course = 'Data Analytics'

message = "Welcome to Python!"

print(student_name)

print(course)

print(message)
```

Both single and double quotation marks can be used. The important thing is to open and close the string using matching quotation marks.

### Why are phone numbers often stored as strings?

A phone number may contain digits, but we do not normally perform arithmetic on it. We also need to preserve leading zeros and symbols such as `+`.

```python
phone_number = "0712345678"

print(phone_number)
```

If we stored the phone number as an integer, a leading zero could be lost.

**Instructor discussion:** Ask students to identify which of these should usually be stored as strings and explain why: a student's age, a student's name, a phone number, and a fee balance.

## 2. String Indexing

Indexing allows us to access one character at a particular position in a string.

Python uses zero-based indexing. This means the first character is at index `0`, not index `1`.

```python
student_name = "Mary"

print(student_name[0])

print(student_name[1])

print(student_name[2])

print(student_name[3])
```

Expected output:

```text
M
a
r
y
```

### Positive and negative indexing

Positive indexing starts from the beginning of the string.

Negative indexing starts from the end of the string. The last character has index `-1`.

```python
course = "Python"

print(course[0])

print(course[1])

print(course[-1])

print(course[-2])
```

Expected output:

```text
P
y
n
o
```

### When is indexing useful?

Imagine that a system needs to inspect the first character of a student ID or access the last character of a code.

```python
student_id = "STU204"

first_character = student_id[0]

last_character = student_id[-1]

print(first_character)

print(last_character)
```

**Important:** Trying to access an index that does not exist raises an `IndexError`.

```python
student_name = "Mary"

print(student_name[5])
```

This raises an error because the valid indexes are `0`, `1`, `2`, and `3`.

## 3. String Slicing

Indexing accesses one character. Slicing extracts a section of a string.

The general syntax is:

```python
string[start:stop]
```

- `start` is the index where extraction begins.
- `stop` is the index where extraction ends, but the character at this index is not included.

```python
course = "Python"

print(course[0:2])

print(course[0:4])

print(course[2:6])
```

Expected output:

```text
Py
Pyth
thon
```

### Leaving out the start or stop

If the start is omitted, Python starts from the beginning.

If the stop is omitted, Python continues to the end.

```python
course = "Python"

print(course[:3])

print(course[2:])

print(course[:])
```

Expected output:

```text
Pyt
thon
Python
```

### Using a step

Slicing can also include a step:

```python
string[start:stop:step]
```

The step determines how many positions Python moves forward between selected characters.

```python
word = "Programming"

print(word[0:10:2])

print(word[::2])
```

A negative step can be used to move backwards.

```python
word = "Python"

print(word[::-1])
```

Expected output:

```text
nohtyP
```

### Practical example: Extracting parts of an ID

Imagine that student IDs follow a consistent format: the first three characters identify the student category, while the remaining characters identify the student number.

```python
student_id = "STU204"

category = student_id[:3]

student_number = student_id[3:]

print(category)

print(student_number)
```

Expected output:

```text
STU
204
```

**Instructor note:** Slicing is useful when text follows a predictable structure. It should not be treated as a reliable way to extract information from text whose format is inconsistent.

## 4. Strings Are Immutable

Strings are immutable. This means that once a string has been created, its individual characters cannot be changed directly.

For example:

```python
student_name = "Mary"

student_name[0] = "H"
```

This raises a `TypeError` because Python does not allow direct modification of a character in a string.

### How do we change a string, then?

We create a new string and assign it to the variable.

```python
student_name = "Mary"

student_name = "H" + student_name[1:]

print(student_name)
```

Expected output:

```text
Hary
```

The original string was not edited character by character. A new string was created and assigned to `student_name`.

String methods also return new strings rather than changing the original string.

```python
student_name = "  mary  "

clean_name = student_name.strip()

print(student_name)

print(clean_name)
```

Expected output:

```text
  mary
mary
```

The original value remains unchanged because `strip()` returns a new string.

### Why does immutability matter?

- It prevents accidental changes to existing text.
- It helps explain why string methods often need to be assigned to a variable.
- It encourages us to distinguish between the original text and the cleaned version.

**Check for understanding:** What will `student_name` contain after the following code?

```python
student_name = "mary"

student_name.upper()

print(student_name)
```

Expected answer: `mary`. The result of `upper()` was not assigned back to the variable.

To keep the uppercase version:

```python
student_name = "mary"

student_name = student_name.upper()

print(student_name)
```

## 5. String Traversal

Traversal means visiting each character in a sequence one at a time.

We can use a `for` loop to traverse a string.

```python
course = "Python"

for character in course:

    print(character)
```

Expected output:

```text
P
y
t
h
o
n
```

### Traversing using indexes and `range()`

Sometimes we need the position of each character as well as the character itself. We can use `range()` and `len()`.

```python
course = "Python"

for index in range(len(course)):

    print(index, course[index])
```

Expected output:

```text
0 P
1 y
2 t
3 h
4 o
5 n
```

### Practical example: Counting a character

Imagine that we want to count how many times the letter `a` appears in a student's response.

```python
response = "Data analysis is about finding patterns"

count = 0

for character in response:

    if character == "a":

        count += 1

print(count)
```

This is a simple example of processing text character by character. In real projects, similar operations can help inspect, validate, or analyse text.

## 6. String Comparison

Strings can be compared using comparison operators such as:

- `==` equal to
- `!=` not equal to
- `<` less than
- `>` greater than

String equality is case-sensitive. This means `"Python"` and `"python"` are considered different strings.

```python
first_word = "Python"

second_word = "python"

print(first_word == second_word)

print(first_word != second_word)
```

Expected output:

```text
False
True
```

### Comparing without case sensitivity

If we want to compare text regardless of capitalisation, we can standardise the case first.

```python
entered_course = "PYTHON"

expected_course = "python"

if entered_course.lower() == expected_course.lower():

    print("The course names match")

else:

    print("The course names do not match")
```

Strings can also be compared alphabetically using `<` and `>`. Python compares character values according to their Unicode code points. This is not the same as comparing the meaning of words.

```python
print("apple" < "banana")

print("Mary" == "mary")
```

**Instructor note:** Focus first on equality and inequality. Explain alphabetical comparisons as an additional feature rather than suggesting that they understand language or dictionary order in every situation.

## 7. Membership Operators

Membership operators check whether a character or sequence of characters exists inside a string.

- `in` checks whether something is present.
- `not in` checks whether something is absent.

```python
email = "mary@example.com"

print("@" in email)

print("gmail" in email)

print("yahoo" not in email)
```

Expected output:

```text
True
False
True
```

### Practical example: Basic email check

```python
email = "mary@example.com"

if "@" in email and "." in email:

    print("The email contains the expected characters")

else:

    print("The email may be invalid")
```

This is only a basic check. It does not guarantee that an email address is valid.

### When is membership useful?

- Checking whether a word appears in a message.
- Checking whether an email contains `@`.
- Checking whether a response contains a particular keyword.
- Checking whether a password contains a required character.
- Searching text for a phrase.

# Part 2: Common String Methods

String methods are built-in operations that help us inspect, clean, transform, and split text.

They are especially useful when working with real-world data because text is not always consistently formatted.

For example, the following values may represent the same course:

```python
course_1 = "Data Analytics"

course_2 = " data analytics "

course_3 = "DATA ANALYTICS"
```

Without cleaning, a program may treat them as different values.

## 8. Changing Letter Case

### `upper()`

Converts all letters to uppercase.

```python
course = "Data Analytics"

print(course.upper())
```

### `lower()`

Converts all letters to lowercase.

```python
course = "DATA ANALYTICS"

print(course.lower())
```

### `capitalize()`

Makes the first character uppercase and the remaining characters lowercase.

```python
student_name = "mARY"

print(student_name.capitalize())
```

### `title()`

Makes the first character of each word uppercase.

```python
student_name = "mary omondi"

print(student_name.title())
```

### `swapcase()`

Changes uppercase characters to lowercase and lowercase characters to uppercase.

```python
message = "Hello Python"

print(message.swapcase())
```

**When are these useful?** Use case conversion when standardising names, categories, search terms, or text labels. Choose the method that matches the formatting you need.

## 9. Removing Unwanted Spaces

### `strip()`

Removes whitespace from both ends of a string.

```python
student_name = "   Mary   "

clean_name = student_name.strip()

print(clean_name)
```

### `lstrip()`

Removes whitespace from the left side.

```python
student_name = "   Mary   "

print(student_name.lstrip())
```

### `rstrip()`

Removes whitespace from the right side.

```python
student_name = "   Mary   "

print(student_name.rstrip())
```

These methods do not remove spaces between words.

```python
student_name = "   Mary Omondi   "

print(student_name.strip())
```

## 10. Replacing Text

`replace()` replaces one part of a string with another.

Syntax:

```python
string.replace(old, new)
```

Example:

```python
message = "I am learning Java"

updated_message = message.replace("Java", "Python")

print(updated_message)
```

Expected output:

```text
I am learning Python
```

We can also replace unwanted characters.

```python
phone_number = "0712-345-678"

clean_number = phone_number.replace("-", "")

print(clean_number)
```

Expected output:

```text
0712345678
```

Remember: `replace()` returns a new string.

## 11. Splitting Strings

`split()` breaks a string into a list of smaller strings.

By default, it splits at whitespace.

```python
full_name = "Mary Wanjiku Omondi"

name_parts = full_name.split()

print(name_parts)
```

Expected output:

```text
['Mary', 'Wanjiku', 'Omondi']
```

We can also specify a separator.

```python
student_record = "Mary,Data Analytics,85"

record_parts = student_record.split(",")

print(record_parts)
```

Expected output:

```text
['Mary', 'Data Analytics', '85']
```

### Why is splitting useful?

Files and datasets often contain information separated by commas, tabs, or other delimiters. Splitting helps us separate that information into individual fields.

## 12. Joining Strings

`join()` combines items from an iterable into one string, using the string before `.join()` as the separator.

```python
name_parts = ["Mary", "Wanjiku", "Omondi"]

full_name = " ".join(name_parts)

print(full_name)
```

Expected output:

```text
Mary Wanjiku Omondi
```

Using a comma as the separator:

```python
subjects = ["Python", "Statistics", "Data Analysis"]

subject_list = ", ".join(subjects)

print(subject_list)
```

Expected output:

```text
Python, Statistics, Data Analysis
```

## 13. Checking the Contents of a String

Some methods return `True` or `False` depending on the contents of a string.

| Method | What it checks |
|---|---|
| `isalpha()` | All characters are letters |
| `isdigit()` | All characters are digits |
| `isalnum()` | All characters are letters or digits |
| `isspace()` | All characters are whitespace |
| `islower()` | Cased letters are lowercase |
| `isupper()` | Cased letters are uppercase |
| `istitle()` | The string follows title-case formatting |

Examples:

```python
student_name = "Mary"

student_age = "25"

student_id = "STU204"

print(student_name.isalpha())

print(student_age.isdigit())

print(student_id.isalnum())
```

Expected output:

```text
True
True
True
```

These methods are useful for basic validation, but they do not replace complete validation rules. For example, a name may contain a space or hyphen, so `isalpha()` may return `False` even when the name is valid.

## 14. Finding Text and Counting Occurrences

### `find()`

Returns the index of the first occurrence of a substring. If the substring is not found, it returns `-1`.

```python
message = "Python is useful for data analysis"

position = message.find("data")

print(position)
```

### `count()`

Counts how many times a substring appears.

```python
message = "Python is popular. Python is useful."

print(message.count("Python"))
```

### `startswith()` and `endswith()`

Check whether a string starts or ends with a particular sequence.

```python
filename = "students.csv"

print(filename.startswith("students"))

print(filename.endswith(".csv"))
```

These methods are useful when checking file extensions, prefixes, codes, or expected text patterns.

# Part 3: File Handling

## 16. Why Do We Need File Handling?

So far, most of our examples have stored information in variables.

But variables exist while the program is running. When the program stops, values stored only in variables are lost.

Imagine a student registration program that collects:

- Student name
- Email address
- Course
- Registration date

If the program closes, how can we access those records the next time it runs?

We need to save the information somewhere.

File handling allows a Python program to:

- Read information that has already been saved.
- Create new files.
- Write information into files.
- Add new information without deleting existing content.
- Process information stored in files.

This is important in data analysis because data is commonly stored in files such as CSV, TXT, and JSON files. Python needs to access those files before we can inspect or analyse their contents.

In this lesson, we will focus on basic text file handling using `.txt` files. CSV and JSON files can be explored separately.

## 17. Opening a File

Python uses the `open()` function to open a file so that we can read or write information.

**Syntax:**

```python
file = open("filename.txt", "mode")
```

The mode determines how we interact with the file.

### Common File Modes

| Mode | Meaning |
|---|---|
| `"r"` | Read an existing file |
| `"w"` | Write to a file; creates it if it does not exist and overwrites existing content |
| `"a"` | Append information to the end of a file without deleting existing content |
| `"x"` | Create a new file; raises an error if it already exists |

**Example: Student Registration System**

- `"r"` – View registered students.
- `"w"` – Create a new registration file.
- `"a"` – Add a newly registered student.
- `"x"` – Create a file only if it does not already exist.

**Important:** Be careful when using `"w"` because it replaces existing content.

## 18. Writing to a File

We use `"w"` mode to write information into a file.

Suppose we want to save student names in `students.txt`.

```python
with open("students.txt", "w") as file:

    file.write("Mary Omondi\n")
    file.write("Brian Kamau\n")
    file.write("John Mwangi\n")
```

This creates the file if it does not exist. If it already exists, its previous content is replaced.

### Understanding `\n`

The `\n` character represents a new line.

Without it, information is written continuously.

```python
with open("courses.txt", "w") as file:

    file.write("Python")
    file.write("Data Analytics")
```

File content:

```text
PythonData Analytics
```

Adding `\n` separates the entries:

```python
with open("courses.txt", "w") as file:

    file.write("Python\n")
    file.write("Data Analytics\n")
```

### Writing User Input to a File

Instead of manually entering names in our code, we can collect them from the user.

```python
student_name = input("Enter student name: ").strip().title()

with open("students.txt", "a") as file:

    file.write(student_name + "\n")

print("Student registered successfully!")
```

This combines user input, string methods and file handling.

## 19. Reading from a File

We use `"r"` mode to retrieve information from an existing file.

```python
with open("students.txt", "r") as file:

    content = file.read()

    print(content)
```

### Different Ways to Read a File

| Method | Purpose | Returns |
|---|---|---|
| `read()` | Reads the entire file | String |
| `readline()` | Reads one line | String |
| `readlines()` | Reads all lines | List of strings |

**Examples:**

```python
# Read the first line
with open("students.txt", "r") as file:
    print(file.readline())


# Read all lines into a list
with open("students.txt", "r") as file:
    students = file.readlines()
    print(students)
```

### Reading a File Line by Line

We can also use a loop to process each record individually.

```python
with open("students.txt", "r") as file:

    for line in file:

        print(line.strip())
```

The `strip()` method removes newline characters and extra whitespace at the beginning and end of each line.

### Using a Function to View Records

We can organise this operation into a reusable function.

```python
def view_students():

    with open("students.txt", "r") as file:

        for line in file:
            print(line.strip())


view_students()
```

## 20. Appending to a File

Suppose another student registers. We want to add their name without deleting existing records.

We use `"a"` mode.

```python
def add_student():

    name = input("Enter student name: ").strip().title()

    if name:

        with open("students.txt", "a") as file:

            file.write(name + "\n")

        print("Student added successfully!")

    else:
        print("Student name cannot be empty.")


add_student()
```

This example combines:
- Functions to organise reusable code.
- User input to collect information.
- Conditional statements to validate the input.
- File handling to save the new record.

**Remember:**

- `"w"` replaces existing content.
- `"a"` preserves existing content and adds new information.

Append mode is useful for registration records, attendance logs and transaction records.

## 21. Closing a File

When we open a file, Python creates a file object that allows us to interact with it.

After using the file, it should be closed to release the associated resources.

```python
file = open("students.txt", "r")

content = file.read()

print(content)

file.close()
```

However, manually closing files can be forgotten, especially when an error occurs before the `close()` statement runs.

Python provides a safer approach using the `with` statement.

## 22. Using `with open()`

The recommended way to work with files is to use `with open()`.

```python
with open("students.txt", "r") as file:

    content = file.read()

    print(content)
```

The file is automatically closed when the indented block finishes, even if an error occurs inside the block.

We can use `with open()` with all the file modes:

- `"r"` for reading.
- `"w"` for writing.
- `"a"` for appending.
- `"x"` for creating a new file.

**Why use `with open()`?**

- Closes files automatically.
- Reduces the risk of leaving files open.
- Makes code cleaner and easier to read.

**Instructor note:** Encourage students to use `with open()` in their examples instead of manually opening and closing files.

- Check that the file exists in the specified location.
- Check the spelling of the filename and extension.
- Understand that the current working directory may differ from the folder containing your Python script.

For portability, practise using relative paths within project folders.

## 24. Reading and Cleaning File Content

This is where strings and file handling come together.

A file may contain information with extra spaces or inconsistent capitalisation.

Imagine `students.txt` contains:

```text
  mary omondi
BRIAN KAMAU
  john mwangi
```

We can read the file and clean each line.

```python
with open("students.txt", "r") as file:

    for line in file:

        student_name = line.strip().title()

        print(student_name)
```

Expected output:

```text
Mary Omondi
Brian Kamau
John Mwangi
```

Notice how we combined:

- File handling to read the information.
- Traversal to process each line.
- `strip()` to remove unwanted spaces and newline characters.
- `title()` to standardise capitalisation.

This is a common data preparation task.

## 25. Writing Multiple Records to a File

We can use a list and a loop to write multiple values into a file.

```python
students = ["Mary", "Brian", "John", "Faith"]

with open("students.txt", "w") as file:

    for student in students:

        file.write(student + "\n")
```

We can then read the file:

```python
with open("students.txt", "r") as file:

    for line in file:

        print(line.strip())
```

This demonstrates how data structures and file handling work together.

# Part 4: Error Handling

## 26. Why Do We Need Error Handling?

Programs do not always run under perfect conditions.

Consider a program that reads student records from a file:

- What if the file does not exist?
- What if the user enters a number where a name is expected?
- What if the file cannot be accessed?
- What if the file contains unexpected information?

Without error handling, a program may stop unexpectedly and display an error message that is not helpful to the user.

Error handling allows us to anticipate possible problems and decide how the program should respond.

It makes programs more reliable and improves the experience for people using them.

## 27. Understanding `try` and `except`

Python uses `try` and `except` to handle exceptions.

- `try` contains code that might raise an error.
- `except` contains the response when a matching error occurs.

```python
try:

    age = int(input("Enter your age: "))

    print(age)

except ValueError:

    print("Please enter a valid whole number")
```

If the user enters `25`, the conversion succeeds.

If the user enters `twenty`, Python raises a `ValueError`, and the `except` block runs.

### Why catch a specific error?

It is better to handle the error we expect rather than catching every possible error without understanding it.

For example, `ValueError` is appropriate when converting text to an integer and the text does not represent a valid integer.

## 28. Handling File Errors

### `FileNotFoundError`

This occurs when Python tries to open a file that does not exist at the specified path.

```python
try:

    with open("students.txt", "r") as file:

        content = file.read()

        print(content)

except FileNotFoundError:

    print("The students file could not be found")
```

### `PermissionError`

This can occur when the program does not have the required permission to access a file.

```python
try:

    with open("students.txt", "r") as file:

        content = file.read()

        print(content)

except PermissionError:

    print("You do not have permission to access this file")
```

### Handling more than one expected error

We can use multiple `except` blocks.

```python
try:

    with open("students.txt", "r") as file:

        content = file.read()

        print(content)

except FileNotFoundError:

    print("The file does not exist")

except PermissionError:

    print("You do not have permission to open this file")
```

Each block handles a different type of error.

## 29. Using `else` and `finally`

Python also provides `else` and `finally` in exception handling.

- `else` runs if no exception occurs in the `try` block.
- `finally` runs whether an exception occurs or not.

```python
try:

    age = int(input("Enter your age: "))

except ValueError:

    print("Please enter a valid whole number")

else:

    print("Your age is", age)

finally:

    print("Age entry process completed")
```

### When are these useful?

- Use `else` when you want to run code only if the risky operation succeeds.
- Use `finally` for cleanup or actions that should happen regardless of whether an error occurred.

**Instructor note:** When using `with open()`, students do not need a `finally` block just to close the file. The context manager handles that automatically.

## 30. A Practical Example: Safely Reading a Student File

```python
try:

    with open("students.txt", "r") as file:

        for line in file:

            student_name = line.strip()

            if student_name:

                print(student_name)

except FileNotFoundError:

    print("Student records were not found")

except PermissionError:

    print("You do not have permission to read the records")
```

This program:

1. Attempts to open the file.
2. Reads it one line at a time.
3. Removes unwanted whitespace.
4. Skips empty lines.
5. Displays a helpful message if the file cannot be accessed.

## 31. Avoiding Broad Exception Handling

You may sometimes see:

```python
try:

    number = int(input("Enter a number: "))

except:

    print("Something went wrong")
```

This catches almost every exception, including errors that the programmer may not have intended to handle.

A better approach is to catch the specific error expected:

```python
try:

    number = int(input("Enter a number: "))

except ValueError:

    print("Please enter a valid number")
```

**Key lesson:** Error handling should make a program more reliable, not hide programming mistakes.

## 32. Integrated Scenario: Student Records Manager

Imagine that a training institution needs a simple program to manage student names.

The program should:

1. Ask the user to enter student names.
2. Remove unwanted spaces.
3. Standardise the names.
4. Save each name to a text file.
5. Read the saved names.
6. Display the names.
7. Handle the situation where the file cannot be read.

This scenario brings together strings, loops, file handling, and error handling.

### Reference solution

```python
students = []

number_of_students = int(input("How many students are you registering? "))

for number in range(number_of_students):

    student_name = input("Enter student name: ")

    student_name = student_name.strip().title()

    students.append(student_name)


with open("students.txt", "w") as file:

    for student in students:

        file.write(student + "\n")


print("\nRegistered students:")

try:

    with open("students.txt", "r") as file:

        for line in file:

            print(line.strip())

except FileNotFoundError:

    print("The student records file could not be found")

except PermissionError:

    print("You do not have permission to read the student records")
```

### Walkthrough

| Part of the program | Concept applied |
|---|---|
| `input()` | Collecting text from the user |
| `strip()` | Removing unwanted spaces |
| `title()` | Standardising names |
| `append()` | Adding names to a list |
| `for` loop | Repeating an operation for each student |
| `open(..., "w")` | Creating or replacing the records file |
| `write()` | Saving each name |
| `open(..., "r")` | Reading saved records |
| `line.strip()` | Cleaning each line read from the file |
| `try` and `except` | Handling expected file errors |

**Instructor discussion:** Ask students what would happen if the program used `"a"` instead of `"w"`. The answer is that each run would add another set of names to the existing file rather than replacing its contents.
