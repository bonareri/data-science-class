# Part 1: Understanding APIs

## 1. What Is an API?

API stands for **Application Programming Interface**.

An API is a set of rules and mechanisms that allows one software application to communicate with another.

Think of an API as a waiter in a restaurant.

Imagine you are sitting in a restaurant:

1. You look at the menu and decide what you want.
2. You give your order to the waiter.
3. The waiter takes your order to the kitchen.
4. The kitchen prepares your food.
5. The waiter brings the food back to you.

You do not need to enter the kitchen or understand how the food is prepared. You simply communicate your request through the waiter.

An API works in a similar way.

- Your application makes a request.
- The API receives the request.
- The external system processes it.
- The API returns a response to your application.

<text></text>

### Real-world examples of APIs

Consider some applications that you may use every day:

- A mobile banking application retrieves account information from banking systems.
- A weather application retrieves weather information from a weather service.
- An e-commerce application retrieves product information and prices.
- A map application retrieves geographical information and directions.
- A payment application communicates with a payment provider to process transactions.

These applications can communicate with external systems through APIs.

### Why are APIs important?

APIs allow developers to:

- Access information from external services.
- Integrate different applications.
- Reuse functionality developed by other organisations.
- Automate the exchange of information.
- Build applications without developing every feature from scratch.
- Retrieve current information from external data sources.

**Instructor discussion:** Ask students to identify an application they use and suggest what external information or services it might need.

## 2. Understanding the Client-Server Model

To understand APIs, we need to understand how applications communicate.

There are two important participants:

**Client:** The application that makes a request for information or a service.

**Server:** The system that receives the request, processes it, and sends back a response.

For example, imagine a Python program requesting information about a particular country.

- Python is the client.
- The API acts as the interface through which the request is made.
- The server processes the request.
- The server sends the response back to Python.

The communication can be represented as follows:

```text
Python Application (Client)
          |
          | 1. Sends API Request
          |
          v
       API Server
          |
          | 2. Processes Request
          |
          v
       API Server
          |
          | 3. Returns API Response
          |
          v
Python Application (Client)
```

The Python application can then use the information returned by the server.

## 3. What Is an API Endpoint?

An API endpoint is a specific URL through which an application can access a particular resource or functionality.

For example, an API might provide information about countries through an endpoint such as:

```text
https://restcountries.com/v3.1/name/kenya
```

This endpoint requests information about Kenya.

Another endpoint from the same API might retrieve information about all countries.

```text
https://restcountries.com/v3.1/all
```

Different endpoints can provide access to different resources.

### Important API terminology

| Term | Meaning |
|---|---|
| API | Allows applications to communicate |
| Client | Application making the request |
| Server | System processing the request |
| Endpoint | URL used to access an API resource |
| Request | Message sent to an API |
| Response | Information returned by an API |
| API documentation | Instructions explaining how to use an API |
| API key | A credential used by some APIs to identify or authorise access |

# Part 2: Understanding HTTP

## 4. What Is HTTP?

HTTP stands for **Hypertext Transfer Protocol**.

It is a communication protocol used by clients and servers to exchange information over the web.

When Python communicates with a web API, it commonly sends an HTTP request and receives an HTTP response.

You will often encounter URLs beginning with:

- `http://`
- `https://`

HTTPS is the secure version of HTTP. It uses encryption to protect information exchanged between the client and server.

Most modern APIs use HTTPS.

## 5. HTTP Methods

HTTP methods describe the action a client wants to perform.

The four common methods are:

| Method | Purpose | Example |
|---|---|---|
| GET | Retrieve information | Get details about a country |
| POST | Submit or create information | Register a new student |
| PUT | Replace an existing resource | Update a student's full profile |
| DELETE | Remove a resource | Delete a student record |

### GET

GET is used when we want to retrieve information from an API.

For example, requesting information about Kenya.

```text
GET https://restcountries.com/v3.1/name/kenya
```

### POST

POST is used to send information to a server, often to create a new resource.

For example, sending details to register a student.

### PUT

PUT is commonly used to replace an existing resource with updated information.

### DELETE

DELETE is used to request the removal of a resource.

**Important:** The examples above describe typical REST API conventions. The exact behaviour of each method depends on the API's documentation and permissions.

For this introductory lesson, we will mainly focus on GET requests because they allow us to retrieve data from public APIs.

# Part 3: Understanding API Responses

## 6. What Happens When an API Receives a Request?

When we send a request to an API, the server processes it and returns a response.

An HTTP response typically contains:

1. A status code.
2. Response headers.
3. A response body, which may contain the requested data.

For example, a server might return information about a country in JSON format.

```json
{
    "name": {
        "common": "Kenya",
        "official": "Republic of Kenya"
    },
    "capital": [
        "Nairobi"
    ],
    "region": "Africa",
    "population": 56432944
}
```

The response contains information about Kenya.

We can see that the data is organised using key-value pairs and nested structures.

This format is called JSON.

## 7. What Is JSON?

JSON stands for **JavaScript Object Notation**.

It is a lightweight format used to exchange data between applications.

Although JSON originated from JavaScript, it is widely used by many programming languages, including Python.

JSON supports the following data types:

| JSON Data Type | Example |
|---|---|
| String | `"Mary"` |
| Number | `25` |
| Boolean | `true` |
| Array | `["Python", "SQL"]` |
| Object | `{"name": "Mary"}` |
| Null | `null` |

### Understanding a JSON object

Consider the following JSON data:

```json
{
    "name": "Mary",
    "age": 25,
    "course": "Data Analytics",
    "is_enrolled": true
}
```

This object contains four key-value pairs.

- `name` is associated with `"Mary"`.
- `age` is associated with `25`.
- `course` is associated with `"Data Analytics"`.
- `is_enrolled` is associated with `true`.

### JSON versus Python dictionaries

JSON objects look similar to Python dictionaries.

Python dictionary:

```python
student = {
    "name": "Mary",
    "age": 25,
    "course": "Data Analytics"
}
```

JSON representation:

```json
{
    "name": "Mary",
    "age": 25,
    "course": "Data Analytics"
}
```

They are similar, but they are not exactly the same format.

For example:

- JSON uses `true`, `false`, and `null`.
- Python uses `True`, `False`, and `None`.

Python can convert JSON data into Python dictionaries and lists.

We will use this when processing API responses.

## 9. Making Your First GET Request

We will use the REST Countries API to retrieve information about Kenya.

API endpoint:

```text
https://restcountries.com/v3.1/name/kenya
```

Python code:

```python
import requests

url = "https://restcountries.com/v3.1/name/kenya"

response = requests.get(url)

print(response)
```

The `requests.get()` function sends a GET request to the specified URL.

The result is stored in the `response` variable.

When you print the response, you may see something like:

```text
<Response [200]>
```

This tells us that the server returned HTTP status code `200`.

But this does not display the actual country information. To access the returned data, we need to inspect the response body.

## 10. Understanding the API Response Object

When we make a request using `requests`, Python returns a response object.

This object contains useful information about the request.

Some common response attributes and methods include:

| Attribute or Method | Purpose |
|---|---|
| `response.status_code` | Returns the HTTP status code |
| `response.text` | Returns the response body as text |
| `response.json()` | Converts a JSON response into Python objects |
| `response.headers` | Provides information about the response headers |
| `response.url` | Returns the URL used for the request |

Example:

```python
import requests

url = "https://restcountries.com/v3.1/name/kenya"

response = requests.get(url)

print(response.status_code)

print(response.url)

print(response.text)
```

We can use `response.json()` to work with the returned JSON data.

```python
import requests

url = "https://restcountries.com/v3.1/name/kenya"

response = requests.get(url)

data = response.json()

print(data)
```

The `json()` method converts the JSON response into Python objects.

For this endpoint, the returned data is a list containing country information.

## 11. Accessing Data from an API Response

Remember that APIs often return nested JSON structures.

The response from the REST Countries API contains a list of countries. Each country is represented by a dictionary.

We can access the first country using index `0`.

```python
import requests

url = "https://restcountries.com/v3.1/name/kenya"

response = requests.get(url)

data = response.json()

country = data[0]

print(country)
```

We can then access specific information using dictionary keys.

```python
import requests

url = "https://restcountries.com/v3.1/name/kenya"

response = requests.get(url)

data = response.json()

country = data[0]

name = country["name"]["common"]

capital = country["capital"][0]

region = country["region"]

population = country["population"]

print("Country:", name)

print("Capital:", capital)

print("Region:", region)

print("Population:", population)
```

Notice how we combined:

- Lists to access the first country.
- Dictionaries to access specific fields.
- Nested dictionary access to retrieve the common country name.
- List indexing to retrieve the capital.

This demonstrates why understanding Python data structures is important when working with APIs.

# Part 5: Query Parameters and Headers

## 12. What Are Query Parameters?

Sometimes we do not want all the information available from an API.

We may want to filter the results or specify what information we need.

Query parameters allow us to pass additional information as part of an API request.

For example:

```text
https://api.example.com/students?course=python
```

Here:

- `course` is the parameter name.
- `python` is the parameter value.
- `?` separates the endpoint from the query parameters.

Multiple parameters are separated using `&`.

```text
https://api.example.com/students?course=python&status=active
```

This request asks for students whose course is Python and whose status is active, assuming the API supports these filters.

### Using query parameters in Python

The `requests` library allows us to pass parameters using the `params` argument.

```python
import requests

url = "https://api.example.com/students"

parameters = {
    "course": "python",
    "status": "active"
}

response = requests.get(url, params=parameters)

print(response.url)
```

This example demonstrates the syntax. The example domain is a placeholder, not a real student API.

### Why are query parameters useful?

- Filtering records.
- Searching for specific information.
- Limiting the number of results.
- Specifying a page of results.
- Selecting information based on a category or date.

The available parameters depend on the API documentation.

## 13. What Are API Headers?

Headers provide additional information about an HTTP request or response.

For example, headers can communicate:

- The format of data expected.
- The format of data being sent.
- Authentication information.
- Other request preferences.

A common request header is:

```text
Accept: application/json
```

This tells the server that the client expects JSON data, if supported.

### Sending headers using Python

```python
import requests

url = "https://api.example.com/students"

headers = {
    "Accept": "application/json"
}

response = requests.get(url, headers=headers)

print(response.status_code)
```

Again, this is a syntax example using a placeholder endpoint.

# Part 6: API Authentication

## 14. What Is API Authentication?

Not all APIs are publicly accessible.

Some APIs require users or applications to prove that they have permission to access the service.

This process is called authentication.

An API may require:

- An API key.
- An access token.
- Username and password authentication.
- OAuth authentication.

### What is an API key?

An API key is a credential that an API provider issues to identify an application or control access.

For example, a provider might require an API key to retrieve weather information.

A request may include an API key in a header:

```python
import requests

url = "https://api.example.com/weather"

headers = {
    "Authorization": "Bearer YOUR_API_KEY"
}

response = requests.get(url, headers=headers)
```

`YOUR_API_KEY` is a placeholder. It must be replaced with a valid credential issued by the API provider.

### Important security practices

- Never publish API keys in public GitHub repositories.
- Do not share secret credentials in screenshots or messages.
- Avoid hardcoding sensitive credentials directly into source code.
- Use environment variables or a secure secrets manager.
- Follow the API provider's authentication instructions.

For example, an API key can be stored in an environment variable and retrieved using Python's `os` module.

```python
import os

api_key = os.getenv("WEATHER_API_KEY")

print(api_key)
```

The environment variable must be configured on the machine running the program.

**Instructor note:** Students should understand what an API key is and why it must be protected. They do not need to implement a complex authentication system in this introductory lesson.

# Part 7: Understanding HTTP Status Codes

## 15. What Are HTTP Status Codes?

Whenever an API receives a request, the server returns a status code indicating the result.

Status codes help us understand whether a request was successful or whether something went wrong.

### Common HTTP status codes

| Status Code | Meaning | Explanation |
|---|---|---|
| 200 | OK | Request was successful |
| 201 | Created | A resource was successfully created |
| 204 | No Content | Request succeeded, but there is no response body |
| 400 | Bad Request | The request contains invalid information |
| 401 | Unauthorized | Authentication is missing or invalid |
| 403 | Forbidden | The client does not have permission |
| 404 | Not Found | The requested resource does not exist |
| 429 | Too Many Requests | The client has exceeded a request limit |
| 500 | Internal Server Error | Something went wrong on the server |
| 503 | Service Unavailable | The server is temporarily unable to handle the request |

### Understanding status code categories

Status codes are grouped into five categories:

| Range | Category | Meaning |
|---|---|---|
| 1xx | Informational | Request is being processed |
| 2xx | Success | Request was successful |
| 3xx | Redirection | Further action is needed to complete the request |
| 4xx | Client Error | There is a problem with the request |
| 5xx | Server Error | There is a problem on the server |

### Checking the status code in Python

```python
import requests

url = "https://restcountries.com/v3.1/name/kenya"

response = requests.get(url)

if response.status_code == 200:

    print("Request successful")

else:

    print("Request failed:", response.status_code)
```

Checking the status code helps us determine whether we should process the response.

# Part 8: Error Handling When Working with APIs

## 16. Why Do We Need Error Handling?

API requests do not always succeed.

Some possible problems include:

- No internet connection.
- The API server is unavailable.
- The endpoint is incorrect.
- The request takes too long.
- The server returns an error status code.
- The API returns data in an unexpected format.

Without error handling, our program may stop unexpectedly.

We can use Python's exception handling mechanisms to make our API programs more reliable.

## 17. Handling Request Exceptions

The `requests` library provides exceptions that help us handle different request failures.

Some common exceptions include:

| Exception | Meaning |
|---|---|
| `requests.exceptions.ConnectionError` | A connection to the server could not be established |
| `requests.exceptions.Timeout` | The request took too long |
| `requests.exceptions.HTTPError` | The server returned an unsuccessful HTTP status when `raise_for_status()` is used |
| `requests.exceptions.RequestException` | A general exception related to making a request |

### Handling connection errors

```python
import requests

url = "https://restcountries.com/v3.1/name/kenya"

try:

    response = requests.get(url, timeout=10)

    print(response.status_code)

except requests.exceptions.ConnectionError:

    print("Unable to connect to the API server")
```

### Handling timeout errors

A timeout occurs when the request takes longer than the specified time.

```python
import requests

url = "https://restcountries.com/v3.1/name/kenya"

try:

    response = requests.get(url, timeout=5)

    print(response.status_code)

except requests.exceptions.Timeout:

    print("The request took too long")
```

The `timeout` argument prevents the program from waiting indefinitely for a response.

## 18. Using `raise_for_status()`

Checking `response.status_code` manually is one way to determine whether a request succeeded.

The `requests` library also provides `raise_for_status()`.

This method raises an `HTTPError` if the server returns an unsuccessful HTTP status code, generally a 4xx or 5xx response.

```python
import requests

url = "https://restcountries.com/v3.1/name/kenya"

try:

    response = requests.get(url, timeout=10)

    response.raise_for_status()

    data = response.json()

    print(data)

except requests.exceptions.HTTPError:

    print("The API returned an unsuccessful status code")

except requests.exceptions.ConnectionError:

    print("Unable to connect to the API")

except requests.exceptions.Timeout:

    print("The API request timed out")
```

### Why is this useful?

Instead of manually checking every unsuccessful status code, we can use `raise_for_status()` to raise an exception when the response indicates an HTTP error.

Remember that a successful HTTP response does not automatically guarantee that the response contains the exact data our program expects. We may still need to validate the returned data.

# Part 9: Practical API Project

## 19. Practical Scenario: Building a Country Information Application

Imagine that you are developing a simple application that allows users to search for information about countries.

The user enters a country name, and the application retrieves information from the REST Countries API.

The program should display:

- Country name.
- Capital city.
- Region.
- Population.
- Currency information.

### API endpoint

```text
https://restcountries.com/v3.1/name/{name}
```

Replace `{name}` with the country you want to search for.

For example:

```text
https://restcountries.com/v3.1/name/kenya
```

### Step 1: Import the required library

```python
import requests
```

### Step 2: Collect the country name

```python
country_name = input("Enter a country name: ")
```

### Step 3: Construct the API endpoint

```python
url = f"https://restcountries.com/v3.1/name/{country_name}"
```

### Step 4: Send the API request

```python
response = requests.get(url, timeout=10)
```

### Step 5: Check the response

```python
if response.status_code == 200:

    data = response.json()

    print(data)

else:

    print("Country information could not be retrieved")
```

### Complete reference solution

```python
import requests

country_name = input("Enter a country name: ").strip()

url = f"https://restcountries.com/v3.1/name/{country_name}"

try:

    response = requests.get(url, timeout=10)

    response.raise_for_status()

    data = response.json()

    country = data[0]

    name = country["name"]["common"]

    capital = country.get("capital", ["Not available"])[0]

    region = country.get("region", "Not available")

    population = country.get("population", "Not available")

    currencies = country.get("currencies", {})

    print("\nCountry Information")

    print("-------------------")

    print("Country:", name)

    print("Capital:", capital)

    print("Region:", region)

    print("Population:", population)

    print("Currencies:", currencies)

except requests.exceptions.HTTPError:

    print("Country not found or the API returned an error")

except requests.exceptions.ConnectionError:

    print("Unable to connect. Check your internet connection")

except requests.exceptions.Timeout:

    print("The request took too long. Please try again")

except requests.exceptions.RequestException:

    print("An error occurred while making the API request")

except (IndexError, KeyError, ValueError):

    print("The API returned data in an unexpected format")
```

### Understanding the program

| Code | Purpose |
|---|---|
| `input()` | Collects the country name |
| `.strip()` | Removes unwanted spaces |
| `f""` | Inserts the country name into the endpoint |
| `requests.get()` | Sends the HTTP GET request |
| `timeout=10` | Limits how long the request waits |
| `raise_for_status()` | Raises an exception for unsuccessful HTTP responses |
| `response.json()` | Converts JSON into Python objects |
| `data[0]` | Retrieves the first country from the returned list |
| `.get()` | Retrieves a dictionary value with a fallback |
| `try` and `except` | Handles possible errors |

**Instructor note:** The API must be reachable for the practical to work. Students should have an internet connection. API responses can change, so encourage them to inspect the returned JSON before assuming a particular field is always present.

# Part 10: APIs and Data Analysis

## 20. Why Are APIs Important in Data Analysis?

As data analysts, we do not always receive datasets as Excel or CSV files.

Sometimes we need to retrieve data directly from online sources.

For example:

- A financial analyst may retrieve exchange rates from a financial API.
- A business analyst may retrieve sales information from an organisation's API.
- A public health analyst may retrieve health statistics from an open-data API.
- A data scientist may retrieve information from a social media or research API.

APIs allow analysts to automate data collection instead of manually downloading information every time.

### How API data becomes useful for analysis

A typical workflow looks like this:

```text
External API
     |
     | Send Request
     v
Retrieve Data
     |
     | Receive JSON Response
     v
Convert JSON into Python Objects
     |
     | Inspect and Clean
     v
Transform Data
     |
     | Store or Analyse
     v
Pandas DataFrame
     |
     v
Exploratory Data Analysis
```

For example, after retrieving data from an API, we can use Pandas to convert the records into a DataFrame.

```python
import requests

import pandas as pd

url = "https://restcountries.com/v3.1/all"

response = requests.get(url, timeout=10)

response.raise_for_status()

data = response.json()

df = pd.json_normalize(data)

print(df.head())
```

This demonstrates how API data can be brought into a data analysis workflow.

The exact columns available depend on the API response. Students should inspect the DataFrame before proceeding with analysis.
