Choosing the number of clusters

### What is WCSS?

WCSS = Within-Cluster Sum of Squares

It measures how far observations are from the centroid of the cluster they belong to.

In simpler terms:

> WCSS tells us how tightly packed our clusters are.

A smaller WCSS means observations are, on average, closer to their cluster centroids.

Imagine K = 1

If we put all customers into one cluster, we have one centroid.

Customers could be quite far from that centroid.

Therefore:

WCSS will be relatively high.

Now increase K

If we use:

K = 2

we have two centroids.

Customers can now be assigned to a centroid that is closer to them.

Therefore, WCSS decreases.

If we increase K again:

K = 3

we have three centroids, so WCSS generally decreases again.

This continues as K increases.

````python
# finding wcss value for different number of clusters

wcss = []

for i in range(1,11):
  kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
  kmeans.fit(X)

  wcss.append(kmeans.inertia_)
````

init='k-means++' - controls how the initial centroids are selected.

Instead of simply choosing the starting centroids completely randomly, K-Means++ chooses initial centroids in a way that generally spreads them out more intelligently.

````python
# plot an elbow graph

sns.set()
plt.plot(range(1,11), wcss)
plt.title('The Elbow Point Graph')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
````

Optimum Number of Clusters = 5

Training the k-Means Clustering Model

````python
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=0)

# return a label for each data point based on their cluster
Y = kmeans.fit_predict(X)

print(Y)
````

### Customer Segementation using Hierachical clustering 

| Linkage      | How it measures distance            | Tends to produce                    | When to consider it                                         |
| ------------ | ----------------------------------- | ----------------------------------- | ----------------------------------------------------------- |
| **Single**   | Closest pair of points              | Long, chain-like clusters           | When clusters may have irregular shapes                     |
| **Complete** | Farthest pair of points             | Compact, tight clusters             | When you want well-separated clusters                       |
| **Average**  | Average distance between all points | Balanced clusters                   | When you want a compromise between single and complete      |
| **Ward**     | Increase in within-cluster variance | Compact, roughly spherical clusters | When working with numerical data and similar-sized clusters |

#### Create the hierarchical clustering tree

For hierarchical clustering, we can use Ward's linkage:

````python
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
````

````python

linkages = ['single', 'complete', 'average', 'ward']

for linkage in linkages:
    model = AgglomerativeClustering(
        n_clusters=5,
        linkage=linkage
    )

    labels = model.fit_predict(X)

    score = silhouette_score(X, labels)

    print(f'{linkage}: {score:.3f}')
````

- `Single (0.269)`: The lowest score. It suggests that single linkage produced clusters with more overlap or chaining.
- `Complete (0.553)`: High silhouette score, indicating relatively well-separated and compact clusters.
- `Average (0.479)`: Reasonable separation, but lower than complete and Ward.
- `Ward (0.553)`: Tied with complete for the highest silhouette score.

`Ward` is particularly natural to consider because it is designed to create compact clusters by minimizing within-cluster variance

````python
Z = linkage(X_scaled, method='ward')
````

### DBSCAN 

 It works differently from hierarchical clustering: instead of deciding the number of clusters first, it identifies groups based on density and can mark unusual points as `noise`.

#### Apply DBSCAN

````python
from sklearn.cluster import DBSCAN
````

````python
dbscan = DBSCAN(
    eps=0.3,
    min_samples=5
)
clusters = dbscan.fit_predict(X_scaled)
````

- `eps` defines the neighborhood around each observation.
- `min_samples` defines how many observations need to be within that neighborhood for an area to be considered sufficiently dense.

````python
customer_data['Cluster'] = clusters
customer_data.head()
````

````python
customer_data['Cluster'].value_counts()
````

Unlike hierarchical clustering, you don't tell `DBSCAN` "give me 5 clusters."

`DBSCAN` determines the clusters based on density.

-1 means that these observations were classified as noise because they did not have enough neighboring observations within the specified eps distance.

#### Visualize the DBSCAN clusters

````python
plt.figure(figsize=(10, 6))

for cluster in sorted(customer_data['Cluster'].unique()):
    cluster_data = customer_data[customer_data['Cluster'] == cluster]
    
    label = 'Noise (-1)' if cluster == -1 else f'Cluster {cluster}'
    
    plt.scatter(
        cluster_data['Annual Income (k$)'],
        cluster_data['Spending Score (1-100)'],
        label=label
    )

plt.title('Customer Segments using DBSCAN')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend(title='Cluster')
plt.show()
````

### Gaussian Mixture Models (GMM)

`GMM` gives each customer a probability of belonging to each cluster, rather than assigning them only to one cluster.

#### Fit the GMM

````python
from sklearn.mixture import GaussianMixture
````

````python
gmm = GaussianMixture(
    n_components=5,
    random_state=42
)

gmm.fit(X_scaled)
````

## Choosing the Covariance Type

In a Gaussian Mixture Model (GMM), the **covariance type** determines the shape and flexibility of the clusters.

Different covariance types make different assumptions about how features vary within each cluster.

### 1. `full`

Each cluster has its own covariance matrix.

* Allows clusters to have different shapes, sizes, and orientations.
* Most flexible option.
* Can model elliptical clusters that are tilted in different directions.
* Uses more parameters.

```python
GaussianMixture(
    n_components=5,
    covariance_type='full',
    random_state=42
)
```

**Use when:** You expect clusters to have different shapes and orientations.

### 2. `tied`

All clusters share the same covariance matrix.

* Clusters can have different centers.
* All clusters have the same overall shape and orientation.
* Less flexible than `full`.

**Use when:** You expect clusters to have similar shapes and orientations.

### 3. `diag`

Each cluster has its own diagonal covariance matrix.

* Allows clusters to have different sizes.
* Assumes features are not correlated within each cluster.
* Clusters are aligned with the feature axes.
* More restrictive than `full`.

**Use when:** You want clusters to have different variances but don't want to model correlations between features.

### 4. `spherical`

Each cluster has a single variance value.

* Creates roughly circular or spherical clusters.
* Each cluster can have a different size.
* Simplest covariance structure.
* Least flexible option.


**Use when:** You expect clusters to be roughly circular or spherical.

## Comparing the Covariance Types

| Covariance Type | Cluster Shape                                | Flexibility |
| --------------- | -------------------------------------------- | ----------- |
| `full`          | Different elliptical shapes and orientations | Highest     |
| `tied`          | Same shape and orientation                   | High        |
| `diag`          | Axis-aligned ellipses                        | Medium      |
| `spherical`     | Circular or spherical                        | Lowest      |

### How do we choose?

We can fit GMMs using different covariance types and compare their **AIC** and **BIC** scores.

#### The interesting part: probabilities

This is where GMM differs from K-Means.

````python
probabilities = gmm.predict_proba(X_scaled)

probabilities[:5]
````

````python
probabilities_df = pd.DataFrame(
    probabilities,
    columns=[
        'Cluster 0',
        'Cluster 1',
        'Cluster 2',
        'Cluster 3',
        'Cluster 4'
    ]
)

# Convert probabilities to percentages
probabilities_df = probabilities_df * 100

# Select the cluster with the highest probability
probabilities_df['Selected Cluster'] = probabilities_df.idxmax(axis=1)

# Format probabilities as percentages
for column in [
    'Cluster 0',
    'Cluster 1',
    'Cluster 2',
    'Cluster 3',
    'Cluster 4'
]:
    probabilities_df[column] = probabilities_df[column].map(
        lambda x: f'{x:.4f}%'
    )

probabilities_df.head()
````

#### Choosing the number of GMM components

Instead of simply assuming 5 components, you can compare different values using BIC and AIC:

### 1. AIC: Akaike Information Criterion

AIC evaluates a model based on:

* How well the model fits the data
* How complex the model is

It penalizes models that have too many parameters.

**Lower AIC = generally better model.**

The idea is:

> Choose a model that explains the data well without being unnecessarily complicated.

### 2. BIC: Bayesian Information Criterion

BIC works similarly to AIC, but it applies a **stronger penalty for model complexity**.

**Lower BIC = generally better model.**

So BIC tends to prefer simpler models when adding more components doesn't provide enough improvement.


````python
from sklearn.mixture import GaussianMixture
````

````python
n_components = range(1, 11)

aic_scores = []
bic_scores = []

for n in n_components:
    model = GaussianMixture(
        n_components=n,
        random_state=42
    )

    model.fit(X_scaled)

    aic_scores.append(model.aic(X_scaled))
    bic_scores.append(model.bic(X_scaled))
````

````python
plt.figure(figsize=(8, 5))

plt.plot(n_components, aic_scores, marker='o', label='AIC')
plt.plot(n_components, bic_scores, marker='o', label='BIC')

plt.xlabel('Number of Components')
plt.ylabel('Score')
plt.title('AIC and BIC for GMM')
plt.legend()

plt.show()
````
