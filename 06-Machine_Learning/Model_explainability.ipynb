## What is Model Explainability?

**Model explainability** is the ability to understand and explain **why a machine learning model made a particular prediction or decision**.

In simple terms:

> **Model explainability answers the question: "Why did the model make this prediction?"**

A machine learning model may give us a prediction, but explainability helps us understand the factors that influenced that prediction.

For example:

- Why did the model classify a customer as likely to churn?
- Why did the model predict a house price of KSh 10 million?
- Why did the model classify a transaction as fraudulent?

---

## Why is Model Explainability Important?

## 1. Builds Trust

Users are more likely to trust a model when they understand how it makes predictions.

For example, a bank may want to know why a model rejected a loan application.

## 2. Helps Identify Errors

Explainability can reveal when a model is relying on irrelevant or unexpected features.

For example, if a house-price model relies heavily on a house ID, there may be a problem with the data.

## 3. Helps Detect Bias

Explainability can help identify whether a model is making decisions based on inappropriate or biased patterns.

## 4. Helps Improve the Model

Understanding which features influence predictions can help us:

- Remove irrelevant features
- Identify data leakage
- Improve feature engineering
- Detect unusual model behaviour
- Select better features

## 5. Supports Decision-Making

In areas such as healthcare, banking, insurance, and recruitment, understanding model decisions can be very important.

---

## Global vs Local Explainability

There are two important ways of explaining a model.

### Global Explainability

Global explainability helps us understand **how the model behaves overall**.

It answers questions such as:

> Which features are most important to the model?

For example, a house-price model might determine that:

1. House size is very important
2. Location is important
3. Number of bedrooms is moderately important
4. Number of bathrooms is less important

---

### Local Explainability

Local explainability focuses on **one specific prediction**.

It answers:

> Why did the model make this particular prediction?

For example:

> Why did the model predict that this particular customer will churn?

SHAP and LIME are particularly useful for local explanations.

---

## Main Explainability Techniques

For this course, we will focus on three commonly used techniques:

1. Feature Importance
2. SHAP
3. LIME

---

## 1. Feature Importance

### What is Feature Importance?

**Feature importance** tells us how much each feature contributes to the model's predictions.

It helps answer:

> Which features does the model consider important?

For example, a house-price model might produce:

| Feature | Importance |
|---|---:|
| House Size | 0.45 |
| Location | 0.30 |
| Bedrooms | 0.15 |
| Bathrooms | 0.07 |
| House Age | 0.03 |

From this, we can see that **House Size** is the most important feature.

---

### Feature Importance in Tree-Based Models

Tree-based models such as:

- Decision Trees
- Random Forests
- Gradient Boosting

can calculate feature importance based on how much each feature contributes to improving the decision-making process of the trees.

For example, in a Decision Tree, a feature that produces useful splits and reduces impurity significantly will generally have higher importance.

---

## Advantages of Feature Importance

- Easy to understand
- Easy to visualize
- Useful for identifying important features
- Useful for feature selection
- Provides a quick overview of the model

---

## Limitations of Feature Importance

Feature importance does not necessarily explain **why a specific prediction was made**.

It tells us what is important to the model overall.

It can also be affected by:

- Correlated features
- High-cardinality features
- The type of model being used

Therefore, feature importance is useful for **global model understanding**, but we may need SHAP or LIME for more detailed explanations.

---

# 2. SHAP

## What is SHAP?

**SHAP** stands for:

> **SHapley Additive exPlanations**

SHAP is one of the most popular model explainability techniques.

It explains how much each feature contributes to a model's prediction.

SHAP is based on the idea of **Shapley values**, which come from cooperative game theory.

---

## Main Idea Behind SHAP

SHAP asks:

> How much did each feature contribute to moving the prediction away from the average prediction?

For example, suppose the average house price predicted by a model is KSh 8 million.

For one particular house:

- Large house size increases the prediction
- Prime location increases the prediction
- More bedrooms increase the prediction
- Older age decreases the prediction

The contributions of these features combine to produce the final prediction.

---

## SHAP Provides Both Global and Local Explanations

One of the major advantages of SHAP is that it can be used to understand:

### Global Behaviour

Which features are generally most important across the dataset?

### Individual Predictions

Why did the model make this prediction for this particular observation?

This makes SHAP very powerful for model interpretation.

---

## SHAP Values

A SHAP value represents the contribution of a feature to a prediction.

### Positive SHAP Value

A positive SHAP value means the feature **pushes the prediction higher**.

### Negative SHAP Value

A negative SHAP value means the feature **pushes the prediction lower**.

The exact meaning of "higher" or "lower" depends on the model and the output being explained.

For example, in a classification problem:

- A positive SHAP value may push the prediction toward Class 1
- A negative SHAP value may push the prediction toward Class 0

---

## Advantages of SHAP

- Can provide global explanations
- Can provide local explanations
- Shows the direction of a feature's contribution
- Works with many different ML models
- Provides detailed explanations
- Helps explain individual predictions

---

## Limitations of SHAP

- Can be computationally expensive
- Can be difficult for beginners to interpret
- Results depend on the model and data
- A SHAP contribution does not necessarily mean that the feature caused the outcome

---

# 3. LIME

## What is LIME?

**LIME** stands for:

> **Local Interpretable Model-agnostic Explanations**

LIME is designed primarily to explain **individual predictions**.

It answers:

> Why did the model make this particular prediction?

---

## How Does LIME Work?

LIME works by creating a simpler, interpretable model around a specific prediction.

The process is approximately:

1. Select one observation
2. Create variations of that observation
3. Get predictions from the original complex model
4. Identify which features influence the prediction
5. Fit a simpler model around that observation
6. Use the simpler model to explain the original prediction

The key idea is:

> **A complex model may be difficult to understand globally, but its behaviour around one particular prediction can be approximated using a simpler model.**

---

## Why is LIME Called Model-Agnostic?

**Model-agnostic** means that LIME does not depend on the internal structure of a particular machine learning algorithm.

It can be used with models such as:

- Logistic Regression
- Decision Trees
- Random Forest
- SVM
- Neural Networks

LIME only needs to interact with the model by providing inputs and receiving predictions.

---

## Advantages of LIME

- Model-agnostic
- Good for explaining individual predictions
- Can be used with many different ML models
- Provides relatively simple explanations

---

## Limitations of LIME

- Mainly provides local explanations
- Results can vary between runs
- The explanation is an approximation of the original model
- Results can depend on how the local neighbourhood is created
- It may not always provide a faithful representation of the model

---

# SHAP vs LIME

SHAP and LIME are both commonly used for explaining complex machine learning models, but they work differently.

| Feature | SHAP | LIME |
|---|---|---|
| Full Name | SHapley Additive exPlanations | Local Interpretable Model-agnostic Explanations |
| Global Explanation | Yes | No |
| Local Explanation | Yes | Yes |
| Model-Agnostic | Generally yes | Yes |
| Shows Feature Contribution | Yes | Yes |
| Shows Direction of Contribution | Yes | Yes |
| Main Strength | Detailed feature contributions | Simple local explanations |
| Main Limitation | Can be computationally expensive | Explanations can be unstable |

---

# Feature Importance vs SHAP vs LIME

| Technique | Main Question |
|---|---|
| Feature Importance | Which features are important to the model overall? |
| SHAP | How much did each feature contribute to a prediction? |
| LIME | Why did the model make this particular prediction? |

A useful way to remember them:

> **Feature Importance → What matters?**

> **SHAP → How much did each feature contribute?**

> **LIME → Why this particular prediction?**

---

# Example: Customer Churn Model

Suppose we build a machine learning model that predicts whether a customer will churn.

The model predicts that a particular customer will churn.

## Feature Importance

Feature importance can tell us that:

- Contract type is important
- Monthly charges are important
- Customer tenure is important
- Support calls are moderately important

This gives us a **global understanding of the model**.

---

## SHAP

SHAP can explain the individual prediction by showing how each feature contributed.

For example:

- Short contract → pushes prediction toward churn
- High monthly charges → pushes prediction toward churn
- Low tenure → pushes prediction toward churn
- Long-term contract → pushes prediction away from churn

This tells us **how each feature contributed to this specific prediction**.

---

## LIME

LIME can create a local explanation for this particular customer.

It identifies the features that are most influential around this prediction and uses them to explain why the model classified the customer as likely to churn.

---

# Important: Explainability Does Not Mean Causation

A very important concept is:

> **Feature contribution does not necessarily mean causation.**

If SHAP shows that a feature strongly contributed to a prediction, this does not mean that the feature **caused** the outcome.

The model has learned a relationship from the training data.

For example:

> A high SHAP contribution from income does not prove that income causes loan default.

It only tells us that income was influential in the model's prediction.

---

# When Should We Use Each Technique?

## Use Feature Importance When:

- You want a quick overview of the model
- You want to identify the most important features
- You are interested in global model behaviour
- You are performing feature selection

## Use SHAP When:

- You want global and local explanations
- You want detailed feature contributions
- You want to know whether features push predictions higher or lower
- You want to investigate individual predictions

## Use LIME When:

- You want to explain individual predictions
- You are working with a complex model
- You want a model-agnostic explanation
- You want a relatively simple local explanation
